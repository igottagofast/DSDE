# FUNCTION

## check duplicate

````python
import pandas as pd
from IPython.display import display

def check_duplicates(df):
    """
    ตรวจสอบหาแถวที่ซ้ำซ้อน (Duplicate Rows) ใน DataFrame
    """
    duplicate_rows = df[df.duplicated()]
    total_duplicates = len(duplicate_rows)

    if total_duplicates > 0:
        print(f"⚠ พบแถวซ้ำทั้งหมด: {total_duplicates} แถว")
        print("ตัวอย่างข้อมูลที่ซ้ำซ้อน:")
        display(duplicate_rows.head(10))
    else:
        print("✓ ไม่พบข้อมูลซ้ำซ้อน (No Duplicates)")

    return duplicate_rows
````

## check null

````python
import pandas as pd
from IPython.display import display

def check_and_show_nulls(df):
    """
    ตรวจสอบหาค่าว่าง (Null/NaN) ใน DataFrame และพิมพ์สรุปจำนวนแถวพร้อมแสดงข้อมูล
    """
    null_counts = df.isnull().sum()
    cols_with_null = null_counts[null_counts > 0]

    if cols_with_null.empty:
        print("✓ ไม่พบค่าว่างในข้อมูลนี้")
        return None

    null_rows = df[df.isnull().any(axis=1)]

    print(f"⚠ พบค่าว่างในคอลัมน์:")
    for col, count in cols_with_null.items():
        print(f"  - '{col}': {count} แถว")

    print(f"\nรวมแถวที่มีค่าว่างทั้งหมด: {len(null_rows)} แถว")
    display(null_rows.head(10))

    return null_rows
````

## delete null

````python
import pandas as pd

def drop_null_rows(df):
    """
    ลบแถวที่มีค่าว่าง (Null/NaN) ใน DataFrame และพิมพ์สรุปจำนวนที่ถูกลบ
    """
    initial_count = len(df)
    df_clean = df.dropna()
    deleted_count = initial_count - len(df_clean)

    if deleted_count > 0:
        print(f"✓ ลบค่าว่างสำเร็จ: ทั้งหมด {deleted_count} แถว")
    else:
        print("✓ ไม่พบค่าว่างในข้อมูลนี้")

    return df_clean
````

## drop column

````python
import pandas as pd

def drop_columns(df, columns_to_drop):
    """
    ลบคอลัมน์ที่ระบุออกจาก DataFrame อย่างปลอดภัย
    """
    # เลือกเฉพาะคอลัมน์ที่มีอยู่ใน DataFrame จริงๆ เพื่อเลี่ยง Error
    existing_cols = [col for col in columns_to_drop if col in df.columns]

    if existing_cols:
        df_clean = df.drop(columns=existing_cols)
        print(f"✓ ลบคอลัมน์สำเร็จ: {existing_cols}")
    else:
        print("✓ ไม่พบคอลัมน์ที่ระบุใน DataFrame (ไม่มีการลบ)")
        df_clean = df.copy()

    return df_clean
````

## Bfill and Ffill

````python
import pandas as pd

def fill_null_values(df):
    """
    จัดการเติมค่าว่าง (Null/NaN) ด้วยวิธี Forward Fill (ffill) ตามด้วย Backward Fill (bfill)
    """
    initial_nulls = df.isnull().sum().sum()

    if initial_nulls == 0:
        print("✓ ไม่พบค่าว่างในข้อมูลนี้")
        return df.copy()

    df_filled = df.ffill().bfill()
    remaining_nulls = df_filled.isnull().sum().sum()

    print(f"✓ จัดการค่าว่างสำเร็จ (เริ่มต้น {initial_nulls} ค่า -> คงเหลือ {remaining_nulls} ค่า)")
    return df_filled

# ค้นหา DataFrame ทั้งหมดที่มีค่าว่างและนำมาจัดการเติมข้อมูล
all_dfs = {name: val for name, val in globals().items() if isinstance(val, pd.DataFrame) and not name.startswith('_')}

print("=== กำลังเริ่มทำการจัดการค่าว่างด้วยวิธี Forward / Backward Fill ===\n")
for name, df in sorted(all_dfs.items()):
    if name == 'df':
        continue

    null_count = df.isnull().sum().sum()
    if null_count > 0:
        print(f"[{name}] พบค่าว่าง {null_count} ตำแหน่ง")
        globals()[name] = fill_null_values(df)
        print()
````

## scatter graph

````python
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns
import pandas as pd
import numpy as np
from IPython.display import display

# ตั้งค่าสไตล์และการแสดงผลเบื้องต้น พร้อมกำหนดฟอนต์สำหรับกรณีทั่วไป
sns.set_theme(style="whitegrid")
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False

def analyze_correlation_generic(df_macro, df_assets,
                               date_col='Date',
                               macro_col='Close',
                               asset_col='Close',
                               group_col='Stock',
                               macro_name='USD_THB',
                               asset_type='หุ้นส่งออก',
                               color='#2ca02c'):
    """
    ฟังก์ชันวิเคราะห์และพล็อต Regression ระหว่างปัจจัยมหภาคกับกลุ่มสินทรัพย์ โดยระบุฟอนต์ภาษาไทยตรงไปยังออบเจกต์ของกราฟโดยตรง
    และเพิ่มพารามิเตอร์ color เพื่อกำหนดสีของจุดใน Scatter plot (Default: สีเขียว '#2ca02c')
    """
    # ดึงออบเจกต์ฟอนต์ Sarabun โดยตรงเพื่อความชัวร์
    thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')

    # 1. เตรียมข้อมูลปัจจัยมหภาค
    macro_renamed = f"{macro_name}_{macro_col}"
    df_macro_clean = df_macro[[date_col, macro_col]].rename(columns={macro_col: macro_renamed})
    df_macro_clean[date_col] = pd.to_datetime(df_macro_clean[date_col])

    # รายชื่อกลุ่มสินทรัพย์ทั้งหมดที่ต้องการวิเคราะห์
    unique_groups = df_assets[group_col].unique()
    num_groups = len(unique_groups)

    # คำนวณจำนวนแถวสำหรับ Subplots (แสดง 2 คอลัมน์)
    num_cols = 2
    num_rows = int(np.ceil(num_groups / num_cols))

    # เคลียร์ข้อมูลสำหรับการบันทึกค่า Correlation
    saved_correlations = []

    # 2. สร้าง Grid สำหรับแสดงผลกราฟ
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(16, 4.8 * num_rows))
    if num_groups == 1:
        axes = np.array([axes])
    else:
        axes = axes.flatten()  # แปลงเป็น 1D array

    for idx, group_name in enumerate(unique_groups):
        if idx >= len(axes):
            break

        # กรองข้อมูลเฉพาะกลุ่มย่อยรายตัว
        asset_renamed = f"Asset_{asset_col}"
        df_sub = df_assets[df_assets[group_col] == group_name][[date_col, asset_col]].rename(columns={asset_col: asset_renamed})
        df_sub[date_col] = pd.to_datetime(df_sub[date_col])

        # รวมข้อมูลปัจจัยมหภาคกับข้อมูลสินทรัพย์รายย่อย
        df_merged = pd.merge(df_macro_clean, df_sub, on=date_col, how='inner').sort_values(date_col)

        if df_merged.empty or len(df_merged) < 2:
            continue

        # คำนวณค่าสหสัมพันธ์ (Correlation)
        correlation = df_merged[macro_renamed].corr(df_merged[asset_renamed])

        # บันทึกค่าลงในรายการ
        saved_correlations.append({
            'Group': group_name,
            'ประเภทสินทรัพย์': asset_type,
            'Macro Factor': macro_name,
            'Correlation': correlation
        })

        # วาด Regression Plot โดยใช้ตัวแปรสีที่ส่งเข้ามา
        sns.regplot(
            data=df_merged,
            x=macro_renamed,
            y=asset_renamed,
            ax=axes[idx],
            scatter_kws={'alpha': 0.25, 'color': color, 's': 10},
            line_kws={'color': '#d62728', 'linewidth': 1.5}
        )

        # ตั้งชื่อและป้ายกำกับด้วยฟอนต์ภาษาไทยอย่างเฉพาะเจาะจง
        axes[idx].set_title(f'{group_name} (Corr = {correlation:.3f})', fontproperties=thai_font, fontsize=12, fontweight='bold')
        axes[idx].set_xlabel(f'{macro_name} ({macro_col})', fontproperties=thai_font, fontsize=11)
        axes[idx].set_ylabel(f'{group_name} ({asset_col})', fontproperties=thai_font, fontsize=11)

    # ซ่อนแกนที่เหลือ (กรณีจำนวนกลุ่มเป็นเลขคี่)
    for j in range(idx + 1, len(axes)):
        fig.delaxes(axes[j])

    # ตั้งค่าหัวข้อใหญ่ด้วยฟอนต์ภาษาไทย
    fig.suptitle(f'Correlation between {macro_name} and {asset_type}', fontproperties=thai_font, fontsize=16, fontweight='bold', y=1.01)
    plt.tight_layout()
    plt.show()

    # --- แสดงตารางสรุป ---
    df_summary = pd.DataFrame(saved_correlations).drop_duplicates(subset=['Group', 'Macro Factor']).reset_index(drop=True)
    df_sorted = df_summary.sort_values(by='Correlation').reset_index(drop=True)

    print(f"\n=== [ตารางสรุป] ค่า Correlation ระหว่าง {macro_name} และ {asset_type} (เรียงจากน้อยไปมาก) ===")
    display(df_sorted)

    return df_sorted
````

## scale corr

````python
def scale_correlation(corr_value, factor=10):
    """
    ฟังก์ชันสำหรับคูณสเกลค่า Correlation เพื่อขยายผลลัพธ์
    """
    if corr_value is None or pd.isna(corr_value):
        return corr_value
    return corr_value * factor
````

````python
import os
import pandas as pd

# ฟังก์ชันเสริมสำหรับโหลดไฟล์ CSV และตั้งชื่อตัวแปรตามชื่อไฟล์แบบอัตโนมัติ
def load_data_from_paths(file_list, prefix="df_"):
    loaded_dfs = {}
    for path in file_list:
        # ดึงชื่อไฟล์ออกมาโดยตัด .csv ออก
        filename = os.path.splitext(os.path.basename(path))[0]
        # ปรับแต่งชื่อไฟล์ให้เป็นภาษาอังกฤษหรือคำที่ระบบเข้าใจง่าย
        clean_name = filename.replace("ราคาย้อนหลัง หุ้น ", "") \
                             .replace("ราคาย้อนหลังของ สัญญาซื้อขายล่วงหน้าน้ำมันดิบ ", "") \
                             .replace("ราคาย้อนหลังของ สัญญาซื้อขายล่วงหน้าน้ำมัน", "") \
                             .replace("ข้อมูลผลตอบแทนย้อนหลังของพันธบัตร ", "") \
                             .replace("อัตราแลกเปลี่ยนย้อนหลัง ", "") \
                             .replace("FM_RT_013_formatted", "TH_RATE")

        # จัดการแปลงชื่อเฉพาะกลุ่มอื่นๆ
        if "สหรัฐอเมริกา 10 ปี" in clean_name or "สหรัฐ 10 ปี" in clean_name:
            clean_name = "US_10Y"
        elif "สหรัฐอเมริกา 30 ปี" in clean_name or "สหรัฐ 30 ปี" in clean_name:
            clean_name = "US_30Y"
        elif "ไทย 10 ปี" in clean_name:
            clean_name = "TH_10Y"
        elif "ไทย 20 ปี" in clean_name:
            clean_name = "TH_20Y"

        clean_name = clean_name.replace(" ", "_")
        var_name = f"{prefix}{clean_name}"

        try:
            df_temp = pd.read_csv(path)
            globals()[var_name] = df_temp
            loaded_dfs[var_name] = df_temp
            print(f"✓ โหลด {var_name} สำเร็จ ({df_temp.shape[0]} แถว x {df_temp.shape[1]} คอลัมน์)")
        except Exception as e:
            print(f"✗ ไม่สามารถโหลด {var_name} จาก {path} ได้: {e}")
    return loaded_dfs

print("=== เริ่มต้นการโหลดข้อมูลทั้งหมดเข้าสู่ระบบ ===\n")

# เรียกใช้ฟังก์ชันโหลดไฟล์จากทุกหมวดหมู่
print("--- [1/6] โหลดข้อมูลหุ้นนำเข้า ---")
load_data_from_paths(files_import)

print("\n--- [2/6] โหลดข้อมูลหุ้นส่งออก ---")
load_data_from_paths(files_export)

print("\n--- [3/6] โหลดข้อมูลอัตราแลกเปลี่ยน ---")
load_data_from_paths(files_fx)

print("\n--- [4/6] โหลดข้อมูลดอกเบี้ยสหรัฐฯ ---")
load_data_from_paths(files_interest)

print("\n--- [5/6] โหลดข้อมูลพันธบัตรไทย ---")
load_data_from_paths(files_th_bond)

print("\n--- [6/6] โหลดข้อมูลราคาน้ำมัน ---")
load_data_from_paths(files_oil)

print("\n--- โหลดดอกเบี้ยอ้างอิงแห่งประเทศไทย ---")
load_data_from_paths(files_th_interest)
````

````python
import pandas as pd

# กำหนดแมปปิ้งคำแปลสำหรับคอลัมน์ภาษาไทยเป็นภาษาอังกฤษตามที่กำหนด
column_mapping = {
    'วันเดือนปี': 'Date',
    'ล่าสุด': 'Close',
    'ราคาเปิด': 'Open',
    'สูงสุด': 'High',
    'ต่ำสุด': 'Low',
    'ปริมาณ': 'Volume',
    '% เปลี่ยน': 'Pct_Change',
    'ประเภทหุ้น': 'Asset_Type'  # เก็บคอลัมน์พิเศษเพิ่มเติมไว้ถ้ามี
}

# ดึงรายชื่อ DataFrame ทั้งหมดในระบบ
all_dfs = {name: val for name, val in globals().items() if isinstance(val, pd.DataFrame) and not name.startswith('_')}

print("=== เริ่มต้นทำการเปลี่ยนหัวคอลัมน์ให้เป็นรูปแบบมาตรฐาน ===\n")

for name, df in all_dfs.items():
    # ทำการแปลงชื่อคอลัมน์ตามแมปปิ้งที่มี
    df.rename(columns=column_mapping, inplace=True)

    # หากขาดคอลัมน์ 'Volume' (เช่นในกรณีของข้อมูลพันธบัตร) จะทำการเพิ่มคอลัมน์ 'Volume' และเติมค่า NaN หรือ '-' เพื่อให้โครงสร้างเหมือนกัน
    if 'Volume' not in df.columns:
        df['Volume'] = None

    # จัดเรียงลำดับคอลัมน์พื้นฐานให้อยู่ลำดับแรกเพื่อความเป็นระเบียบ
    base_cols = ['Date', 'Close', 'Open', 'High', 'Low', 'Volume', 'Pct_Change']
    existing_base = [col for col in base_cols if col in df.columns]
    extra_cols = [col for col in df.columns if col not in base_cols]

    # กำหนดลำดับคอลัมน์ใหม่
    df = df[existing_base + extra_cols]
    globals()[name] = df # อัปเดตกลับไปยังตัวแปร Global

    print(f"✓ {name} ได้รับการปรับปรุงหัวคอลัมน์แล้ว -> {list(df.columns)}")
````

````python
import pandas as pd

# ดึงรายชื่อ DataFrame ทั้งหมดในระบบ
all_dfs = {name: val for name, val in globals().items() if isinstance(val, pd.DataFrame) and not name.startswith('_')}

target_cols = ['Date', 'Close', 'Open', 'High', 'Low', 'Volume', 'Pct_Change']

print("=== เริ่มต้นการลบคอลัมน์ที่ไม่เกี่ยวข้องออก ===\n")

for name, df in all_dfs.items():
    # เลือกเฉพาะคอลัมน์ที่อยู่ในเป้าหมายและมีอยู่ใน DataFrame จริง
    existing_target_cols = [col for col in target_cols if col in df.columns]

    # อัปเดต DataFrame โดยเก็บไว้เฉพาะ 7 คอลัมน์หลัก
    df_filtered = df[existing_target_cols]
    globals()[name] = df_filtered

    print(f"✓ {name} คงเหลือเฉพาะคอลัมน์ -> {list(df_filtered.columns)}")
````

````python
# ค้นหาและเก็บทุก DataFrame ที่อยู่ในระบบมาวนลูปตรวจสอบค่าว่าง
all_dfs = {name: val for name, val in globals().items() if isinstance(val, pd.DataFrame) and not name.startswith('_')}

print(f"--- กำลังเริ่มตรวจสอบค่าว่างในทั้งหมด {len(all_dfs)} DataFrames ---\n")

for name, df in sorted(all_dfs.items()):
    # ข้ามตัวแปรพักข้อมูลชั่วคราวที่เป็น df เพื่อไม่ให้ซ้ำซ้อน
    if name == 'df':
        continue

    # ตรวจสอบเบื้องต้นว่ามีค่าว่างใน DataFrame นี้หรือไม่
    null_sum = df.isnull().sum().sum()
    if null_sum > 0:
        print("=" * 60)
        print(f"📌 [พบค่าว่างใน DataFrame: {name}] (รวมทั้งหมด {null_sum} ตำแหน่ง)")
        print("=" * 60)
        check_and_show_nulls(df)
        print("\n")
    else:
        print(f"✓ {name}: ไม่มีค่าว่าง")
````

### นำเข้า

````python
import pandas as pd
import numpy as np

# ปรับแต่งข้อมูลเปอร์เซ็นต์ (Pct_Change) ให้เป็น Float
def clean_pct_series(df):
    df_clean = df.copy()
    if 'Pct_Change' in df_clean.columns and df_clean['Pct_Change'].dtype == 'object':
        df_clean['Pct_Change_Numeric'] = df_clean['Pct_Change'].str.rstrip('%').astype('float') / 100.0
    else:
        df_clean['Pct_Change_Numeric'] = pd.to_numeric(df_clean['Pct_Change'], errors='coerce')
    return df_clean

# 1. จัดทำข้อมูล USD_THB ในรูปแบบเปอร์เซ็นต์
df_usd_thb_pct = clean_pct_series(df_USD_THB)
df_macro_input = df_usd_thb_pct[['Date', 'Pct_Change_Numeric']].rename(columns={'Pct_Change_Numeric': 'Pct_Change'})

# 2. รวมข้อมูลหุ้นนำเข้าแต่ละตัวจากรายชื่อ พร้อมระบุชื่อหุ้น (Stock)
import_stock_dfs = {
    'PTTGC': df_PTTGC, 'TOP': df_TOP, 'IRPC': df_IRPC, 'IVL': df_IVL, 'COM7': df_COM7,
    'BGRIM': df_BGRIM, 'GPSC': df_GPSC, 'EGCO': df_EGCO, 'RATCH': df_RATCH, 'BCP': df_BCP
}

merged_import_list = []
for name, df_stock in import_stock_dfs.items():
    if df_stock is not None and not df_stock.empty:
        df_temp = df_stock.copy()
        df_temp['Stock'] = name
        merged_import_list.append(df_temp)

df_import_all_with_stock = pd.concat(merged_import_list, ignore_index=True)

# 3. จัดทำข้อมูลหุ้นนำเข้าในรูปแบบเปอร์เซ็นต์
df_import_all_pct = clean_pct_series(df_import_all_with_stock)
df_assets_input = df_import_all_pct[['Date', 'Pct_Change_Numeric', 'Stock']].rename(columns={'Pct_Change_Numeric': 'Pct_Change'})

# 4. เรียกใช้งานฟังก์ชันสำเร็จรูป analyze_correlation_generic ที่สร้างไว้ทันที
df_summary_result = analyze_correlation_generic(
    df_macro=df_macro_input,
    df_assets=df_assets_input,
    date_col='Date',
    macro_col='Pct_Change',
    asset_col='Pct_Change',
    group_col='Stock',
    macro_name='USD_THB',
    asset_type='หุ้นนำเข้า'
)
````

## Heatmap

### นำเข้า

````python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

def plot_correlation_heatmap(stocks_dict, macro_dict, title="แผนภูมิความร้อนแสดงความสัมพันธ์ (Correlation Heatmap)"):
    """
    สร้าง Heatmap ความสัมพันธ์ระหว่างหุ้นและปัจจัยภายนอก

    Parameters:
    stocks_dict (dict): Dictionary ของหุ้น { 'Stock_Name': df_stock }
    macro_dict (dict): Dictionary ของปัจจัยภายนอก { 'Factor_Name': df_factor }
    title (str): หัวข้อของกราฟ
    """
    # เตรียมฟอนต์ภาษาไทย
    thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')

    # ฟังก์ชันช่วยแปลง % Change ให้เป็นตัวเลข
    def clean_pct(df):
        df_clean = df.copy()
        df_clean['Date'] = pd.to_datetime(df_clean['Date'])
        if 'Pct_Change' in df_clean.columns:
            if df_clean['Pct_Change'].dtype == 'object':
                df_clean['Pct_Val'] = df_clean['Pct_Change'].str.rstrip('%').astype('float') / 100.0
            else:
                df_clean['Pct_Val'] = pd.to_numeric(df_clean['Pct_Change'], errors='coerce')
        else:
            # หากไม่มี Pct_Change ให้คำนวณจาก Close
            df_clean = df_clean.sort_values('Date')
            df_clean['Pct_Val'] = df_clean['Close'].pct_change()
        return df_clean[['Date', 'Pct_Val']]

    # 1. รวบรวมข้อมูลปัจจัยภายนอก
    df_macro_merged = None
    for factor_name, df_factor in macro_dict.items():
        df_cleaned = clean_pct(df_factor).rename(columns={'Pct_Val': factor_name})
        if df_macro_merged is None:
            df_macro_merged = df_cleaned
        else:
            df_macro_merged = pd.merge(df_macro_merged, df_cleaned, on='Date', how='inner')

    # 2. รวบรวมข้อมูลหุ้น 10 ตัว
    df_stocks_merged = None
    for stock_name, df_stock in stocks_dict.items():
        df_cleaned = clean_pct(df_stock).rename(columns={'Pct_Val': stock_name})
        if df_stocks_merged is None:
            df_stocks_merged = df_cleaned
        else:
            df_stocks_merged = pd.merge(df_stocks_merged, df_cleaned, on='Date', how='inner')

    # 3. รวมข้อมูลทั้งหมดเข้าด้วยกันด้วย Date
    if df_macro_merged is not None and df_stocks_merged is not None:
        df_final = pd.merge(df_stocks_merged, df_macro_merged, on='Date', how='inner')

        # ลบคอลัมน์ Date ออกเพื่อคำนวณเฉพาะ Correlation
        df_corr_data = df_final.drop(columns=['Date'])

        # คำนวณ Correlation Matrix
        corr_matrix = df_corr_data.corr()

        # เลือกดึงเฉพาะความสัมพันธ์ระหว่าง หุ้น (Rows) กับ ปัจจัยภายนอก (Columns)
        stock_names = list(stocks_dict.keys())
        macro_names = list(macro_dict.keys())
        sub_corr_matrix = corr_matrix.loc[stock_names, macro_names]

        # 4. พล็อต Heatmap
        plt.figure(figsize=(10, 8))
        sns.heatmap(
            sub_corr_matrix,
            annot=True,
            cmap='RdBu_r',
            vmin=-1,
            vmax=1,
            center=0,
            fmt=".3f",
            linewidths=0.5,
            cbar_kws={'label': 'Correlation Coefficient'}
        )

        # ตั้งค่าภาษาไทยให้กับส่วนต่างๆ ของกราฟ
        plt.title(title, fontproperties=thai_font, fontsize=16, fontweight='bold', pad=20)
        plt.xticks(fontproperties=thai_font, fontsize=12, rotation=45, ha='right')
        plt.yticks(fontproperties=thai_font, fontsize=12, rotation=0)

        # ปรับการแสดงผล Label ของ Colorbar
        ax = plt.gca()
        cbar = ax.collections[0].colorbar
        cbar.ax.yaxis.label.set_fontproperties(thai_font)
        cbar.ax.yaxis.label.set_size(12)

        plt.tight_layout()
        plt.show()

        return sub_corr_matrix
    else:
        print("ข้อมูลไม่เพียงพอในการสร้าง Heatmap")
        return None

# --- ตัวอย่างการเรียกใช้งานฟังก์ชันกับกลุ่มหุ้นนำเข้า 10 ตัว และปัจจัยภายนอกหลัก ---
# สร้าง Dict สำหรับหุ้น 10 ตัว
import_stocks = {
    'PTTGC': df_PTTGC, 'TOP': df_TOP, 'IRPC': df_IRPC, 'IVL': df_IVL, 'COM7': df_COM7,
    'BGRIM': df_BGRIM, 'GPSC': df_GPSC, 'EGCO': df_EGCO, 'RATCH': df_RATCH, 'BCP': df_BCP
}

# สร้าง Dict สำหรับปัจจัยภายนอก
external_factors = {
    'USD_THB (อัตราแลกเปลี่ยน)': df_USD_THB,
    'US_10Y (บอนด์ยีลด์สหรัฐ 10ปี)': df_US_10Y,
    'WTI (ราคาน้ำมัน WTI)': df_WTI,
    'Brent (ราคาน้ำมัน Brent)': df_เบรนท์,
    'TH_10Y (บอนด์ยีลด์ไทย 10ปี)': df_TH_10Y
}

# เรียกใช้ฟังก์ชันพล็อตกราฟ
corr_results = plot_correlation_heatmap(
    stocks_dict=import_stocks,
    macro_dict=external_factors,
    title="ความสัมพันธ์ (Correlation) ระหว่างหุ้นนำเข้า 10 ตัวกับปัจจัยภายนอก"
)
````

````python
# สร้าง Dict สำหรับหุ้นส่งออก 10 ตัว
export_stocks = {
    'DELTA': df_DELTA, 'HANA': df_HANA, 'KCE': df_KCE, 'CPF': df_CPF, 'TU': df_TU,
    'AAI': df_AAI, 'STA': df_STA, 'BDMS': df_BDMS, 'BH': df_BH, 'ITC': df_ITC
}

# สร้าง Dict สำหรับปัจจัยภายนอก (ใช้กลุ่มเดียวกับของหุ้นนำเข้าเพื่อความเปรียบเทียบที่ชัดเจน)
external_factors = {
    'USD_THB (อัตราแลกเปลี่ยน)': df_USD_THB,
    'US_10Y (บอนด์ยีลด์สหรัฐ 10ปี)': df_US_10Y,
    'WTI (ราคาน้ำมัน WTI)': df_WTI,
    'Brent (ราคาน้ำมัน Brent)': df_เบรนท์,
    'TH_10Y (บอนด์ยีลด์ไทย 10ปี)': df_TH_10Y
}

# เรียกใช้ฟังก์ชันพล็อตกราฟ Heatmap สำหรับกลุ่มส่งออก
corr_results_export = plot_correlation_heatmap(
    stocks_dict=export_stocks,
    macro_dict=external_factors,
    title="ความสัมพันธ์ (Correlation) ระหว่างหุ้นส่งออก 10 ตัวกับปัจจัยภายนอก"
)
````

## Correlation Matrix (20 หุ้น x 8 ปัจจัย)

````python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from IPython.display import display

# ฟังก์ชันดึงและทำความสะอาด Pct_Change เพื่อคำนวณ Correlation แบบจับคู่ (Pairwise)
def get_numeric_pct(df):
    df_clean = df.copy()
    df_clean['Date'] = pd.to_datetime(df_clean['Date'])
    if 'Pct_Change' in df_clean.columns:
        if df_clean['Pct_Change'].dtype == 'object':
            df_clean['Pct_Val'] = df_clean['Pct_Change'].str.rstrip('%').astype('float') / 100.0
        else:
            df_clean['Pct_Val'] = pd.to_numeric(df_clean['Pct_Change'], errors='coerce')
    else:
        df_clean = df_clean.sort_values('Date')
        df_clean['Pct_Val'] = df_clean['Close'].pct_change()
    return df_clean[['Date', 'Pct_Val']].dropna()

# รายชื่อกลุ่มหุ้น
import_stocks_dict = {
    'PTTGC': df_PTTGC, 'TOP': df_TOP, 'IRPC': df_IRPC, 'IVL': df_IVL, 'COM7': df_COM7,
    'BGRIM': df_BGRIM, 'GPSC': df_GPSC, 'EGCO': df_EGCO, 'RATCH': df_RATCH, 'BCP': df_BCP
}

export_stocks_dict = {
    'DELTA': df_DELTA, 'HANA': df_HANA, 'KCE': df_KCE, 'CPF': df_CPF, 'TU': df_TU,
    'AAI': df_AAI, 'STA': df_STA, 'BDMS': df_BDMS, 'BH': df_BH, 'ITC': df_ITC
}

# รายชื่อปัจจัยภายนอกและตัวแปรที่ใช้ใน Scatter Plot ก่อนหน้านี้
macro_factors_dict = {
    'USD_THB': df_USD_THB,
    'US_10Y': df_US_10Y,
    'US_30Y': df_US_30Y,
    'TH_10Y': df_TH_10Y,
    'TH_20Y': df_TH_20Y,
    'TH_RATE': df_TH_RATE,
    'WTI': df_WTI,
    'Brent': df_เบรนท์
}

# เก็บข้อมูลผลลัพธ์
scatter_corr_results = []

# 1. คำนวณฝั่งหุ้นนำเข้า
for stock_name, df_stock in import_stocks_dict.items():
    df_s_clean = get_numeric_pct(df_stock)
    for factor_name, df_factor in macro_factors_dict.items():
        df_f_clean = get_numeric_pct(df_factor)
        # Join กันเฉพาะคู่นั้น ๆ เพื่อเลียนแบบขั้นตอนพล็อต Scatter ของคุณ
        merged = pd.merge(df_s_clean, df_f_clean, on='Date', how='inner')
        if len(merged) >= 2:
            corr_val = merged['Pct_Val_x'].corr(merged['Pct_Val_y'])
        else:
            corr_val = np.nan
        scatter_corr_results.append({
            'Stock': stock_name,
            'Type': 'หุ้นนำเข้า',
            'Macro_Factor': factor_name,
            'Scatter_Correlation': corr_val
        })

# 2. คำนวณฝั่งหุ้นส่งออก
for stock_name, df_stock in export_stocks_dict.items():
    df_s_clean = get_numeric_pct(df_stock)
    for factor_name, df_factor in macro_factors_dict.items():
        df_f_clean = get_numeric_pct(df_factor)
        # Join กันเฉพาะคู่นั้น ๆ
        merged = pd.merge(df_s_clean, df_f_clean, on='Date', how='inner')
        if len(merged) >= 2:
            corr_val = merged['Pct_Val_x'].corr(merged['Pct_Val_y'])
        else:
            corr_val = np.nan
        scatter_corr_results.append({
            'Stock': stock_name,
            'Type': 'หุ้นส่งออก',
            'Macro_Factor': factor_name,
            'Scatter_Correlation': corr_val
        })

# แปลงเป็น DataFrame
df_scatter_corr = pd.DataFrame(scatter_corr_results)

# Pivot ตารางให้อยู่ในรูปแบบ Matrix เพื่อเปรียบเทียบง่าย ๆ
matrix_scatter_corr = df_scatter_corr.pivot(index='Stock', columns='Macro_Factor', values='Scatter_Correlation')
# จัดเรียงลำดับแถวตามกลุ่มนำเข้าและส่งออก
ordered_stocks = list(import_stocks_dict.keys()) + list(export_stocks_dict.keys())
matrix_scatter_corr = matrix_scatter_corr.reindex(ordered_stocks)

print("=== ตารางค่า Correlation ที่มาจากการจับคู่พล็อต Scatter Plot แต่ละคู่โดยตรง ===")
display(matrix_scatter_corr)

# 3. สร้างกราฟ Heatmap
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
sns.set_theme(style="white")

plt.figure(figsize=(14, 11))
sns.heatmap(
    matrix_scatter_corr,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    center=0,
    fmt=".3f",
    linewidths=0.5,
    cbar_kws={'label': 'Correlation Coefficient'}
)

# ตั้งค่าหัวข้อแกนและฟอนต์
plt.title("Heatmap: สหสัมพันธ์ (Pairwise Correlation) ตามแนวทาง Scatter Plot\n(หุ้นนำเข้า 10 ตัวแรก / หุ้นส่งออก 10 ตัวหลัง x 8 ปัจจัยมหภาค)", fontproperties=thai_font, fontsize=15, fontweight='bold', pad=20)
plt.xticks(fontproperties=thai_font, fontsize=11, rotation=45, ha='right')
plt.yticks(fontproperties=thai_font, fontsize=11, rotation=0)

# ปรับเปลี่ยนฟอนต์ของ Colorbar
ax = plt.gca()
cbar = ax.collections[0].colorbar
cbar.ax.yaxis.label.set_fontproperties(thai_font)
cbar.ax.yaxis.label.set_size(12)

plt.tight_layout()
plt.show()
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm

# ดึงข้อมูลจาก df_scatter_corr ที่คำนวณไว้ในขั้นตอนก่อนหน้า
# กรองเฉพาะปัจจัยราคาน้ำมัน WTI และ Brent
df_oil_corr = df_scatter_corr[df_scatter_corr['Macro_Factor'].isin(['WTI', 'Brent'])].copy()

# 1. คำนวณค่าสถิติเชิงพรรณนาจำแนกตามประเภทหุ้นและราคาน้ำมัน
oil_summary = df_oil_corr.groupby(['Type', 'Macro_Factor'])['Scatter_Correlation'].agg(['mean', 'std', 'min', 'max']).reset_index()

print("=== ตารางสรุปค่าเฉลี่ยและช่วงค่า Correlation จำแนกตามกลุ่ม ===")
display(oil_summary)

# 2. ทำการพล็อตกราฟเปรียบเทียบความแตกต่างจำแนกรายหุ้น
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6), sharey=True)

# กรองข้อมูลรายตัวเพื่อพล็อต
import_oil = df_oil_corr[df_oil_corr['Type'] == 'หุ้นนำเข้า']
export_oil = df_oil_corr[df_oil_corr['Type'] == 'หุ้นส่งออก']

# พล็อตฝั่งหุ้นนำเข้า
sns.barplot(data=import_oil, x='Stock', y='Scatter_Correlation', hue='Macro_Factor', ax=ax1, palette='Greens_d')
ax1.set_title('ค่าความสัมพันธ์ของ "หุ้นกลุ่มนำเข้า" กับราคาน้ำมัน', fontproperties=thai_font, fontsize=12, fontweight='bold')
ax1.set_xlabel('รายชื่อหุ้นนำเข้า', fontproperties=thai_font, fontsize=10)
ax1.set_ylabel('Correlation Coefficient', fontproperties=thai_font, fontsize=10)
ax1.set_xticklabels(ax1.get_xticklabels(), rotation=45)
ax1.axhline(0, color='grey', linestyle='--', linewidth=0.8)

# พล็อตฝั่งหุ้นส่งออก
sns.barplot(data=export_oil, x='Stock', y='Scatter_Correlation', hue='Macro_Factor', ax=ax2, palette='Purples_d')
ax2.set_title('ค่าความสัมพันธ์ของ "หุ้นกลุ่มส่งออก" กับราคาน้ำมัน', fontproperties=thai_font, fontsize=12, fontweight='bold')
ax2.set_xlabel('รายชื่อหุ้นส่งออก', fontproperties=thai_font, fontsize=10)
ax2.set_ylabel('', fontproperties=thai_font, fontsize=10)
ax2.set_xticklabels(ax2.get_xticklabels(), rotation=45)
ax2.axhline(0, color='grey', linestyle='--', linewidth=0.8)

fig.suptitle('การเปรียบเทียบค่าความสัมพันธ์เชิงเส้นจำแนกรายตัวระหว่างกลุ่มนำเข้าและส่งออกต่อราคาน้ำมัน', fontproperties=thai_font, fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
````

## feature engineering

### สูตร feature engineering

````python
import pandas as pd
import numpy as np

# ฟังก์ชันเสริมสำหรับแปลงคอลัมน์ข้อมูลต่าง ๆ ให้เป็นตัวเลขอย่างปลอดภัย
def safe_to_numeric(series):
    if series.dtype == 'object':
        return pd.to_numeric(series.astype(str).str.replace(',', '').str.rstrip('%'), errors='coerce')
    return pd.to_numeric(series, errors='coerce')

# =====================================================================
# 1. กลุ่มฟังก์ชันคำนวณภายในหลักทรัพย์ (Intraday & Lagged Features)
# =====================================================================

def calculate_intraday_features(df):
    """
    คำนวณหาความผันผวนระหว่างวัน (Volatility) และทิศทางความเคลื่อนไหวราคาภายในวัน (Movement)
    สูตร:
      - Daily_Volatility = High - Low
      - Daily_Movement = Close - Open
    """
    df_out = df.copy()
    df_out['Date'] = pd.to_datetime(df_out['Date'])

    # แปลงราคารองรับกรณีเป็น String
    high = safe_to_numeric(df_out['High'])
    low = safe_to_numeric(df_out['Low'])
    close = safe_to_numeric(df_out['Close'])
    open_val = safe_to_numeric(df_out['Open'])

    df_out['Daily_Volatility'] = high - low
    df_out['Daily_Movement'] = close - open_val
    return df_out

def calculate_price_change_prev_day(df, stock_name=None, close_col='Close'):
    """
    คำนวณส่วนต่างราคาปิดของวันนี้เทียบกับราคาปิดของวันทำการก่อนหน้า
    สูตร: Price_Change_Prev_Day = Close_today - Close_yesterday
    """
    df_out = df.copy()
    df_out['Date'] = pd.to_datetime(df_out['Date'])
    df_out = df_out.sort_values(by='Date')

    # ดึงค่า Close เป็นตัวเลข
    close_series = safe_to_numeric(df_out[close_col])

    # คำนวณหาผลต่างย้อนหลัง 1 วัน
    df_out['Price_Change_Prev_Day'] = close_series.diff()

    if stock_name:
        df_out['Stock'] = stock_name

    return df_out.reset_index(drop=True)

def calculate_volatility_to_open(df):
    """
    คำนวณอัตราส่วนความผันผวนระหว่างวันเทียบราคาเปิด
    สูตร: Volatility_to_Open = (High - Low) / Open
    """
    df_out = df.copy()
    high = safe_to_numeric(df_out['High'])
    low = safe_to_numeric(df_out['Low'])
    open_val = safe_to_numeric(df_out['Open'])

    df_out['Volatility_to_Open'] = (high - low) / open_val
    return df_out

# =====================================================================
# 2. กลุ่มฟังก์ชันเปรียบเทียบข้ามหลักทรัพย์ และ ปัจจัยภายนอก (Cross-File Features)
# =====================================================================

def calculate_relative_return(df_stock, df_factor, factor_name='FX', stock_name=None):
    """
    คำนวณผลตอบแทนส่วนต่างเทียบกับปัจจัยภายนอก (เช่น อัตราแลกเปลี่ยน หรือดอกเบี้ย)
    สูตร: Relative_Return = Pct_Change_Stock - Pct_Change_Factor
    """
    df_stk_cp = df_stock.copy()
    df_fac_cp = df_factor.copy()

    df_stk_cp['Date'] = pd.to_datetime(df_stk_cp['Date'])
    df_fac_cp['Date'] = pd.to_datetime(df_fac_cp['Date'])

    # แปลงเปอร์เซ็นต์การเปลี่ยนแปลงเป็นตัวเลข
    df_stk_cp['Stock_Pct_Change'] = safe_to_numeric(df_stk_cp['Pct_Change'])
    df_fac_cp['Factor_Pct_Change'] = safe_to_numeric(df_fac_cp['Pct_Change'])

    # รวมข้อมูล
    merged = pd.merge(
        df_stk_cp[['Date', 'Stock_Pct_Change', 'Close', 'Open', 'High', 'Low']],
        df_fac_cp[['Date', 'Factor_Pct_Change']].rename(columns={'Factor_Pct_Change': f'{factor_name}_Pct_Change'}),
        on='Date',
        how='inner'
    )

    merged[f'Relative_Return_vs_{factor_name}'] = merged['Stock_Pct_Change'] - merged[f'{factor_name}_Pct_Change']

    if stock_name:
        merged['Stock'] = stock_name

    return merged.sort_values(by='Date').reset_index(drop=True)

def calculate_yield_slope(df_long, df_short, long_name='Yield_20Y', short_name='Yield_10Y'):
    """
    คำนวณความชันระหว่างอัตราผลตอบแทนพันธบัตรรัฐบาลระยะยาวและระยะสั้น
    สูตร: Yield_Curve_Slope = Yield_Long - Yield_Short
    """
    df_l = df_long.copy()
    df_s = df_short.copy()

    df_l['Date'] = pd.to_datetime(df_l['Date'])
    df_s['Date'] = pd.to_datetime(df_s['Date'])

    # แปลง Yield เป็นตัวเลข
    df_l['Yield_Long'] = safe_to_numeric(df_l['Close'])
    df_s['Yield_Short'] = safe_to_numeric(df_s['Close'])

    merged = pd.merge(
        df_l[['Date', 'Yield_Long']].rename(columns={'Yield_Long': long_name}),
        df_s[['Date', 'Yield_Short']].rename(columns={'Yield_Short': short_name}),
        on='Date',
        how='inner'
    )

    merged['Yield_Curve_Slope'] = merged[long_name] - merged[short_name]
    return merged.sort_values(by='Date').reset_index(drop=True)

def calculate_relative_strength(df_export, df_import, exp_stock_name="Export_Stock", imp_stock_name="Import_Stock"):
    """
    คำนวณอัตราส่วนความแข็งแกร่งเปรียบเทียบเชิงสัมพัทธ์ และส่วนต่างผลตอบแทนของคู่หุ้น
    สูตร:
      - Relative_Strength_Ratio = Export_Close / Import_Close
      - Return_Diff_Export_vs_Import = Export_Pct_Change - Import_Pct_Change
    """
    df_exp_sub = df_export.copy()
    df_imp_sub = df_import.copy()

    df_exp_sub['Date'] = pd.to_datetime(df_exp_sub['Date'])
    df_imp_sub['Date'] = pd.to_datetime(df_imp_sub['Date'])

    # แปลง Close และ Pct_Change เป็นตัวเลข
    df_exp_sub['Export_Close'] = safe_to_numeric(df_exp_sub['Close'])
    df_exp_sub['Export_Pct_Change'] = safe_to_numeric(df_exp_sub['Pct_Change'])

    df_imp_sub['Import_Close'] = safe_to_numeric(df_imp_sub['Close'])
    df_imp_sub['Import_Pct_Change'] = safe_to_numeric(df_imp_sub['Pct_Change'])

    merged = pd.merge(
        df_exp_sub[['Date', 'Export_Close', 'Export_Pct_Change']],
        df_imp_sub[['Date', 'Import_Close', 'Import_Pct_Change']],
        on='Date',
        how='inner'
    )

    merged['Stock_Pair'] = f"{exp_stock_name} vs {imp_stock_name}"
    merged['Relative_Strength_Ratio'] = merged['Export_Close'] / merged['Import_Close']
    merged['Return_Diff_Export_vs_Import'] = merged['Export_Pct_Change'] - merged['Import_Pct_Change']

    return merged.sort_values(by='Date').reset_index(drop=True)

def convert_price_to_usd(df_stock, df_fx, stock_name=None):
    """
    แปลงราคาหุ้นปิดสกุลเงินบาท (THB) ให้เป็นสกุลเงินดอลลาร์สหรัฐ (USD) อ้างอิงอัตราแลกเปลี่ยนรายวัน
    สูตร: Stock_Close_USD = Stock_Close_THB / USD_THB_Close
    """
    df_stk_cp = df_stock.copy()
    df_fx_cp = df_fx.copy()

    df_stk_cp['Date'] = pd.to_datetime(df_stk_cp['Date'])
    df_fx_cp['Date'] = pd.to_datetime(df_fx_cp['Date'])

    # แปลงราคาปิดเป็นตัวเลข
    df_stk_cp['Stock_Close_THB'] = safe_to_numeric(df_stk_cp['Close'])
    df_fx_cp['USD_THB_Close'] = safe_to_numeric(df_fx_cp['Close'])

    merged = pd.merge(
        df_stk_cp,
        df_fx_cp[['Date', 'USD_THB_Close']],
        on='Date',
        how='inner'
    )

    merged['Stock_Close_USD'] = merged['Stock_Close_THB'] / merged['USD_THB_Close']

    if stock_name:
        merged['Stock'] = stock_name

    return merged.sort_values(by='Date').reset_index(drop=True)
````

### ฟังก์ชั่นพลอตกราฟจาก feature engineering

````python
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns
import pandas as pd
import numpy as np
from IPython.display import display

def plot_engineered_feature_scatter(
    stocks_dict,
    feature_formula,
    feature_col,
    feature_label_th,
    title,
    color='#2ca02c'
):
    """
    ฟังก์ชันที่มีความยืดหยุ่นสูงสำหรับพล็อตกราฟ Scatter และหาสหสัมพันธ์ (Correlation)

    Parameters:
    -----------
    stocks_dict : dict
        Dictionary ของข้อมูลหุ้น เช่น {'PTTGC': df_PTTGC, ...}
    feature_formula : function
        ฟังก์ชันคำนวณทาง Feature Engineering เช่น calculate_intraday_features
    feature_col : str
        ชื่อคอลัมน์ของฟีเจอร์ที่เราต้องการวิเคราะห์ เช่น 'Daily_Volatility' หรือ 'Volatility_to_Open'
    feature_label_th : str
        ชื่อภาษาไทยของฟีเจอร์นั้นเพื่อใช้แสดงบนแกนกราฟ
    title : str
        หัวข้อหลักของกราฟทั้งหมด
    color : str
        รหัสสีของจุด Scatter (Default: สีเขียว '#2ca02c')
    """
    # 1. ตั้งค่าฟอนต์ภาษาไทย
    thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
    sns.set_theme(style="whitegrid")
    plt.rcParams['font.family'] = 'Sarabun'
    plt.rcParams['axes.unicode_minus'] = False

    # 2. คำนวณจำนวนแถวสำหรับแสดงผล Grid อัตโนมัติ
    stock_names = list(stocks_dict.keys())
    num_stocks = len(stock_names)
    num_cols = 2
    num_rows = int(np.ceil(num_stocks / num_cols))

    fig, axes = plt.subplots(num_rows, num_cols, figsize=(16, 4.8 * num_rows))
    axes = axes.flatten() if num_stocks > 1 else np.array([axes])

    summary_list = []

    # 3. วนลูปคำนวณและพล็อตกราฟทีละหุ้น
    for idx, name in enumerate(stock_names):
        # ใช้สูตรคำนวณยืดหยุ่นที่ส่งเข้ามาทางพารามิเตอร์
        df_feat = feature_formula(stocks_dict[name])

        # จัดการค่าว่างและดึงเฉพาะข้อมูลที่พร้อมวิเคราะห์
        df_feat = df_feat.dropna(subset=[feature_col, 'Close'])

        vol_series = pd.to_numeric(df_feat[feature_col], errors='coerce')
        close_series = pd.to_numeric(df_feat['Close'], errors='coerce')

        # คำนวณค่าสหสัมพันธ์ (Correlation)
        correlation = vol_series.corr(close_series)

        summary_list.append({
            'หุ้น': name,
            f'ค่าเฉลี่ย {feature_label_th}': vol_series.mean(),
            'Correlation (vs Close)': correlation
        })

        # พล็อตกราฟความสัมพันธ์พร้อม Regression Line
        sns.regplot(
            data=df_feat,
            x='Close',
            y=feature_col,
            ax=axes[idx],
            scatter_kws={'alpha': 0.25, 'color': color, 's': 10},
            line_kws={'color': '#d62728', 'linewidth': 1.5}
        )

        # กำหนดข้อความต่าง ๆ บนกราฟด้วยฟอนต์ไทย
        axes[idx].set_title(f'{name} (Corr = {correlation:.3f})', fontproperties=thai_font, fontsize=12, fontweight='bold')
        axes[idx].set_xlabel('ราคาปิด (Close Price)', fontproperties=thai_font, fontsize=11)
        axes[idx].set_ylabel(feature_label_th, fontproperties=thai_font, fontsize=11)

    # ซ่อนส่วนแกนย่อยที่เหลือ (ในกรณีที่จำนวนกลุ่มย่อยเป็นเลขคี่)
    for j in range(idx + 1, len(axes)):
        fig.delaxes(axes[j])

    fig.suptitle(title, fontproperties=thai_font, fontsize=16, fontweight='bold', y=1.01)
    plt.tight_layout()
    plt.show()

    # 4. แสดงตารางผลลัพธ์
    df_summary = pd.DataFrame(summary_list)
    print(f"\n=== [ตารางสรุปผล] ความสัมพันธ์ระหว่าง ราคาปิด และ {feature_label_th} ===")
    display(df_summary)
````

### calculate_yield_slope (ความชันของเส้นอัตราผลตอบแทนพันธบัตรไทย: TH_20Y - TH_10Y)

````python
import_stocks_dict = {
    'PTTGC': df_PTTGC, 'TOP': df_TOP, 'IRPC': df_IRPC, 'IVL': df_IVL, 'COM7': df_COM7,
    'BGRIM': df_BGRIM, 'GPSC': df_GPSC, 'EGCO': df_EGCO, 'RATCH': df_RATCH, 'BCP': df_BCP
}

export_stocks_dict = {
    'DELTA': df_DELTA, 'HANA': df_HANA, 'KCE': df_KCE, 'CPF': df_CPF, 'TU': df_TU,
    'AAI': df_AAI, 'STA': df_STA, 'BDMS': df_BDMS, 'BH': df_BH, 'ITC': df_ITC
}

# 1. คำนวณ Yield Curve Slope (TH_20Y - TH_10Y)
df_slope = calculate_yield_slope(df_TH_20Y, df_TH_10Y, long_name='Yield_20Y', short_name='Yield_10Y')

# 2. ฟังก์ชัน helper เพื่อรวมข้อมูล Slope เข้ากับราคาปิดของหุ้นแต่ละตัว
def calc_yield_slope_vs_stock(df_stock):
    # ทำความสะอาดข้อมูลวันที่ก่อน
    df_stk_cp = df_stock.copy()
    df_stk_cp['Date'] = pd.to_datetime(df_stk_cp['Date'])

    # รวมกับข้อมูล Slope
    merged = pd.merge(df_stk_cp, df_slope[['Date', 'Yield_Curve_Slope']], on='Date', how='inner')
    return merged

# --- 3. วิเคราะห์ความสัมพันธ์ของ Yield Curve Slope กลุ่มหุ้นนำเข้า ---
plot_engineered_feature_scatter(
    stocks_dict=import_stocks_dict,
    feature_formula=calc_yield_slope_vs_stock,
    feature_col='Yield_Curve_Slope',
    feature_label_th='ความชันพันธบัตรไทย (Yield Curve Slope: 20Y - 10Y)',
    title='การวิเคราะห์ Yield Curve Slope เทียบกับราคาปิดของหุ้นกลุ่มนำเข้า',
    color='#2ca02c'
)

# --- 4. วิเคราะห์ความสัมพันธ์ของ Yield Curve Slope กลุ่มหุ้นส่งออก ---
plot_engineered_feature_scatter(
    stocks_dict=export_stocks_dict,
    feature_formula=calc_yield_slope_vs_stock,
    feature_col='Yield_Curve_Slope',
    feature_label_th='ความชันพันธบัตรไทย (Yield Curve Slope: 20Y - 10Y)',
    title='การวิเคราะห์ Yield Curve Slope เทียบกับราคาปิดของหุ้นกลุ่มส่งออก',
    color='#1f77b4'
)
````

### calculate_relative_strength (อัตราส่วนความแข็งแกร่งเปรียบเทียบเชิงสัมพัทธ์: Export Stock / Import Stock)

**รายละเอียดและคำอธิบายสูตร:**
วิเคราะห์อัตราส่วนระหว่างราคาปิดของกลุ่มหุ้นส่งออกเปรียบเทียบกับกลุ่มหุ้นนำเข้าที่เป็นคู่ค้าหรืออยู่ในสเปกตรัมธุรกิจที่ตรงกันข้ามกัน เพื่อหาว่าเมื่อราคาของหุ้นตัวหนึ่งแข็งแกร่งขึ้น จะส่งผลต่อระดับราคาและความสัมพันธ์ของอีกตัวหนึ่งอย่างไร

* **Relative_Strength_Ratio** = ราคาปิดหุ้นส่งออก (Export Close) / ราคาปิดหุ้นนำเข้า (Import Close)
* **Return_Diff_Export_vs_Import** = เปอร์เซ็นต์การเปลี่ยนแปลงของหุ้นส่งออก - เปอร์เซ็นต์การเปลี่ยนแปลงของหุ้นนำเข้า (คำนวณเพื่อใช้ดูในมิติ % Change)

````python
# 1. กำหนดคู่หุ้นส่งออกและหุ้นนำเข้าเพื่อนำมาจับคู่เปรียบเทียบวิเคราะห์เชิงสัมพัทธ์
stock_pairs = [
    ('DELTA', 'PTTGC'),
    ('HANA', 'COM7'),
    ('KCE', 'IVL'),
    ('TU', 'CPF'),
    ('STA', 'BCP')
]

# 2. ฟังก์ชันประยุกต์ใช้เพื่อรวมคำนวณสำหรับสร้าง Scatter Plot
def calc_relative_strength_for_pair(df_export, df_import, exp_name, imp_name):
    df_pair = calculate_relative_strength(df_export, df_import, exp_stock_name=exp_name, imp_stock_name=imp_name)
    # คืนค่า DataFrame พร้อมกำหนดคอลัมน์หลักเป็นราคาปิดฝั่งส่งออกและคอลัมน์ฟีเจอร์เป็น อัตราส่วนความแข็งแกร่ง
    df_pair = df_pair.rename(columns={'Export_Close': 'Close'})
    return df_pair

# 3. สร้างโครงสร้าง dictionary สำหรับส่งผ่านพล็อตเตอร์
pairs_data_dict = {}
for exp_name, imp_name in stock_pairs:
    df_exp = globals()[f'df_{exp_name}']
    df_imp = globals()[f'df_{imp_name}']
    label = f'{exp_name} vs {imp_name}'
    # บันทึกเป็น lambda ให้สอดคล้องกับพารามิเตอร์ของฟังก์ชันยืดหยุ่น
    pairs_data_dict[label] = (df_exp, df_imp, exp_name, imp_name)

# ปรับฟังก์ชันแสดงผลเพื่อให้ทำงานกับข้อมูลโครงสร้างคู่หุ้น
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns
import numpy as np

def plot_relative_strength_pairs_scatter(pairs_dict, color='#1f77b4'):
    thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
    sns.set_theme(style="whitegrid")

    num_pairs = len(pairs_dict)
    num_cols = 2
    num_rows = int(np.ceil(num_pairs / num_cols))

    fig, axes = plt.subplots(num_rows, num_cols, figsize=(16, 4.8 * num_rows))
    axes = axes.flatten() if num_pairs > 1 else np.array([axes])

    summary_list = []

    for idx, (label, (df_exp, df_imp, exp_name, imp_name)) in enumerate(pairs_dict.items()):
        df_feat = calculate_relative_strength(df_exp, df_imp, exp_stock_name=exp_name, imp_stock_name=imp_name)
        df_feat = df_feat.dropna(subset=['Relative_Strength_Ratio', 'Export_Close'])

        ratio_series = pd.to_numeric(df_feat['Relative_Strength_Ratio'], errors='coerce')
        close_series = pd.to_numeric(df_feat['Export_Close'], errors='coerce')

        correlation = ratio_series.corr(close_series)

        summary_list.append({
            'คู่หุ้น (Export vs Import)': label,
            'ค่าเฉลี่ย Relative Strength Ratio': ratio_series.mean(),
            'Correlation (vs Export Close)': correlation
        })

        sns.regplot(
            data=df_feat,
            x='Export_Close',
            y='Relative_Strength_Ratio',
            ax=axes[idx],
            scatter_kws={'alpha': 0.25, 'color': color, 's': 10},
            line_kws={'color': '#d62728', 'linewidth': 1.5}
        )

        axes[idx].set_title(f'{label} (Corr = {correlation:.3f})', fontproperties=thai_font, fontsize=12, fontweight='bold')
        axes[idx].set_xlabel(f'ราคาปิดของ {exp_name}', fontproperties=thai_font, fontsize=11)
        axes[idx].set_ylabel('อัตราความแข็งแกร่ง (Relative Strength Ratio)', fontproperties=thai_font, fontsize=11)

    for j in range(idx + 1, len(axes)):
        fig.delaxes(axes[j])

    fig.suptitle('การวิเคราะห์ Relative Strength Ratio คู่หุ้นส่งออกเปรียบเทียบกับหุ้นนำเข้า', fontproperties=thai_font, fontsize=16, fontweight='bold', y=1.01)
    plt.tight_layout()
    plt.show()

    df_summary = pd.DataFrame(summary_list)
    print("\n=== [ตารางสรุปผล] ความสัมพันธ์ระหว่าง ราคาปิด และ Relative Strength Ratio ===")
    display(df_summary)

# ดำเนินการพล็อตกราฟ
plot_relative_strength_pairs_scatter(pairs_data_dict, color='#9467bd')
````

### convert_price_to_usd (การแปลงราคาปิดหุ้นจากสกุลเงินบาทเป็นดอลลาร์สหรัฐ: THB -> USD)

**รายละเอียดและคำอธิบายสูตร:**
วิเคราะห์ความสัมพันธ์ระหว่างราคาปิดของหุ้นในหน่วยดอลลาร์สหรัฐ (USD) กับราคาปิดสกุลเงินบาท (THB) เพื่อดูทิศทางและความแข็งแกร่งเมื่อรับรู้ผลกระทบจากความผันผวนของอัตราแลกเปลี่ยนโดยตรง

* **Stock_Close_USD** = ราคาปิดหุ้นสกุลบาท (Stock_Close_THB) / อัตราแลกเปลี่ยน USD_THB (USD_THB_Close)

````python
import_stocks_dict = {
    'PTTGC': df_PTTGC, 'TOP': df_TOP, 'IRPC': df_IRPC, 'IVL': df_IVL, 'COM7': df_COM7,
    'BGRIM': df_BGRIM, 'GPSC': df_GPSC, 'EGCO': df_EGCO, 'RATCH': df_RATCH, 'BCP': df_BCP
}

export_stocks_dict = {
    'DELTA': df_DELTA, 'HANA': df_HANA, 'KCE': df_KCE, 'CPF': df_CPF, 'TU': df_TU,
    'AAI': df_AAI, 'STA': df_STA, 'BDMS': df_BDMS, 'BH': df_BH, 'ITC': df_ITC
}

# ฟังก์ชันประยุกต์ใช้เพื่อแปลงข้อมูลสำหรับส่งเข้าฟังก์ชันแสดงผล
def calc_close_usd(df_stock):
    return convert_price_to_usd(df_stock, df_USD_THB)

# --- 1. วิเคราะห์ราคาหุ้นปิดสกุลดอลลาร์สหรัฐ (Stock Close USD) ของกลุ่มหุ้นนำเข้า ---
plot_engineered_feature_scatter(
    stocks_dict=import_stocks_dict,
    feature_formula=calc_close_usd,
    feature_col='Stock_Close_USD',
    feature_label_th='ราคาปิดหน่วยดอลลาร์สหรัฐ (Stock Close USD)',
    title='การวิเคราะห์ราคาปิดในหน่วย USD (Stock Close USD) ของหุ้นกลุ่มนำเข้า',
    color='#2ca02c'
)

# --- 2. วิเคราะห์ราคาหุ้นปิดสกุลดอลลาร์สหรัฐ (Stock Close USD) ของกลุ่มหุ้นส่งออก ---
plot_engineered_feature_scatter(
    stocks_dict=export_stocks_dict,
    feature_formula=calc_close_usd,
    feature_col='Stock_Close_USD',
    feature_label_th='ราคาปิดหน่วยดอลลาร์สหรัฐ (Stock Close USD)',
    title='การวิเคราะห์ราคาปิดในหน่วย USD (Stock Close USD) ของหุ้นกลุ่มส่งออก',
    color='#1f77b4'
)
````

# PIPELINE

### 1. Data Integration and Alignment
- ทำการคัดกรองข้อมูลหุ้นส่งออก (Export) โดยตัด `AAI` และ `ITC` ออก คงเหลือหุ้น 8 ตัว
- ทำการคัดกรองข้อมูลหุ้นนำเข้า (Import) โดยตัด `COM7` และ `BGRIM` ออก คงเหลือหุ้น 8 ตัว
- คำนวณหาค่าเฉลี่ยรายวันของกลุ่มส่งออก (`Export_Avg`) และกลุ่มนำเข้า (`Import_Avg`)
- จัดเตรียมและรวมข้อมูลแกน X (ฟีเจอร์): `Brent`, `US_10Y`, `USD_THB`, `Volatility_to_Open`, `Yield_Curve_Slope` (TH_20Y - TH_10Y)
- จัดเตรียมและรวมข้อมูลแกน Y (เป้าหมาย): `Import_Avg`, `Export_Avg` และหุ้นรายตัวตัวแทน ได้แก่ `BH`, `BDMS` (กลุ่มส่งออก) และ `EGCO`, `BCP` (กลุ่มนำเข้า)
- ทำการคลีนข้อมูล จัดการค่าว่างแบบ Forward/Backward Fill และทำการ Aligned ข้อมูลให้อยู่บนอนุกรมเวลา (Date) เดียวกันทั้งหมด

````python
import pandas as pd
import numpy as np

# 1. คัดเลือกรายชื่อหุ้น (คัดออกกลุ่มละ 2 ตัวตามบรีฟ)
import_8_names = ['PTTGC', 'TOP', 'IRPC', 'IVL', 'GPSC', 'EGCO', 'RATCH', 'BCP']
export_8_names = ['DELTA', 'HANA', 'KCE', 'CPF', 'TU', 'STA', 'BDMS', 'BH']

def get_clean_return_and_close(df_name):
    df = globals()[df_name].copy()
    df['Date'] = pd.to_datetime(df['Date'])

    # ทำความสะอาด % Change (ถ้าเป็น string ให้แปลงเป็น float)
    if 'Pct_Change' in df.columns:
        if df['Pct_Change'].dtype == 'object':
            df['Return'] = df['Pct_Change'].str.rstrip('%').astype('float') / 100.0
        else:
            df['Return'] = pd.to_numeric(df['Pct_Change'], errors='coerce')
    else:
        df = df.sort_values('Date')
        df['Return'] = df['Close'].pct_change()

    df['Close'] = pd.to_numeric(df['Close'], errors='coerce')
    df['Open'] = pd.to_numeric(df['Open'], errors='coerce')
    df['High'] = pd.to_numeric(df['High'], errors='coerce')
    df['Low'] = pd.to_numeric(df['Low'], errors='coerce')

    return df[['Date', 'Close', 'Open', 'High', 'Low', 'Return']]

# โหลดและแปลงข้อมูลหุ้นนำเข้าและส่งออก
import_dfs = {name: get_clean_return_and_close(f'df_{name}') for name in import_8_names}
export_dfs = {name: get_clean_return_and_close(f'df_{name}') for name in export_8_names}

# 2. คำนวณผลตอบแทนเฉลี่ยรายวันของกลุ่มนำเข้า (Import_Avg) และส่งออก (Export_Avg)
import_pct_merged = pd.DataFrame({'Date': df_USD_THB['Date']})
import_pct_merged['Date'] = pd.to_datetime(import_pct_merged['Date'])
for name, df_stk in import_dfs.items():
    import_pct_merged = pd.merge(import_pct_merged, df_stk[['Date', 'Return']].rename(columns={'Return': name}), on='Date', how='left')

import_pct_merged['Import_Avg'] = import_pct_merged[import_8_names].mean(axis=1)

export_pct_merged = pd.DataFrame({'Date': df_USD_THB['Date']})
export_pct_merged['Date'] = pd.to_datetime(export_pct_merged['Date'])
for name, df_stk in export_dfs.items():
    export_pct_merged = pd.merge(export_pct_merged, df_stk[['Date', 'Return']].rename(columns={'Return': name}), on='Date', how='left')

export_pct_merged['Export_Avg'] = export_pct_merged[export_8_names].mean(axis=1)

# 3. เตรียมข้อมูลฟีเจอร์ภายนอก (X Factors)
df_brent_clean = get_clean_return_and_close('df_เบรนท์')[['Date', 'Return']].rename(columns={'Return': 'Brent'})
df_us10y_clean = get_clean_return_and_close('df_US_10Y')[['Date', 'Return']].rename(columns={'Return': 'US_10Y'})
df_usdthb_clean = get_clean_return_and_close('df_USD_THB')[['Date', 'Return']].rename(columns={'Return': 'USD_THB'})

# ฟีเจอร์ Volatility_to_Open เฉลี่ยของทั้ง 16 หุ้นในระบบ
vol_merged = pd.DataFrame({'Date': df_USD_THB['Date']})
vol_merged['Date'] = pd.to_datetime(vol_merged['Date'])
vol_cols = []
for name, df_stk in {**import_dfs, **export_dfs}.items():
    df_stk['Vol_Open'] = (df_stk['High'] - df_stk['Low']) / df_stk['Open']
    vol_merged = pd.merge(vol_merged, df_stk[['Date', 'Vol_Open']].rename(columns={'Vol_Open': f'Vol_Open_{name}'}), on='Date', how='left')
    vol_cols.append(f'Vol_Open_{name}')

vol_merged['Volatility_to_Open'] = vol_merged[vol_cols].mean(axis=1)

# ฟีเจอร์ Yield_Curve_Slope (TH_20Y - TH_10Y)
df_slope_clean = calculate_yield_slope(df_TH_20Y, df_TH_10Y)[['Date', 'Yield_Curve_Slope']]

# 4. ประกอบร่างตารางสุดท้ายด้วยการทำ Alignment
# ใช้ Date ตลาดหลักทรัพย์ไทย (อ้างอิงจากอัตราแลกเปลี่ยน USD_THB เป็นหลัก)
base_df = pd.DataFrame({'Date': sorted(df_USD_THB['Date'].unique())})
base_df['Date'] = pd.to_datetime(base_df['Date'])

final_pipeline_df = base_df.copy()
final_pipeline_df = pd.merge(final_pipeline_df, df_brent_clean, on='Date', how='inner')
final_pipeline_df = pd.merge(final_pipeline_df, df_us10y_clean, on='Date', how='inner')
final_pipeline_df = pd.merge(final_pipeline_df, df_usdthb_clean, on='Date', how='inner')
final_pipeline_df = pd.merge(final_pipeline_df, vol_merged[['Date', 'Volatility_to_Open']], on='Date', how='inner')
final_pipeline_df = pd.merge(final_pipeline_df, df_slope_clean, on='Date', how='inner')

# รวมเป้าหมาย (Y Targets)
final_pipeline_df = pd.merge(final_pipeline_df, import_pct_merged[['Date', 'Import_Avg']], on='Date', how='inner')
final_pipeline_df = pd.merge(final_pipeline_df, export_pct_merged[['Date', 'Export_Avg']], on='Date', how='inner')
final_pipeline_df = pd.merge(final_pipeline_df, export_dfs['BH'][['Date', 'Return']].rename(columns={'Return': 'BH_Pct'}), on='Date', how='inner')
final_pipeline_df = pd.merge(final_pipeline_df, export_dfs['BDMS'][['Date', 'Return']].rename(columns={'Return': 'BDMS_Pct'}), on='Date', how='inner')
final_pipeline_df = pd.merge(final_pipeline_df, import_dfs['EGCO'][['Date', 'Return']].rename(columns={'Return': 'EGCO_Pct'}), on='Date', how='inner')
final_pipeline_df = pd.merge(final_pipeline_df, import_dfs['BCP'][['Date', 'Return']].rename(columns={'Return': 'BCP_Pct'}), on='Date', how='inner')

# จัดเรียงตามมิติเวลาจากอดีตไปหาปัจจุบัน
final_pipeline_df = final_pipeline_df.sort_values(by='Date').reset_index(drop=True)

# เติมค่าว่างด้วย Forward Fill เท่านั้น และลบค่าว่างที่เหลือ (ห้าม bfill)
final_pipeline_df = final_pipeline_df.ffill().dropna()

# แยกส่วนประกอบ X และ Y
x_cols = ['Brent', 'US_10Y', 'USD_THB', 'Volatility_to_Open', 'Yield_Curve_Slope']
y_cols = ['Import_Avg', 'Export_Avg', 'BH_Pct', 'BDMS_Pct', 'EGCO_Pct', 'BCP_Pct']

X = final_pipeline_df[x_cols]
Y = final_pipeline_df[y_cols]

print(f"✓ รวมและจัดเรียงข้อมูลอนุกรมเวลาเรียบร้อยแล้ว: จำนวนข้อมูลทั้งหมด {final_pipeline_df.shape[0]} วัน")
display(final_pipeline_df.head(5))
````

### 2. Temporal Train-Test Split
- จัดลำดับข้อมูลรวมทั้งหมดตามมิติเวลา (Chronological order) จากอดีตไปหาปัจจุบันอย่างถูกต้องเพื่อป้องกันปัญหา Data Leakage
- ทำการแบ่งชุดข้อมูลออกเป็น ชุดฝึกสอน (Train Set) 80% และชุดทดสอบ (Test Set) 20% โดยอิงตามเส้นแบ่งเวลา (Temporal Cutoff Split) แทนการสุ่มสุ่มทั่วไป

````python
import pandas as pd
import numpy as np

# 1. มั่นใจว่าข้อมูลเรียงลำดับตามวันที่จากอดีตไปปัจจุบัน
final_pipeline_df = final_pipeline_df.sort_values(by='Date').reset_index(drop=True)

# 2. แยก Features (X) และ Targets (Y) พร้อมเก็บคอลัมน์ Date ไว้ใช้อ้างอิงตอนพลอต
x_cols = ['Brent', 'US_10Y', 'USD_THB', 'Volatility_to_Open', 'Yield_Curve_Slope']
y_cols = ['Import_Avg', 'Export_Avg', 'BH_Pct', 'BDMS_Pct', 'EGCO_Pct', 'BCP_Pct']

dates = final_pipeline_df['Date']
X = final_pipeline_df[x_cols]
Y = final_pipeline_df[y_cols]

# 3. คำนวณจุดตัดแบ่งข้อมูล (Temporal Cutoff) ที่ 80%
split_ratio = 0.8
split_idx = int(len(final_pipeline_df) * split_ratio)

# 4. แบ่งข้อมูลออกจากกันโดยไม่มีการสุ่มสลับ (No Shuffle)
X_train, X_test = X.iloc[:split_idx].reset_index(drop=True), X.iloc[split_idx:].reset_index(drop=True)
Y_train, Y_test = Y.iloc[:split_idx].reset_index(drop=True), Y.iloc[split_idx:].reset_index(drop=True)
train_dates, test_dates = dates.iloc[:split_idx].reset_index(drop=True), dates.iloc[split_idx:].reset_index(drop=True)

# 5. แสดงผลการแบ่งข้อมูลเพื่อตรวจสอบ
print("=== [สรุปผลการทำ Temporal Train-Test Split (80:20)] ===\n")
print(f"จำนวนข้อมูลทั้งหมด: {len(final_pipeline_df)} วัน")
print(f"ชุดข้อมูลฝึกสอน (Train Set): {X_train.shape[0]} วัน ({split_ratio*100:.1f}%)")
print(f"  - วันเริ่มต้น: {train_dates.min().strftime('%Y-%m-%d')} ถึง วันสิ้นสุด: {train_dates.max().strftime('%Y-%m-%d')}")
print(f"ชุดข้อมูลทดสอบ (Test Set): {X_test.shape[0]} วัน ({ (1-split_ratio)*100:.1f}%)")
print(f"  - วันเริ่มต้น: {test_dates.min().strftime('%Y-%m-%d')} ถึง วันสิ้นสุด: {test_dates.max().strftime('%Y-%m-%d')}")
print(f"\nขนาด X_train: {X_train.shape} | ขนาด Y_train: {Y_train.shape}")
print(f"ขนาด X_test:  {X_test.shape}  | ขนาด Y_test:  {Y_test.shape}")
````

### 3. Pipeline Design and Preprocessing
- ออกแบบ Scikit-Learn Pipeline ที่มีระเบียบและมาตรฐาน
- กำหนดการ Preprocessing: การใช้ `StandardScaler` เพื่อปรับขนาดของฟีเจอร์แกน X ทั้งหมดให้อยู่ในสเกลมาตรฐานเดียวกัน ป้องกันความลำเอียงจากการใช้ขนาดหน่วยข้อมูลที่ต่างกันต่อตัวแบบพยากรณ์

````python
import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# 1. นิยามและสร้าง Preprocessing Pipeline
preprocessing_pipeline = Pipeline([
    ('scaler', StandardScaler())
])

# 2. Fit และ Transform บน X_train เฉพาะชุดเทรนเท่านั้น เพื่อป้องกัน Data Leakage
X_train_scaled_array = preprocessing_pipeline.fit_transform(X_train)

# 3. Transform บน X_test ด้วยพารามิเตอร์ที่ได้จาก X_train เท่านั้น
X_test_scaled_array = preprocessing_pipeline.transform(X_test)

# 4. แปลงกลับมาเป็น DataFrame เพื่อการตรวจสอบและใช้งานที่ง่ายขึ้น
X_train_scaled = pd.DataFrame(X_train_scaled_array, columns=x_cols)
X_test_scaled = pd.DataFrame(X_test_scaled_array, columns=x_cols)

# 5. แสดงตารางเปรียบเทียบค่าทางสถิติก่อนและหลังสเกลข้อมูล
comparison_summary = pd.DataFrame({
    'Mean (Before Scaled)': X_train.mean(),
    'Std (Before Scaled)': X_train.std(),
    'Mean (After Scaled)': X_train_scaled.mean(),
    'Std (After Scaled)': X_train_scaled.std()
})

print("=== [สรุปการทำ Feature Scaling ด้วย StandardScaler บน X_train] ===\n")
display(comparison_summary)
print("\nตัวอย่างข้อมูล X_train_scaled (5 แถวแรก):")
display(X_train_scaled.head(5))
print(f"ขนาดของ X_train_scaled: {X_train_scaled.shape} | ขนาดของ X_test_scaled: {X_test_scaled.shape}")
````

### 4. Model Training and Baseline Evaluation
- ดำเนินการสร้างโมเดล Machine Learning หลายประเภท (อาทิ `Linear Regression`, `Ridge`, `Random Forest`, หรือ `Gradient Boosting`)
- ทำการฝึกสอนโมเดล (Fit) บนชุดข้อมูลสอน และประเมินประสิทธิภาพเบื้องต้นเทียบกับค่า Baseline บนชุดข้อมูลทดสอบ (Test Set)
- คำนวณมาตรวัดความแม่นยำ เช่น `R² Score`, `Mean Absolute Error (MAE)`, และ `Root Mean Squared Error (RMSE)` สำหรับเป้าหมาย Y แต่ละตัว

````python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from IPython.display import display

# 1. แปลงเป้าหมาย Y_train และ Y_test เป็น Binary Classification (1 เมื่อ Return > 0, นอกเหนือจากนั้นเป็น 0)
Y_train_class = Y_train.apply(lambda x: np.where(x > 0, 1, 0))
Y_test_class = Y_test.apply(lambda x: np.where(x > 0, 1, 0))

# 2. นิยามโมเดลที่ใช้งาน
models = {
    'Logistic Regression': LogisticRegression(random_state=42, max_iter=1000),
    'Random Forest': RandomForestClassifier(random_state=42, n_jobs=-1),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42)
}

# เตรียมลิสต์สำหรับเก็บผลลัพธ์การประเมิน
classification_results = []

# 3. วนลูปเทรนและประเมินผลแยกตามแต่ละตัวแปรเป้าหมาย (Y Targets)
for col in Y_train_class.columns:
    y_tr = Y_train_class[col]
    y_te = Y_test_class[col]

    # --- A. Baseline Model (Majority Class) ---
    majority_class = y_tr.mode()[0]
    baseline_preds = np.full(shape=len(y_te), fill_value=majority_class)

    base_acc = accuracy_score(y_te, baseline_preds)
    base_prec = precision_score(y_te, baseline_preds, zero_division=0)
    base_rec = recall_score(y_te, baseline_preds, zero_division=0)
    base_f1 = f1_score(y_te, baseline_preds, zero_division=0)
    # สำหรับ Majority class ทิศทางเดียว จะได้ค่า ROC-AUC เป็น 0.5
    base_auc = 0.5

    classification_results.append({
        'Target': col,
        'Model': 'Baseline (Majority Class)',
        'Accuracy': base_acc,
        'Precision': base_prec,
        'Recall': base_rec,
        'F1-Score': base_f1,
        'ROC-AUC': base_auc
    })

    # --- B. Machine Learning Models ---
    for model_name, clf in models.items():
        # ฟิตเฉพาะบนข้อมูล Train เท่านั้น ป้องกัน Data Leakage
        clf.fit(X_train_scaled, y_tr)

        # ทำนายผลบน Test Set
        preds = clf.predict(X_test_scaled)

        # ดึงความน่าจะเป็น (Probability) สำหรับการคำนวณ ROC-AUC
        if hasattr(clf, "predict_proba"):
            probs = clf.predict_proba(X_test_scaled)[:, 1]
            auc_score = roc_auc_score(y_te, probs)
        else:
            auc_score = np.nan

        acc = accuracy_score(y_te, preds)
        prec = precision_score(y_te, preds, zero_division=0)
        rec = recall_score(y_te, preds, zero_division=0)
        f1 = f1_score(y_te, preds, zero_division=0)

        classification_results.append({
            'Target': col,
            'Model': model_name,
            'Accuracy': acc,
            'Precision': prec,
            'Recall': rec,
            'F1-Score': f1,
            'ROC-AUC': auc_score
        })

# 4. แปลงผลการประเมินเป็น DataFrame สรุปข้อมูล
df_classification_summary = pd.DataFrame(classification_results)

# แสดงตารางสรุปแบบแยกตามแต่ละ Target ให้เห็นชัดเจน
print("=== ตารางสรุปผลประสิทธิภาพโมเดล Classification แยกตามตัวแปรเป้าหมาย ===")
for target in Y_train_class.columns:
    print(f"\nTarget: {target}")
    df_sub = df_classification_summary[df_classification_summary['Target'] == target].copy()
    display(df_sub.sort_values(by='Accuracy', ascending=False).reset_index(drop=True))
````

### 5. Visualizing Predictions
- สร้างและแสดงผลกราฟพล็อตเปรียบเทียบระหว่าง ค่าจริง (Actual Values) และค่าที่โมเดลพยากรณ์ได้ (Predicted Values) บนชุดข้อมูลทดสอบ (Test Set)
- แสดงกราฟอนุกรมเวลาเพื่อตรวจดูความสอดคล้องและการล้าหลังของค่าทำนายในแต่ละตัวแปรเป้าหมาย Y ทั้ง 6 ตัวเพื่อประเมินความแม่นยำทางกายภาพ

````python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
import numpy as np
import pandas as pd
import matplotlib.font_manager as fm

# 1. ตั้งค่าฟอนต์ภาษาไทยสำหรับการแสดงผลบนกราฟ
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False

# 2. ทำนายและเตรียมข้อมูลทิศทาง (Direction) สำหรับ Import_Avg และ Export_Avg
# เลือกโมเดลหลักเป็น Logistic Regression
model_import = LogisticRegression(random_state=42, max_iter=1000)
model_export = LogisticRegression(random_state=42, max_iter=1000)

# ฟิตโมเดลด้วย Train set
model_import.fit(X_train_scaled, Y_train_class['Import_Avg'])
model_export.fit(X_train_scaled, Y_train_class['Export_Avg'])

# พยากรณ์บน Test set
preds_import = model_import.predict(X_test_scaled)
preds_export = model_export.predict(X_test_scaled)

actual_import = Y_test_class['Import_Avg'].values
actual_export = Y_test_class['Export_Avg'].values

# 3. สร้าง DataFrame เพื่อใช้พล็อตกราฟเปรียบเทียบตามอนุกรมเวลา (ดึง 60 วันล่าสุดของ Test Set เพื่อความชัดเจนในการแสดงผล)
plot_df = pd.DataFrame({
    'Date': test_dates,
    'Actual_Import': actual_import,
    'Pred_Import': preds_import,
    'Actual_Export': actual_export,
    'Pred_Export': preds_export
}).sort_values('Date').tail(60).reset_index(drop=True)

# =========================================================
# พล็อตกราฟที่ 1: เปรียบเทียบ Actual vs Predicted Direction
# =========================================================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(15, 8), sharex=True)

# กราฟย่อยที่ 1: Import_Avg
ax1.step(plot_df['Date'], plot_df['Actual_Import'], where='mid', label='ทิศทางจริง (Actual)', color='#2ca02c', linewidth=2)
ax1.step(plot_df['Date'], plot_df['Pred_Import'], where='mid', label='ทิศทางที่ทำนาย (Predicted)', color='#d62728', linestyle='--', linewidth=2)
ax1.set_title('การเปรียบเทียบทิศทางราคาจริงและราคาพยากรณ์: Import_Avg (60 วันล่าสุดใน Test Set)', fontproperties=thai_font, fontsize=14, fontweight='bold')
ax1.set_ylabel('ทิศทาง (1 = ขึ้น, 0 = ลง)', fontproperties=thai_font, fontsize=11)
ax1.set_yticks([0, 1])
ax1.legend(prop=thai_font, loc='upper left')
ax1.grid(True, linestyle=':', alpha=0.6)

# กราฟย่อยที่ 2: Export_Avg
ax2.step(plot_df['Date'], plot_df['Actual_Export'], where='mid', label='ทิศทางจริง (Actual)', color='#1f77b4', linewidth=2)
ax2.step(plot_df['Date'], plot_df['Pred_Export'], where='mid', label='ทิศทางที่ทำนาย (Predicted)', color='#ff7f0e', linestyle='--', linewidth=2)
ax2.set_title('การเปรียบเทียบทิศทางราคาจริงและราคาพยากรณ์: Export_Avg (60 วันล่าสุดใน Test Set)', fontproperties=thai_font, fontsize=14, fontweight='bold')
ax2.set_xlabel('วันที่ (Date)', fontproperties=thai_font, fontsize=11)
ax2.set_ylabel('ทิศทาง (1 = ขึ้น, 0 = ลง)', fontproperties=thai_font, fontsize=11)
ax2.set_yticks([0, 1])
ax2.legend(prop=thai_font, loc='upper left')
ax2.grid(True, linestyle=':', alpha=0.6)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# =========================================================
# พล็อตกราฟที่ 2: Heatmap Confusion Matrix
# =========================================================
cm_import = confusion_matrix(actual_import, preds_import)
cm_export = confusion_matrix(actual_export, preds_export)

fig_cm, (ax_cm1, ax_cm2) = plt.subplots(1, 2, figsize=(14, 5.5))

# Confusion Matrix: Import_Avg
sns.heatmap(cm_import, annot=True, fmt='d', cmap='Greens', ax=ax_cm1, cbar=False,
            xticklabels=['ทำนาย ลง (0)', 'ทำนาย ขึ้น (1)'],
            yticklabels=['จริง ลง (0)', 'จริง ขึ้น (1)'])
ax_cm1.set_title('Confusion Matrix: ดัชนีกลุ่มนำเข้า (Import_Avg)', fontproperties=thai_font, fontsize=13, fontweight='bold')
ax_cm1.set_xlabel('ค่าที่โมเดลทำนาย (Predicted)', fontproperties=thai_font, fontsize=11)
ax_cm1.set_ylabel('ค่าจริง (Actual)', fontproperties=thai_font, fontsize=11)
for label in (ax_cm1.get_xticklabels() + ax_cm1.get_yticklabels()):
    label.set_fontproperties(thai_font)

# Confusion Matrix: Export_Avg
sns.heatmap(cm_export, annot=True, fmt='d', cmap='Blues', ax=ax_cm2, cbar=False,
            xticklabels=['ทำนาย ลง (0)', 'ทำนาย ขึ้น (1)'],
            yticklabels=['จริง ลง (0)', 'จริง ขึ้น (1)'])
ax_cm2.set_title('Confusion Matrix: ดัชนีกลุ่มส่งออก (Export_Avg)', fontproperties=thai_font, fontsize=13, fontweight='bold')
ax_cm2.set_xlabel('ค่าที่โมเดลทำนาย (Predicted)', fontproperties=thai_font, fontsize=11)
ax_cm2.set_ylabel('ค่าจริง (Actual)', fontproperties=thai_font, fontsize=11)
for label in (ax_cm2.get_xticklabels() + ax_cm2.get_yticklabels()):
    label.set_fontproperties(thai_font)

plt.tight_layout()
plt.show()
````

### 6. Final Task
- สรุปประสิทธิภาพและวิเคราะห์ความยากง่ายในการทำนายเป้าหมายแต่ละตัว (เปรียบเทียบว่ากลุ่มดัชนีเฉลี่ย `Export_Avg` / `Import_Avg` กับหุ้นรายตัว `BH`, `BDMS`, `EGCO`, `BCP` ฝั่งไหนมีประสิทธิภาพของโมเดลที่ดีกว่ากัน)
- วิเคราะห์หา Feature Importance ของปัจจัยมหภาคและตัวแปรเสริมเพื่อดูว่าปัจจัยใดมีอิทธิพลต่อผลตอบแทนของกลุ่มพอร์ตเหล่านั้นมากที่สุด

````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.font_manager as fm
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score

# 1. โหลดและตั้งค่าฟอนต์ภาษาไทยสำหรับการทำกราฟสรุป
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False

# 2. ทำการคำนวณแบบจำลอง Logistic Regression สำหรับ Y ทั้ง 6 ตัวแปร
target_cols = ['Import_Avg', 'Export_Avg', 'BH_Pct', 'BDMS_Pct', 'EGCO_Pct', 'BCP_Pct']
x_cols = ['Brent', 'US_10Y', 'USD_THB', 'Volatility_to_Open', 'Yield_Curve_Slope']

# ข้อมูลสำหรับทำนายทิศทาง 5 วันข้างหน้า
Y_train_5d = Y_train.rolling(window=5).sum().shift(-5)
Y_test_5d = Y_test.rolling(window=5).sum().shift(-5)

train_mask = Y_train_5d.notna().all(axis=1)
test_mask = Y_test_5d.notna().all(axis=1)

X_train_clean = X_train_scaled[train_mask]
X_test_clean = X_test_scaled[test_mask]

Y_train_class_5d = Y_train_5d[train_mask].apply(lambda x: np.where(x > 0, 1, 0))
Y_test_class_5d = Y_test_5d[test_mask].apply(lambda x: np.where(x > 0, 1, 0))

# สร้างดิคเก็บค่าสัมประสิทธิ์และผลลัพธ์โมเดล
coef_data = {}
performance_results = []

for col in target_cols:
    model = LogisticRegression(random_state=42, max_iter=1000)
    model.fit(X_train_clean, Y_train_class_5d[col])

    # บันทึกสัมประสิทธิ์
    coef_data[col] = model.coef_[0]

    # วัดประสิทธิภาพบน Test Set
    preds = model.predict(X_test_clean)
    acc = accuracy_score(Y_test_class_5d[col], preds)
    f1 = f1_score(Y_test_class_5d[col], preds, zero_division=0)

    performance_results.append({
        'Target': col,
        'Type': 'พอร์ตเฉลี่ย (Portfolio Index)' if 'Avg' in col else 'หุ้นรายตัว (Single Stock)',
        'Accuracy (5-Day Direction)': acc,
        'F1-Score': f1
    })

df_performance = pd.DataFrame(performance_results)
df_coefs = pd.DataFrame(coef_data, index=x_cols)

# =========================================================
# 3. จัดทำกราฟแท่งเปรียบเทียบ Feature Importance ( Coefficients )
# =========================================================
fig, axes = plt.subplots(3, 2, figsize=(15, 12), sharey=True)
axes = axes.flatten()

colors = ['#2ca02c', '#1f77b4', '#9467bd', '#bcbd22', '#e377c2', '#ff7f0e']

for idx, col in enumerate(target_cols):
    importance = df_coefs[col]
    sns.barplot(x=importance.values, y=importance.index, ax=axes[idx], palette='RdBu_r', hue=importance.index, legend=False)
    axes[idx].set_title(f'อิทธิพลปัจจัยมหภาคต่อทิศทางของ: {col}', fontproperties=thai_font, fontsize=12, fontweight='bold')
    axes[idx].axvline(0, color='black', linestyle='--', linewidth=0.8)
    axes[idx].set_xlabel('ขนาดและทิศทางอิทธิพล (Coefficient)', fontproperties=thai_font, fontsize=9)
    for tick in axes[idx].get_yticklabels():
        tick.set_fontproperties(thai_font)

fig.suptitle('Feature Importance (สัมประสิทธิ์อิทธิพลของปัจจัยมหภาคต่อการทำนายทิศทาง 5 วัน)', fontproperties=thai_font, fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# แสดงตารางวิเคราะห์ความแม่นยำเปรียบเทียบ
print("\n=== ตารางเปรียบเทียบประสิทธิภาพการทำนายทิศทางระยะ 5 วัน (พอร์ตดัชนีเฉลี่ย VS หุ้นรายตัว) ===")
display(df_performance.sort_values(by='Accuracy (5-Day Direction)', ascending=False).reset_index(drop=True))
````

### 7. Executive Summary (บทสรุปวิเคราะห์เชิงลึกสำหรับผู้บริหาร)

#### **1) สรุปประสิทธิภาพการพยากรณ์เชิงโครงสร้าง (Portfolio vs. Single Stock)**
* **พอร์ตเฉลี่ย (Import_Avg & Export_Avg) ทำนายง่ายและแม่นยำสูงกว่าหุ้นรายตัว:**
  จากผลลัพธ์พบว่าแบบจำลองที่พยากรณ์ทิศทางของดัชนีเฉลี่ยของกลุ่มพอร์ตนำเข้าและส่งออกมีประสิทธิภาพของความแม่นยำสูงกว่าการทำนายหุ้นรายตัว (เช่น `BH`, `BDMS`, `EGCO`, `BCP`)
* **เหตุผลสนับสนุนเชิงการเงิน (Financial Noise & Diversification):**
  ราคาของหุ้นรายตัวมีปัจจัยความเสี่ยงเฉพาะตัว (Idiosyncratic Risk) หรือมี Noise ค่อนข้างสูง เช่น ข่าวการประกาศผลประกอบการเฉพาะตัวแปร การเก็งกำไรในตลาดรายวัน และสภาพคล่องของหุ้นแต่ละตัว ทำให้ทิศทางราคามีความผันผวนแบบไม่เป็นระบบ (Unsystematic Risk) สูงมาก ในขณะที่การนำหุ้นมารวมกลุ่มเป็นพอร์ตโฟลิโอและใช้ค่าเฉลี่ยตัวแปรเป้าหมาย (`Import_Avg` และ `Export_Avg`) จะช่วย **ถัวเฉลี่ยและตัดสัญญาณรบกวน (Noise)** เฉพาะตัวเหล่านั้นออกไป ทำให้แบบจำลองสามารถดึงเฉพาะความสัมพันธ์เชิงโครงสร้างหลักจากตัวแปรมหภาคเพื่อทำนายแนวโน้มทิศทางภาพใหญ่ได้เสถียรและแม่นยำกว่า

#### **2) บทวิเคราะห์ Feature Importance (อิทธิพลของปัจจัยมหภาค)**
* **พอร์ตนำเข้า (Import_Avg):** ได้รับแรงขับเคลื่อนเชิงลบหลักจากค่าอัตราแลกเปลี่ยน `USD_THB` (เงินบาทอ่อนค่าเป็นผลลบต่อหุ้นนำเข้าเนื่องจากทำให้ต้นทุนนำเข้าสูงขึ้น) และมีความเชื่อมโยงเชิงบวกสูงกับความชันเส้นอัตราผลตอบแทนพันธบัตรรัฐบาลไทย (`Yield_Curve_Slope`)
* **พอร์ตส่งออก (Export_Avg):** ได้รับอิทธิพลเชิงบวกสูงอย่างเด่นชัดจากค่า `USD_THB` (เงินบาทอ่อนค่าช่วยให้ขีดความสามารถการแข่งขันและรายได้จากการส่งออกเมื่อแปลงเป็นสกุลเงินบาทเติบโตสูงขึ้น) รวมถึงค่าความผันผวนของตลาดภาพรวม (`Volatility_to_Open`)
* **ราคาน้ำมันเบรนท์ (Brent):** เป็นปัจจัยหลักที่ส่งผลบวกอย่างมากต่อกลุ่มนำเข้าพลังงานและปิโตรเคมีรายตัว (เช่น `BCP`) ในแง่ของ Inventory Gain และสเปรดราคาผลิตภัณฑ์ แต่จะแสดงผลกระทบจำกัดลงเมื่อกระจายในภาพใหญ่ของพอร์ตอุตสาหกรรมอื่น

# หัวข้อ 3: การออกแบบวิศวกรรมข้อมูล (Data Engineering Design)

ในการพัฒนาแบบจำลองพยากรณ์ทิศทางราคาสินทรัพย์ทางการเงินระยะ 5 วัน (5-Day Horizon) จากปัจจัยมหภาค จำจำเป็นต้องมีระบบการรวบรวมและจัดเตรียมข้อมูลที่มีความถูกต้องและน่าเชื่อถือสูง เพื่อป้องกันปัญหา **Data Leakage** และการบิดเบือนของอนุกรมเวลา (Temporal Inconsistency) ส่วนนี้จะอธิบายถึงรายละเอียดของสถาปัตยกรรมการไหลเวียนของข้อมูล เทคโนโลยี และวิธีการจัดเตรียม Analytical Dataset อย่างเป็นระบบ

---

## 1) ภาพรวมสถาปัตยกรรมการประมวลผลข้อมูล (Data Pipeline Architecture)

การออกแบบระบบจัดระบบท่อข้อมูล (Data Pipeline) ของโครงการนี้ขับเคลื่อนด้วยโครงสร้างแบบ **Batch Processing** ที่มีการประมวลผลตามลำดับขั้นตอนเวลา (Chronological Pipeline Flow) ดังแผนภาพด้านล่าง:

```
[ Raw CSV Files ] (ใน Google Drive แยกรายหมวดหมู่)
       │
       ▼
[ 1. Ingestion & Standardization ]
   - โหลดข้อมูลจาก 7 โฟลเดอร์หลักผ่าน Glob
   - แปลงชื่อคอลัมน์ภาษาไทยให้เป็นรูปแบบมาตรฐานภาษาอังกฤษ (Date, Close, Open, High, Low, Volume, Pct_Change)
   - เปลี่ยนประเภทข้อมูล (Cast Types) เช่น เปอร์เซ็นต์ (%) ให้เป็น Float
       │
       ▼
[ 2. Data Cleaning & Null Handling ]
   - ลบคอลัมน์ที่ว่างเปล่าออก เช่น คอลัมน์ Volume ของพันธบัตร
   - ลบและจัดการค่าว่างด้วยวิธี Forward Fill และ Backward Fill เพื่อลดสัญญานรบกวน
       │
       ▼
[ 3. Feature Engineering & Aggregation ]
   - คำนวณความผันผวนระหว่างวัน (Volatility_to_Open)
   - คำนวณส่วนต่างอัตราผลตอบแทนพันธบัตร (Yield_Curve_Slope = TH_20Y - TH_10Y)
   - คำนวณพอร์ตดัชนีเฉลี่ยส่งออก (Export_Avg) และนำเข้า (Import_Avg) จากหุ้นกลุ่มละ 8 ตัว
       │
       ▼
[ 4. Temporal Alignment (Join) ]
   - เชื่อมโยงข้อมูลทุกไฟล์ตามอนุกรมเวลา (Date) อิงตามปฏิทินเวลาเปิดทำการของตลาดไทย
   - เติมค่าว่างด้วย Forward Fill และ Drop ส่วนที่ขาดหายเพื่อให้ตารางสอดคล้องกันแบบวันต่อวัน
       │
       ▼
[ 5. Split & Scaling ]
   - ทำ Temporal Train-Test Split (80:20) เพื่อหลีกเลี่ยงการเห็นข้อมูลล่วงหน้า
   - ประยุกต์ใช้ StandardScaler ลงใน Scikit-Learn Pipeline เฉพาะบนชุดข้อมูลฝึกสอน (Train Set)
       │
       ▼
[ 6. Analytical Dataset Store ]
   - บันทึกไฟล์ Analytical Dataset ลง Google Drive และ Local Workspace เพื่อใช้เทรนโมเดล
```

---

## 2) การรับเข้าข้อมูล (Data Ingestion)

ระบบรองรับการเชื่อมต่อข้อมูลดิบ (Raw Data Extraction) จากไดเรกทอรี Google Drive ในรูปแบบไฟล์ CSV หลายแหล่งข้อมูลย้อนหลังมากกว่า 2,000 วันทำการ โดยมีกระบวนการดังนี้:
* **Multi-Source Extraction:** ดึงข้อมูลจากไฟล์ประวัติย้อนหลังในโฟลเดอร์จำแนกตามประเภทสินทรัพย์และดัชนีชี้วัดหลัก ได้แก่:
  1. โฟลเดอร์หุ้นนำเข้า (10 หลักทรัพย์ดิบ)
  2. โฟลเดอร์หุ้นส่งออก (10 หลักทรัพย์ดิบ)
  3. โฟลเดอร์อัตราแลกเปลี่ยน (USD/THB)
  4. โฟลเดอร์ราคาน้ำมันโลก (WTI, Brent)
  5. โฟลเดอร์ดอกเบี้ยสหรัฐฯ และพันธบัตรไทยย้อนหลัง (US_10Y, US_30Y, TH_10Y, TH_20Y)
  6. ดอกเบี้ยอ้างอิงของธนาคารแห่งประเทศไทย (TH_RATE)
* **Automated Column Mapping:** ออกแบบฟังก์ชันตรวจสอบและตั้งชื่อตัวแปรให้เป็นมาตรฐานร่วมกันโดยอัตโนมัติ โดยตัดคำเฉพาะและปรับให้เป็นชื่อย่อสากล เช่น ปรับปรุงชื่อฟอนต์แกนคอลัมน์จาก `วันเดือนปี` เป็น `Date`, `ล่าสุด` เป็น `Close`, และ `% เปลี่ยน` เป็น `Pct_Change` เพื่อให้กระบวนการทำความสะอาดในลำดับถัดไปทำงานร่วมกันได้ทันที

---

## 3) การจัดเก็บข้อมูลดิบ (Raw Data Storage)

* **Storage Platform:** จัดเก็บข้อมูลดิบทั้งหมดแบบถาวรบน **Google Drive Storage** ภายใต้ไดเรกทอรีโครงการแบ่งแยกโฟลเดอร์ย่อยอย่างชัดเจนตามกลุ่มตัวแปร (เช่น `/DSDN/หุ้นส่งออก/`, `/DSDN/ราคาน้ำมัน/` เป็นต้น)
* **ลักษณะข้อมูลดิบ:** ข้อมูลเก็บอยู่ในรูปแบบไฟล์ `.csv` ที่มีโครงสร้างแถวและคอลัมน์เป็นไปตามรอบการดาวน์โหลดของผู้ให้บริการข้อมูลการเงินเชิงประวัติ (Historical Financial Data Providers) โดยรักษาโครงสร้างดั้งเดิม ไม่มีการแก้ไขใดๆ ก่อนเข้าสู่กระบวนการ ETL เพื่อความโปร่งใสในการตรวจสอบย้อนกลับ (Data Lineage)

---

## 4) การจัดเก็บข้อมูลที่ผ่านการประมวลผลแล้ว (Processed Data & Analytical Dataset)

การออกแบบและสร้างฐานข้อมูลเพื่อใช้ในการวิเคราะห์ขั้นสูงและฝึกสอนโมเดล (Analytical Dataset Store) มีขั้นตอนการจัดเก็บดังนี้:
* **การคัดเลือกตัวแปรและขอบเขตหลักทรัพย์ (Scope Alignment):**
  - คัดเลือกหุ้นส่งออกกลุ่มหลักจำนวน 8 หลักทรัพย์ (คัดหุ้นที่มีขนาดข้อมูลสั้นอย่าง AAI และ ITC ออก)
  - คัดเลือกหุ้นนำเข้ากลุ่มหลักจำนวน 8 หลักทรัพย์ (คัด COM7 และ BGRIM ออก)
* **การเชื่อมโยงข้อมูลข้ามมิติเวลา (Multi-variate Temporal Join):** ผนวกข้อมูลแกน X (ปัจจัยชี้วัด) และแกน Y (ผลตอบแทนรายกลุ่มและหุ้นตัวแทน) เข้าด้วยกันแบบ Inner Join ตามค่าดัชนีวันทำการ `Date` ของฝั่งอัตราแลกเปลี่ยนเงินบาท (ตลาดทุนไทย) เป็นฐาน
* **Storage Formats & Destinations:**
  - **Workspace Output:** บันทึกโครงสร้างไฟล์ลงใน Workspace ของระบบที่ปลายทาง `/content/final_pipeline_data.csv` เพื่อความรวดเร็วในการเข้าถึงใน Runtime ปัจจุบัน
  - **Enterprise Persistence:** บันทึกสำเนาหลักลงบน Google Drive `/MyDrive/Colab Notebooks/DSDN/final_pipeline_data.csv` เพื่อให้ทีมนักวิเคราะห์สามารถดึงข้อมูลที่ผ่านกระบวนการจัดเตรียมข้อมูลที่สะอาดแล้วไปใช้วิเคราะห์ซ้ำได้โดยไม่ต้องผ่านกระบวนการ ETL ใหม่อีกครั้ง
  - **คุณลักษณะตารางสุดท้าย:** ตารางประกอบด้วยคอลัมน์จำนวน 12 ตัวแปร ได้แก่ อนุกรมเวลา (Date) ปัจจัยชี้นำมหภาค 5 ตัวแปร และตัวแปรเป้าหมายทางการเงินและดัชนีเฉลี่ยอีก 6 ตัวแปร รวมขนาดข้อมูลสุทธิ 2,107 แถว

---

## 5) เครื่องมือและเทคโนโลยีหลักที่ใช้ (Technology Stack)

เพื่อให้ระบบการจัดเตรียมข้อมูลมีความยืดหยุ่น ทำงานได้รวดเร็ว และมีมาตรฐานในแวดวงวิทยาศาสตร์ข้อมูล จึงเลือกใช้เครื่องมือดังต่อไปนี้:

1. **Python (Core Language):** เป็นภาษาหลักในการประมวลผลและควบคุมกระบวนการดึงข้อมูล เขียนโค้ดกระชับ เข้าใจง่าย
2. **Pandas:** ใช้ในการทำความสะอาดข้อมูล (Data Cleaning), คำนวณคณิตศาสตร์ทางการเงินแบบเวกเตอร์ (Vectorized Financial Calculations), จัดการค่าสูญหาย (Interpolation & Imputation), และรวมตาราง (Merge & Align)
3. **NumPy:** ประมวลผลทางคณิตศาสตร์และจัดการตัวแปรสถานะแบบ Array หลายมิติอย่างมีประสิทธิภาพ
4. **Scikit-Learn (StandardScaler & Pipeline):** ใช้จัดสเกลฟีเจอร์ให้อยู่ในรูปแบบเดียวกัน (Mean=0, Std=1) ป้องกันการรั่วไหลของข้อมูลระหว่างการทดสอบ
5. **Glob & OS Libraries:** จัดการเส้นทางโฟลเดอร์ ค้นหาและดึงไฟล์ CSV จำนวนมากจากระบบจัดเก็บไฟล์ Google Drive แบบอัตโนมัติ

###เอาข้อมูลใหม่เข้า

````python
import glob
import os
import pandas as pd

# 1. กำหนดโฟลเดอร์สำหรับข้อมูลชุดใหม่
path_new_fx = '/content/drive/MyDrive/Colab Notebooks/DSDN/อัตราแลกเปลี่ยน/*.csv'
path_new_index = '/content/drive/MyDrive/Colab Notebooks/DSDN/ข้อมูลดัชนี/*.csv'

files_new_fx = glob.glob(path_new_fx)
files_new_index = glob.glob(path_new_index)

print(f"จำนวนไฟล์อัตราแลกเปลี่ยนใหม่: {len(files_new_fx)} ไฟล์")
print(f"จำนวนไฟล์ข้อมูลดัชนีใหม่: {len(files_new_index)} ไฟล์")

# 2. ฟังก์ชันเสริมสำหรับโหลดไฟล์ CSV ใหม่และตั้งชื่อตัวแปรให้อัตโนมัติ
def load_new_datasets(file_list, prefix="df_new_"):
    loaded = {}
    for path in file_list:
        filename = os.path.splitext(os.path.basename(path))[0]
        clean_name = filename.replace("ข้อมูลย้อนหลัง ", "")\
                             .replace("อัตราแลกเปลี่ยนย้อนหลัง ", "")\
                             .replace(" ", "_")
        var_name = f"{prefix}{clean_name}"

        try:
            df_temp = pd.read_csv(path)
            globals()[var_name] = df_temp
            loaded[var_name] = df_temp
            print(f"✓ โหลด {var_name} สำเร็จ ({df_temp.shape[0]} แถว x {df_temp.shape[1]} คอลัมน์)")
        except Exception as e:
            print(f"✗ ไม่สามารถโหลด {var_name} จาก {path} ได้: {e}")
    return loaded

print("\n=== เริ่มต้นดาวน์โหลดชุดข้อมูลดิบใหม่ ===\n")

print("--- [1/3] โหลดข้อมูลอัตราแลกเปลี่ยนใหม่ ---")
fx_loaded = load_new_datasets(files_new_fx, prefix="df_new_fx_")

print("\n--- [2/3] โหลดข้อมูลดัชนีใหม่ ---")
index_loaded = load_new_datasets(files_new_index, prefix="df_new_index_")
````

###แยกตัวแปรต้นตั้งต้น (Brent, US_10Y, USD_THB, Volatility_to_Open, Yield_Curve_Slope) และตัวแปรเป้าหมาย (Y) ออกจากกัน

````python
import pandas as pd

# 1. ตรวจสอบและดึงข้อมูลจาก final_pipeline_df
if 'final_pipeline_df' in globals():
    # แยกคอลัมน์ตัวแปรต้นตั้งต้น 5 ตัวแปร
    x_cols_base = ['Brent', 'US_10Y', 'USD_THB', 'Volatility_to_Open', 'Yield_Curve_Slope']
    X_base = final_pipeline_df[x_cols_base].copy()

    # แยกคอลัมน์ตัวแปรเป้าหมาย 6 ตัวแปร
    y_cols_all = ['Import_Avg', 'Export_Avg', 'BH_Pct', 'BDMS_Pct', 'EGCO_Pct', 'BCP_Pct']
    Y_all = final_pipeline_df[y_cols_all].copy()

    # เก็บข้อมูลคอลัมน์วันที่แยกไว้ต่างหากเพื่อรักษาลำดับเวลา
    dates = final_pipeline_df['Date'].copy()

    # 2. แสดงผลตรวจสอบขนาดของข้อมูล
    print("=== ตรวจสอบมิติข้อมูล (Shapes) ===")
    print(f"ขนาดของ X_base (Features): {X_base.shape}")
    print(f"ขนาดของ Y_all (Targets):  {Y_all.shape}")
    print(f"ขนาดของ dates (Temporal):  {dates.shape}\n")

    # 3. แสดงตัวอย่างข้อมูล 5 แถวแรก
    print("=== ตัวอย่างข้อมูล X_base (5 แถวแรก) ===")
    display(X_base.head(5))

    print("\n=== ตัวอย่างข้อมูล Y_all (5 แถวแรก) ===")
    display(Y_all.head(5))
else:
    print("ไม่พบตัวแปร 'final_pipeline_df' ในระบบ กรุณารันเซลล์จัดเตรียมข้อมูล (Pipeline) ก่อนหน้านี้")
````

###รวบทั้ง 3 กลุ่มฟีเจอร์ (Lagged 5 ตัว + Percentage Change/Momentum 3 ตัว + Rolling Volatility 5 วัน 2 ตัว)

````python
import pandas as pd
import numpy as np

# 1. สร้างตัวแปร X_fe โดยการคัดลอกจาก X_base
X_fe = X_base.copy()

# 2. สร้าง Lagged Features (5 ตัว)
X_fe['Brent_lag1'] = X_fe['Brent'].shift(1)
X_fe['Brent_lag2'] = X_fe['Brent'].shift(2)
X_fe['USD_THB_lag1'] = X_fe['USD_THB'].shift(1)
X_fe['USD_THB_lag2'] = X_fe['USD_THB'].shift(2)
X_fe['US_10Y_lag1'] = X_fe['US_10Y'].shift(1)

# 3. สร้าง Percentage Change / Momentum (3 ตัว) และจัดการแปลง inf เป็น NaN
X_fe['USD_THB_pct_1d'] = X_fe['USD_THB'].pct_change(1).replace([np.inf, -np.inf], np.nan)
X_fe['USD_THB_pct_5d'] = X_fe['USD_THB'].pct_change(5).replace([np.inf, -np.inf], np.nan)
X_fe['Brent_pct_5d'] = X_fe['Brent'].pct_change(5).replace([np.inf, -np.inf], np.nan)

# 4. สร้าง Rolling Volatility 5 วัน (2 ตัว)
X_fe['Volatility_roll5_std'] = X_fe['Volatility_to_Open'].rolling(window=5).std()
X_fe['USD_THB_roll5_std'] = X_fe['USD_THB'].rolling(window=5).std()

# 5. ตรวจสอบจำนวนค่าว่าง (NaN) ในแต่ละคอลัมน์ของ X_fe
print("=== ตรวจสอบค่าว่าง (NaN) ใน X_fe ===")
print(X_fe.isnull().sum())
print()

# 6. แสดงผลลัพธ์ 5 แถวแรกของ X_fe
print("=== ตัวอย่างข้อมูล X_fe (5 แถวแรก) ===")
display(X_fe.head(5))
````

###หลังจากล้างค่า NaN และจัดแนว Index ของ X_feat, Y_feat และ dates_feat

````python
import pandas as pd
import numpy as np

# 1. แปลงค่า np.inf และ -np.inf ใน X_fe ให้เป็น np.nan อีกครั้งเพื่อความปลอดภัย
X_fe_clean = X_fe.replace([np.inf, -np.inf], np.nan)

# 2. หา index ของแถวที่ไม่มีค่า NaN เลยใน X_fe_clean
valid_idx = X_fe_clean.dropna().index

# 3. กรองข้อมูลให้เหลือเฉพาะแถวที่สมบูรณ์และรีเซ็ต index ของทุกตัวแปร
X_feat = X_fe_clean.loc[valid_idx].reset_index(drop=True)
Y_feat = Y_all.loc[valid_idx].reset_index(drop=True)
dates_feat = dates.loc[valid_idx].reset_index(drop=True)

# 4. ตรวจสอบความถูกต้องและแสดงผลลัพธ์
print("=== [ผลการจัดการค่าว่างและการสร้างชุดฟีเจอร์สมบูรณ์ (X_feat)] ===\n")
print(f"จำนวนฟีเจอร์ทั้งหมดใน X_feat: {X_feat.shape[1]} คอลัมน์")
print(f"รายชื่อฟีเจอร์ทั้งหมด: {list(X_feat.columns)}\n")

print("=== ตรวจสอบมิติข้อมูล (Shapes) ===")
print(f"ขนาดของ X_feat (Features): {X_feat.shape}")
print(f"ขนาดของ Y_feat (Targets):  {Y_feat.shape}")
print(f"ขนาดของ dates_feat:         {dates_feat.shape}")
print("✓ จำนวนแถวของข้อมูลทุกตัวแปรตรงกันเรียบร้อยแล้ว" if len(X_feat) == len(Y_feat) == len(dates_feat) else "⚠ คำเตือน: จำนวนแถวของข้อมูลไม่เท่ากัน")
print()

# ตรวจสอบหาค่า NaN หรือ inf ที่อาจหลงเหลือ
remaining_nulls = X_feat.isnull().sum().sum()
remaining_infs = np.isinf(X_feat.select_dtypes(include=[np.number])).sum().sum()
print(f"จำนวนค่า NaN คงเหลือใน X_feat: {remaining_nulls} ค่า")
print(f"จำนวนค่า Infinity คงเหลือใน X_feat: {remaining_infs} ค่า\n")

# ตรวจสอบช่วงวันที่ของข้อมูล
print("=== ตรวจสอบช่วงเวลาของข้อมูล (Temporal Range) ===")
print(f"วันเริ่มต้น: {dates_feat.min().strftime('%Y-%m-%d')}")
print(f"วันสิ้นสุด: {dates_feat.max().strftime('%Y-%m-%d')}\n")

# แสดงข้อมูล 5 แถวแรกของ X_feat
print("=== ตัวอย่างข้อมูล X_feat (5 แถวแรก) ===")
display(X_feat.head(5))
````

### คำนวณ Correlation Matrix ระหว่างฟีเจอร์ทั้ง 15 ตัวใน `X_feat` (`X_feat.corr()`) แล้วพล็อต Heatmap

````python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# ตั้งค่าฟอนต์ภาษาไทยสำหรับการแสดงผล
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
sns.set_theme(style="white")
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False

# =====================================================================
# 1. พล็อต Heatmap ของฟีเจอร์ทั้ง 15 ตัวใน X_feat
# =====================================================================
corr_features = X_feat.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(
    corr_features,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    linewidths=0.5,
    cbar_kws={'label': 'Correlation Coefficient'}
)
plt.title("Correlation Heatmap of All 15 Features (X_feat)", fontproperties=thai_font, fontsize=14, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# =====================================================================
# 2. พล็อต Heatmap แสดงความสัมพันธ์ระหว่าง 15 ฟีเจอร์ กับ Targets (Y_feat)
# =====================================================================
# ผนวกข้อมูลและหา Correlation Matrix เฉพาะจุดเชื่อมโยงข้ามกลุ่ม
combined_df = pd.concat([X_feat, Y_feat], axis=1)
corr_features_vs_targets = combined_df.corr().loc[X_feat.columns, Y_feat.columns]

plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_features_vs_targets,
    annot=True,
    fmt=".2f",
    cmap="RdBu_r",
    vmin=-0.5,
    vmax=0.5,
    center=0,
    linewidths=0.5,
    cbar_kws={'label': 'Correlation Coefficient'}
)
plt.title("Correlation Heatmap: 15 Features vs Targets (Y_feat)", fontproperties=thai_font, fontsize=14, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# =====================================================================
# 3. ค้นหาและพิมพ์คู่ฟีเจอร์ที่มีความสัมพันธ์สูงมาก (|r| >= 0.85)
# =====================================================================
print("=== [รายงานคู่ฟีเจอร์ที่มีความสัมพันธ์กันสูงมาก (|r| >= 0.85)] ===\n")

# สร้าง Mask เพื่อดูเฉพาะครึ่งบนของ Matrix (เลี่ยงคู่ที่ซ้ำกันและตัวเอง)
upper_tri = corr_features.where(np.triu(np.ones(corr_features.shape), k=1).astype(bool))

high_corr_pairs = []
for col in upper_tri.columns:
    for row in upper_tri.index:
        val = upper_tri.loc[row, col]
        if not pd.isna(val) and abs(val) >= 0.85:
            high_corr_pairs.append((row, col, val))

if high_corr_pairs:
    print(f"พบคู่ฟีเจอร์ที่ซ้ำซ้อนสูงทั้งหมด {len(high_corr_pairs)} คู่ ดังนี้:")
    for idx, (f1, f2, r_val) in enumerate(high_corr_pairs, 1):
        print(f"  {idx}. {f1} <-> {f2}  (r = {r_val:+.3f})")
    print("\n💡 ข้อแนะนำ: คู่ฟีเจอร์ข้างต้นอาจส่งผลให้เกิดปัญหา Multicollinearity ควรพิจารณาเลือกใช้ตัวแปรใดตัวแปรหนึ่งในการทำนาย")
else:
    print("✓ ไม่พบคู่ฟีเจอร์ที่มีความสัมพันธ์กันเองเกิน 0.85 (ฟีเจอร์เป็นอิสระต่อกันดี)")
````

###Feature Selection & Statistical Validation (การตรวจสอบสัญญาณทางสถิติของตัวแปรต้นบน Train Set)

````python
import pandas as pd
import numpy as np
from scipy.stats import spearmanr, mannwhitneyu
from sklearn.feature_selection import mutual_info_classif
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm

# 1. ทำ Temporal Split ที่ 80% เช่นเดิมเพื่อป้องกัน Data Leakage
split_ratio = 0.8
split_idx = int(len(X_feat) * split_ratio)

X_train = X_feat.iloc[:split_idx].reset_index(drop=True)
Y_train = Y_feat.iloc[:split_idx].reset_index(drop=True)

# แปลง Target ของชุด Train เป็น Binary Classification
Y_train_class = Y_train.apply(lambda x: np.where(x > 0, 1, 0))

targets_to_analyze = ['Import_Avg', 'Export_Avg']
feature_names = X_train.columns

thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False

# 2. คำนวณค่าทางสถิติทีละ Target บน Train Set เท่านั้น
for target in targets_to_analyze:
    y_tr = Y_train_class[target]

    # คำนวณ Mutual Information
    mi_scores = mutual_info_classif(X_train, y_tr, random_state=42)

    stats_results = []
    for idx, col in enumerate(feature_names):
        feat_series = X_train[col]

        # คำนวณ Spearman Correlation
        corr_coef, _ = spearmanr(feat_series, y_tr)

        # ทดสอบทางสถิติด้วย Mann-Whitney U Test (คลาส 0 vs คลาส 1)
        group0 = feat_series[y_tr == 0]
        group1 = feat_series[y_tr == 1]

        if len(group0) > 0 and len(group1) > 0:
            u_stat, p_value = mannwhitneyu(group0, group1, alternative='two-sided')
        else:
            p_value = np.nan

        status = "Significant" if p_value < 0.05 else "Not Significant"

        stats_results.append({
            'Feature': col,
            'Spearman_Corr': corr_coef,
            'Mutual_Information': mi_scores[idx],
            'p-value (MWU)': p_value,
            'Status (alpha=0.05)': status
        })

    df_stats = pd.DataFrame(stats_results).sort_values(by='Mutual_Information', ascending=False).reset_index(drop=True)

    print(f"\n=== [ตารางสรุป Feature Selection & Statistical Validation] Target: {target} ===")
    display(df_stats)

    # 3. พล็อตกราฟเปรียบเทียบสำหรับแต่ละ Target
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

    # กราฟแท่ง Mutual Information
    sns.barplot(data=df_stats, x='Mutual_Information', y='Feature', ax=ax1, palette='viridis', hue='Feature', legend=False)
    ax1.set_title(f'Mutual Information Scores: {target}', fontproperties=thai_font, fontsize=12, fontweight='bold')
    ax1.set_xlabel('MI Score', fontproperties=thai_font, fontsize=10)
    ax1.set_ylabel('Features', fontproperties=thai_font, fontsize=10)
    for tick in ax1.get_yticklabels():
        tick.set_fontproperties(thai_font)

    # กราฟแท่ง Spearman Correlation
    df_sorted_corr = df_stats.sort_values(by='Spearman_Corr', key=abs, ascending=False)
    sns.barplot(data=df_sorted_corr, x='Spearman_Corr', y='Feature', ax=ax2, palette='coolwarm', hue='Feature', legend=False)
    ax2.set_title(f'Spearman Correlation: {target}', fontproperties=thai_font, fontsize=12, fontweight='bold')
    ax2.set_xlabel('Correlation Coefficient', fontproperties=thai_font, fontsize=10)
    ax2.set_ylabel('', fontproperties=thai_font, fontsize=10)
    ax2.axvline(0, color='black', linestyle='--', linewidth=0.8)
    for tick in ax2.get_yticklabels():
        tick.set_fontproperties(thai_font)

    plt.tight_layout()
    plt.show()
````

````python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# 1. คัดเลือกเฉพาะ 8 ฟีเจอร์ที่กำหนดจาก X_feat
selected_features = ['Brent', 'US_10Y', 'USD_THB', 'Brent_lag1', 'Brent_lag2', 'USD_THB_lag1', 'USD_THB_lag2', 'USD_THB_pct_1d']
X_selected = X_feat[selected_features].copy()

# 2. แบ่งชุดข้อมูลแบบ Temporal Split ตามลำดับเวลา (80% Train, 20% Test)
split_ratio = 0.8
split_idx = int(len(X_selected) * split_ratio)

X_train_raw = X_selected.iloc[:split_idx].reset_index(drop=True)
X_test_raw = X_selected.iloc[split_idx:].reset_index(drop=True)

dates_train = dates_feat.iloc[:split_idx].reset_index(drop=True)
dates_test = dates_feat.iloc[split_idx:].reset_index(drop=True)

# 3. แปลงตัวแปรเป้าหมายใน Y_feat เป็น Binary Classification (1 เมื่อผลตอบแทน > 0, 0 เมื่อผลตอบแทน <= 0)
Y_train_cls = Y_feat.iloc[:split_idx].apply(lambda x: np.where(x > 0, 1, 0)).reset_index(drop=True)
Y_test_cls = Y_feat.iloc[split_idx:].apply(lambda x: np.where(x > 0, 1, 0)).reset_index(drop=True)

# 4. ทำ StandardScaler แบบป้องกัน Data Leakage
scaler = StandardScaler()

# fit และ transform เฉพาะบนชุด Train
X_train_scaled_arr = scaler.fit_transform(X_train_raw)
# transform บนชุด Test ด้วยพารามิเตอร์จากชุด Train เท่านั้น
X_test_scaled_arr = scaler.transform(X_test_raw)

# แปลงกลับเป็น DataFrame
X_train_scaled = pd.DataFrame(X_train_scaled_arr, columns=selected_features)
X_test_scaled = pd.DataFrame(X_test_scaled_arr, columns=selected_features)

# 5. แสดงผลการตรวจสอบข้อมูล
print("=== 1. ตรวจสอบมิติข้อมูล (Shapes) ===")
print(f"X_train_scaled: {X_train_scaled.shape}")
print(f"X_test_scaled:  {X_test_scaled.shape}")
print(f"Y_train_cls:    {Y_train_cls.shape}")
print(f"Y_test_cls:     {Y_test_cls.shape}")
print(f"dates_train:    {dates_train.shape}")
print(f"dates_test:     {dates_test.shape}\n")

print("=== 2. ตรวจสอบช่วงเวลาข้อมูล (Temporal Coverage) ===")
print(f"Train Set: เริ่มต้น {dates_train.min().strftime('%Y-%m-%d')} ถึง สิ้นสุด {dates_train.max().strftime('%Y-%m-%d')}")
print(f"Test Set:  เริ่มต้น {dates_test.min().strftime('%Y-%m-%d')} ถึง สิ้นสุด {dates_test.max().strftime('%Y-%m-%d')}")
# ตรวจสอบการทับซ้อนของวัน
overlap = set(dates_train).intersection(set(dates_test))
print(f"จำนวนวันที่ซ้อนทับกันระหว่าง Train และ Test: {len(overlap)} วัน\n")

print("=== 3. ตรวจสอบความสมบูรณ์ของข้อมูล (NaN & Infinity Check) ===")
train_nans = X_train_scaled.isnull().sum().sum()
test_nans = X_test_scaled.isnull().sum().sum()
train_infs = np.isinf(X_train_scaled.to_numpy()).sum()
test_infs = np.isinf(X_test_scaled.to_numpy()).sum()
print(f"พบค่า NaN ใน X_train_scaled: {train_nans} | ใน X_test_scaled: {test_nans}")
print(f"พบค่า Inf ใน X_train_scaled: {train_infs} | ใน X_test_scaled: {test_infs}")
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from IPython.display import display

# 1. ตั้งค่าฟอนต์ภาษาไทยสำหรับกราฟแสดงผล
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False

# 2. นิยามโมเดลที่ต้องการฝึกสอนและประเมิน
models = {
    'Logistic Regression': LogisticRegression(random_state=42, max_iter=1000),
    'Random Forest': RandomForestClassifier(random_state=42, n_jobs=-1),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42)
}

all_results = []
best_models = {}

# 3. วนลูปเทรนและประเมินผลทีละตัวแปรเป้าหมาย (6 Targets)
for col in Y_train_cls.columns:
    y_tr = Y_train_cls[col]
    y_te = Y_test_cls[col]

    # --- A. Baseline Model (Majority Class) ---
    majority_class = y_tr.mode()[0]
    baseline_preds = np.full(shape=len(y_te), fill_value=majority_class)

    # คำนวณอัตราส่วนการทำนาย (ควรเป็น 100% สำหรับคลาสส่วนใหญ่)
    base_class_1_ratio = np.mean(baseline_preds == 1)
    base_pred_dist = f"1: {base_class_1_ratio:.1%}, 0: {1 - base_class_1_ratio:.1%}"

    all_results.append({
        'Target': col,
        'Model': 'Baseline (Majority Class)',
        'Accuracy': accuracy_score(y_te, baseline_preds),
        'Precision': precision_score(y_te, baseline_preds, zero_division=0),
        'Recall': recall_score(y_te, baseline_preds, zero_division=0),
        'F1-Score': f1_score(y_te, baseline_preds, zero_division=0),
        'Prediction Distribution': base_pred_dist
    })

    # --- B. Machine Learning Models ---
    best_acc = -1
    best_model_name = None
    best_clf_obj = None

    for model_name, clf in models.items():
        # ฝึกสอนโมเดลบนชุดข้อมูลที่ Scale แล้ว
        clf.fit(X_train_scaled, y_tr)
        preds = clf.predict(X_test_scaled)

        acc = accuracy_score(y_te, preds)
        prec = precision_score(y_te, preds, zero_division=0)
        rec = recall_score(y_te, preds, zero_division=0)
        f1 = f1_score(y_te, preds, zero_division=0)

        # คำนวณสัดส่วนการทายคลาส 1 และ คลาส 0
        class_1_ratio = np.mean(preds == 1)
        pred_dist = f"1: {class_1_ratio:.1%}, 0: {1 - class_1_ratio:.1%}"

        all_results.append({
            'Target': col,
            'Model': model_name,
            'Accuracy': acc,
            'Precision': prec,
            'Recall': rec,
            'F1-Score': f1,
            'Prediction Distribution': pred_dist
        })

        # เก็บโมเดลที่ดีที่สุดอ้างอิงจาก Accuracy บน Test Set เพื่อนำมาวิเคราะห์ Feature Importance
        if acc > best_acc:
            best_acc = acc
            best_model_name = model_name
            best_clf_obj = clf

    best_models[col] = (best_model_name, best_clf_obj)

# 4. แสดงตารางเปรียบเทียบผลลัพธ์
df_summary = pd.DataFrame(all_results)
print("=== [ตารางสรุปผลประสิทธิภาพโมเดลจำแนกตามเป้าหมายบน Test Set] ===")
for col in Y_train_cls.columns:
    print(f"\n ตัวแปรเป้าหมาย: {col}")
    df_target_sub = df_summary[df_summary['Target'] == col].sort_values(by='Accuracy', ascending=False).reset_index(drop=True)
    display(df_target_sub)

# 5. แสดงผล Feature Importance หรือ Coefficients สำหรับโมเดลที่ดีที่สุดในแต่ละ Target
print("\n=== [การวิเคราะห์อิทธิพลของตัวแปรต้น (Feature Weight / Importance Analysis)] ===")
fig, axes = plt.subplots(3, 2, figsize=(16, 14))
axes = axes.flatten()

for idx, col in enumerate(Y_train_cls.columns):
    best_name, clf = best_models[col]
    ax = axes[idx]

    if best_name == 'Logistic Regression':
        # ใช้ค่า Coefficient
        importance_vals = clf.coef_[0]
        importance_series = pd.Series(importance_vals, index=selected_features).sort_values()
        sns.barplot(x=importance_series.values, y=importance_series.index, ax=ax, palette='RdBu_r', hue=importance_series.index, legend=False)
        ax.set_title(f"{col} Best Model: {best_name}\n(Coefficient Value)", fontproperties=thai_font, fontsize=11, fontweight='bold')
        ax.axvline(0, color='black', linestyle='--', linewidth=0.8)
    else:
        # ใช้ค่า Feature Importances สำหรับ Tree-based models
        importance_vals = clf.feature_importances_
        importance_series = pd.Series(importance_vals, index=selected_features).sort_values()
        sns.barplot(x=importance_series.values, y=importance_series.index, ax=ax, palette='viridis', hue=importance_series.index, legend=False)
        ax.set_title(f"{col} Best Model: {best_name}\n(Gini Feature Importance)", fontproperties=thai_font, fontsize=11, fontweight='bold')

    ax.set_xlabel('ระดับความสำคัญ / น้ำหนักของตัวแปร', fontproperties=thai_font, fontsize=9)
    for tick in ax.get_yticklabels():
        tick.set_fontproperties(thai_font)

plt.tight_layout()
plt.show()
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm
from sklearn.linear_model import LogisticRegression
from sklearn.base import clone
from sklearn.metrics import accuracy_score, f1_score
from IPython.display import display

# 1. โหลดฟอนต์ภาษาไทยสำหรับการพล็อตกราฟ
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False
sns.set_theme(style="whitegrid")

# 2. นิยามโครงสร้างโมเดล Logistic Regression
base_lr = LogisticRegression(max_iter=1000, random_state=42)
selected_features = ['Brent', 'US_10Y', 'USD_THB', 'Brent_lag1', 'Brent_lag2', 'USD_THB_lag1', 'USD_THB_lag2', 'USD_THB_pct_1d']

results_lr = []
coef_dict = {}
preds_dict = {}

# 3. เทรนและประเมินผลรายตัวแปรเป้าหมาย
for col in Y_train_cls.columns:
    y_tr = Y_train_cls[col]
    y_te = Y_test_cls[col]

    # คำนวณ Baseline (Majority Class)
    majority_val = y_tr.mode()[0]
    base_preds = np.full(shape=len(y_te), fill_value=majority_val)
    base_acc = accuracy_score(y_te, base_preds)
    base_f1 = f1_score(y_te, base_preds, zero_division=0)

    # เทรน Logistic Regression Clone
    model_lr = clone(base_lr)
    model_lr.fit(X_train_scaled, y_tr)
    preds_lr = model_lr.predict(X_test_scaled)

    lr_acc = accuracy_score(y_te, preds_lr)
    lr_f1 = f1_score(y_te, preds_lr, zero_division=0)

    # บันทึกสัมประสิทธิ์และผลลัพธ์การพยากรณ์
    coef_dict[col] = model_lr.coef_[0]
    preds_dict[col] = preds_lr

    # สรุปอัตราส่วนคลาสที่ทำนาย
    class_1_ratio = np.mean(preds_lr == 1)

    results_lr.append({
        'Target': col,
        'Baseline_Accuracy': base_acc,
        'LR_Accuracy': lr_acc,
        'Baseline_F1': base_f1,
        'LR_F1': lr_f1,
        'Pred_Class_1_Ratio': class_1_ratio
    })

df_lr_summary = pd.DataFrame(results_lr)
print("=== [ตารางสรุปเปรียบเทียบผลลัพธ์ Logistic Regression vs Baseline] ===")
display(df_lr_summary)

# =========================================================
# กราฟชุดที่ 1: เปรียบเทียบประสิทธิภาพ (Accuracy & F1-Score)
# =========================================================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

x_indices = np.arange(len(df_lr_summary['Target']))
bar_width = 0.35

# พล็อต Accuracy
ax1.bar(x_indices - bar_width/2, df_lr_summary['Baseline_Accuracy'], width=bar_width, label='Baseline', color='#7f7f7f', alpha=0.7)
ax1.bar(x_indices + bar_width/2, df_lr_summary['LR_Accuracy'], width=bar_width, label='Logistic Regression', color='#1f77b4')
ax1.set_title('เปรียบเทียบความแม่นยำ (Accuracy) ของแต่ละเป้าหมาย', fontproperties=thai_font, fontsize=14, fontweight='bold')
ax1.set_xticks(x_indices)
ax1.set_xticklabels(df_lr_summary['Target'], rotation=30)
ax1.set_ylabel('Accuracy Score', fontproperties=thai_font, fontsize=11)
ax1.legend(prop=thai_font)

# พล็อต F1-Score
ax2.bar(x_indices - bar_width/2, df_lr_summary['Baseline_F1'], width=bar_width, label='Baseline', color='#7f7f7f', alpha=0.7)
ax2.bar(x_indices + bar_width/2, df_lr_summary['LR_F1'], width=bar_width, label='Logistic Regression', color='#2ca02c')
ax2.set_title('เปรียบเทียบค่า F1-Score ของแต่ละเป้าหมาย', fontproperties=thai_font, fontsize=14, fontweight='bold')
ax2.set_xticks(x_indices)
ax2.set_xticklabels(df_lr_summary['Target'], rotation=30)
ax2.set_ylabel('F1-Score', fontproperties=thai_font, fontsize=11)
ax2.legend(prop=thai_font)

plt.tight_layout()
plt.show()

# =========================================================
# กราฟชุดที่ 2: น้ำหนักตัวแปร Coefficients Bar Chart แยกตาม 6 เป้าหมาย
# =========================================================
df_coefs = pd.DataFrame(coef_dict, index=selected_features)

fig2, axes2 = plt.subplots(3, 2, figsize=(16, 14), sharey=True)
axes2 = axes2.flatten()

for idx, col in enumerate(Y_train_cls.columns):
    importance = df_coefs[col].sort_values()
    sns.barplot(x=importance.values, y=importance.index, ax=axes2[idx], palette='RdBu_r', hue=importance.index, legend=False)
    axes2[idx].set_title(f'สัมประสิทธิ์ (Coefficients) สำหรับเป้าหมาย: {col}', fontproperties=thai_font, fontsize=12, fontweight='bold')
    axes2[idx].axvline(0, color='black', linestyle='--', linewidth=0.8)
    axes2[idx].set_xlabel('Coefficient Value', fontproperties=thai_font, fontsize=10)
    for tick in axes2[idx].get_yticklabels():
        tick.set_fontproperties(thai_font)

fig2.suptitle('น้ำหนักและทิศทางของตัวแปรต้นต่อเป้าหมายพยากรณ์แต่ละตัว', fontproperties=thai_font, fontsize=16, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

# =========================================================
# กราฟชุดที่ 3: เปรียบเทียบทิศทางพยากรณ์ในช่วง dates_test (60 วันล่าสุด)
# =========================================================
plot_days = 60
test_dates_dt = pd.to_datetime(dates_test)

fig3, axes3 = plt.subplots(6, 1, figsize=(15, 18), sharex=True)

for idx, col in enumerate(Y_train_cls.columns):
    y_act = Y_test_cls[col].values[-plot_days:]
    y_prd = preds_dict[col][-plot_days:]
    dates_sub = test_dates_dt.values[-plot_days:]

    axes3[idx].step(dates_sub, y_act, where='mid', label='Actual Direction', color='#1f77b4', alpha=0.8, linewidth=2)
    axes3[idx].step(dates_sub, y_prd, where='mid', label='Predicted Direction', color='#d62728', linestyle='--', linewidth=1.8)
    axes3[idx].set_title(f'อนุกรมเวลาเปรียบเทียบทิศทางจริง vs คำทำนาย: {col} ({plot_days} วันล่าสุด)', fontproperties=thai_font, fontsize=11, fontweight='bold')
    axes3[idx].set_yticks([0, 1])
    axes3[idx].set_yticklabels(['ลง/คงที่ (0)', 'ขึ้น (1)'], fontproperties=thai_font, fontsize=9)
    axes3[idx].legend(prop=thai_font, loc='upper right')
    axes3[idx].grid(True, linestyle=':', alpha=0.5)

plt.xticks(rotation=45)
fig3.suptitle('เปรียบเทียบทิศทางการขึ้นลงจริงเทียบกับคำทำนายของ Logistic Regression ในชุดข้อมูลทดสอบ', fontproperties=thai_font, fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
````

````python
print("=== [ขั้นตอนที่ 2] ตรวจสอบค่าว่างและทำความสะอาดสำหรับข้อมูลดัชนี ===\n")

for name, df in sorted(new_index_dfs.items()):
    print(f"📌 ดำเนินการตรวจสอบและทำความสะอาด: {name}")
    # ตรวจสอบว่ามีค่าว่างหรือไม่
    check_and_show_nulls(df)

    # ดรอปคอลัมน์ ปริมาณ (Volume) ออกหากมีค่าเป็น Null ทั้งหมดหรือเกือบทั้งหมด เพื่อป้องกันข้อผิดพลาด
    if 'ปริมาณ' in df.columns and df['ปริมาณ'].isnull().all():
        df_cleaned = drop_columns(df, ['ปริมาณ'])
    elif 'ปริมาณ' in df.columns and df['ปริมาณ'].isnull().sum() > 0:
        # สำหรับดัชนีที่คอลัมน์ปริมาณเป็น NaN ในบางวัน แต่เราจำเป็นต้องใช้ข้อมูลราคาปิด แนะนำให้ดรอปคอลัมน์นี้
        df_cleaned = drop_columns(df, ['ปริมาณ'])
    else:
        df_cleaned = df.copy()

    # ลบแถวที่มีค่าว่างอื่น ๆ เพื่อให้ข้อมูลสมบูรณ์
    df_final = drop_null_rows(df_cleaned)

    # บันทึกค่าที่ทำความสะอาดเสร็จกลับไปยัง Global variable
    globals()[name] = df_final
    print(f"✓ {name} ทำความสะอาดสำเร็จ คงเหลือขนาด: {df_final.shape[0]} แถว x {df_final.shape[1]} คอลัมน์")
    print("-" * 60)
````

###สร้าง df ข้อมูลดัชนีใหม่

````python
import pandas as pd
from IPython.display import display

# 1. กำหนดแมปปิ้งชื่อตัวแปรดัชนีภาษาไทยเป็นภาษาอังกฤษมาตรฐานสำหรับชุดข้อมูลใหม่
index_variable_mapping = {
    'df_new_index_ข้อมูลย้อนหลังของ_ดัชนี_SET_50': 'df_new_SET50',
    'df_new_index_ข้อมูลย้อนหลังของ_ดัชนี_US_Dollar_Index': 'df_new_DXY',
    'df_new_index_ข้อมูลย้อนหลังของ_ดัชนี_S&P_500': 'df_new_SP500',
    'df_new_index_ข้อมูลย้อนหลังของ_ดัชนี_CBOE_Volatility_Index': 'df_new_VIX',
    'df_new_index_ข้อมูลย้อนหลังของ_ดัชนี_US_Dollar_Index_(1)': 'df_new_DXY_alt',
    'df_new_index_ข้อมูลย้อนหลังของ_ดัชนี_Baltic_Dry_Index': 'df_new_BDI',
    'df_new_index_ข้อมูลย้อนหลังของ_ดัชนี_Philadelphia_Semiconductor_Index': 'df_new_SOX'
}

print("=== เริ่มสร้างตัวแปรมาตรฐานภาษาอังกฤษสำหรับข้อมูลดัชนีชุดใหม่ ===\n")

for thai_var, eng_var in index_variable_mapping.items():
    df = globals().get(thai_var)
    if df is not None:
        # จัดการแปลงชื่อคอลัมน์ให้อยู่ในมาตรฐานเดียวกัน (Date, Close, Open, High, Low, Pct_Change)
        column_mapping = {
            'วันเดือนปี': 'Date',
            'ล่าสุด': 'Close',
            'ราคาเปิด': 'Open',
            'สูงสุด': 'High',
            'ต่ำสุด': 'Low',
            '% เปลี่ยน': 'Pct_Change'
        }
        df_renamed = df.rename(columns=column_mapping)

        # จัดเรียงลำดับคอลัมน์ให้เป็นระเบียบ
        cols_order = ['Date', 'Close', 'Open', 'High', 'Low', 'Pct_Change']
        existing_cols = [c for c in cols_order if c in df_renamed.columns]
        df_renamed = df_renamed[existing_cols]

        # ประกาศตัวแปร Global ใหม่
        globals()[eng_var] = df_renamed
        print(f"✓ สร้างตัวแปร {eng_var} สำเร็จ! (ขนาด: {df_renamed.shape[0]} แถว x {df_renamed.shape[1]} คอลัมน์)")
    else:
        print(f"⚠ ไม่พบตัวแปร {thai_var} ในระบบ")
````

````python
# 2. แสดงตัวอย่างข้อมูลดัชนีชุดใหม่ที่ถูกจัดมาตรฐานเรียบร้อยแล้ว
print("=== ตัวอย่างข้อมูลดัชนีใหม่บางส่วน ===")
for eng_var in ['df_new_SET50', 'df_new_SP500', 'df_new_VIX', 'df_new_BDI', 'df_new_SOX']:
    df = globals().get(eng_var)
    if df is not None:
        print(f"\n• ตัวอย่างข้อมูลของ {eng_var}: ")
        display(df.head(3))
````

#หาความสัมพันธ์

````python
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from IPython.display import display

# 1. ตั้งค่าฟอนต์ภาษาไทยสำหรับการแสดงผลบนกราฟ
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False
sns.set_theme(style="white")

# ฟังก์ชันดึงและทำความสะอาด Pct_Change หรือคำนวณ % Change รายวัน
def get_daily_return(df, name_col):
    df_clean = df.copy()

    # ตรวจสอบและแปลงคอลัมน์วันที่
    date_col = 'Date' if 'Date' in df_clean.columns else ('วันเดือนปี' if 'วันเดือนปี' in df_clean.columns else df_clean.columns[0])
    df_clean['Date_Parsed'] = pd.to_datetime(df_clean[date_col])
    df_clean = df_clean.sort_values('Date_Parsed')

    pct_col = 'Pct_Change' if 'Pct_Change' in df_clean.columns else ('% เปลี่ยน' if '% เปลี่ยน' in df_clean.columns else None)

    if pct_col and pct_col in df_clean.columns:
        if df_clean[pct_col].dtype == 'object':
            df_clean['Return'] = df_clean[pct_col].str.rstrip('%').astype('float') / 100.0
        else:
            df_clean['Return'] = pd.to_numeric(df_clean[pct_col], errors='coerce')
    else:
        close_col = 'Close' if 'Close' in df_clean.columns else ('ล่าสุด' if 'ล่าสุด' in df_clean.columns else df_clean.columns[1])
        df_clean['Return'] = pd.to_numeric(df_clean[close_col], errors='coerce').pct_change()

    return df_clean[['Date_Parsed', 'Return']].rename(columns={'Date_Parsed': 'Date', 'Return': name_col}).dropna()

# 2. ตรวจสอบและโหลด final_pipeline_df จากไฟล์หากไม่มีใน Memory
if 'final_pipeline_df' not in globals():
    local_path = '/content/final_pipeline_data.csv'
    drive_path = '/content/drive/MyDrive/Colab Notebooks/DSDN/final_pipeline_data.csv'

    if os.path.exists(local_path):
        print(f"✓ ไม่พบตัวแปรในหน่วยความจำ แต่ทำการโหลดข้อมูลสำเร็จจาก Local: {local_path}")
        final_pipeline_df = pd.read_csv(local_path)
    elif os.path.exists(drive_path):
        print(f"✓ ไม่พบตัวแปรในหน่วยความจำ แต่ทำการโหลดข้อมูลสำเร็จจาก Drive: {drive_path}")
        final_pipeline_df = pd.read_csv(drive_path)
    else:
        raise FileNotFoundError("ไม่พบไฟล์ข้อมูล final_pipeline_data.csv กรุณาย้อนกลับไปรันขั้นตอนจัดเตรียม Pipeline")

# 3. รวบรวมข้อมูลปัจจัยรายวันชุดใหม่
daily_factors = {
    'DXY': df_new_DXY,
    'SET50': df_new_SET50,
    'SP500': df_new_SP500,
    'VIX': df_new_VIX,
    'BDI': df_new_BDI,
    'SOX': df_new_SOX,
    'EUR_THB': df_new_fx_EUR_THB,
    'JPY_THB': df_new_fx_JPY_THB,
    'CNY_THB': df_new_fx_CNY_THB,
    'USD_THB': df_new_fx_USD_THB
}

# ทำการแปลงข้อมูลปัจจัยนำเข้าเป็น Daily Return
df_factors_merged = None
for name, df_fac in daily_factors.items():
    if df_fac is not None:
        df_ret = get_daily_return(df_fac, name)
        if df_factors_merged is None:
            df_factors_merged = df_ret
        else:
            df_factors_merged = pd.merge(df_factors_merged, df_ret, on='Date', how='inner')

# 4. เตรียมตัวแปรเป้าหมาย (Y) จาก final_pipeline_df
# ตรวจสอบชื่อคอลัมน์จริงใน DataFrame ที่บันทึกไว้
target_cols_y = [col for col in ['Import_Avg', 'Export_Avg', 'BH_Pct', 'BDMS_Pct', 'EGCO_Pct', 'BCP_Pct'] if col in final_pipeline_df.columns]

df_targets = final_pipeline_df[['Date'] + target_cols_y].copy()
df_targets['Date'] = pd.to_datetime(df_targets['Date'])

# 5. รวมข้อมูลทั้งหมดเข้าด้วยกัน
df_daily_analysis = pd.merge(df_factors_merged, df_targets, on='Date', how='inner')

# 6. คำนวณ Correlation Matrix
daily_corr = df_daily_analysis.drop(columns=['Date']).corr()
sub_daily_corr = daily_corr.loc[list(daily_factors.keys()), target_cols_y]

# 7. พล็อต Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(
    sub_daily_corr,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    center=0,
    fmt=".3f",
    linewidths=0.5,
    cbar_kws={'label': 'Correlation Coefficient'}
)
plt.title("ความสัมพันธ์รายวัน (Daily % Change Correlation)\nระหว่างปัจจัยนำเข้าใหม่ กับ ตัวแปรเป้าหมาย Y", fontproperties=thai_font, fontsize=14, fontweight='bold', pad=15)
plt.xticks(fontproperties=thai_font, fontsize=11, rotation=45, ha='right')
plt.yticks(fontproperties=thai_font, fontsize=11, rotation=0)
plt.tight_layout()
plt.show()

print("=== [ตารางสรุปผล] ค่าสหสัมพันธ์รายวันระหว่าง ปัจจัยนำเข้าใหม่ กับ ตัวแปรเป้าหมาย Y ===")
display(sub_daily_corr)
````

###เช็กขนาดข้อมูลหลังรวม (shape และช่วงวันที่)

````python
import pandas as pd
import numpy as np

# 1. มั่นใจว่าข้อมูลวันที่ของทั้งสอง DataFrame อยู่ในรูปแบบ Datetime
df_daily_analysis['Date'] = pd.to_datetime(df_daily_analysis['Date'])
final_pipeline_df['Date'] = pd.to_datetime(final_pipeline_df['Date'])

# 2. ตรวจสอบรายละเอียดเบื้องต้นของ df_daily_analysis
shape_daily = df_daily_analysis.shape
min_date_daily = df_daily_analysis['Date'].min()
max_date_daily = df_daily_analysis['Date'].max()
null_count_daily = df_daily_analysis.isnull().sum().sum()
duplicated_dates_daily = df_daily_analysis['Date'].duplicated().sum()

# 3. ตรวจสอบรายละเอียดเบื้องต้นของ final_pipeline_df
shape_pipe = final_pipeline_df.shape
min_date_pipe = final_pipeline_df['Date'].min()
max_date_pipe = final_pipeline_df['Date'].max()

# 4. วิเคราะห์เปรียบเทียบวันทำการที่หายไปจากการ Aligned (Inner Merge)
dates_pipe_set = set(final_pipeline_df['Date'])
dates_daily_set = set(df_daily_analysis['Date'])

lost_dates = dates_pipe_set - dates_daily_set
num_lost_days = len(lost_dates)
pct_lost_days = (num_lost_days / len(dates_pipe_set)) * 100

# 5. แสดงรายงานการวิเคราะห์เชิงลึก
print("=== [รายงานผลการวิเคราะห์ข้อมูลหลังทำ Temporal Inner Alignment] ===\n")
print(f"📊 ข้อมูล DataFrame: df_daily_analysis")
print(f"  - ขนาดของข้อมูล (Shape): {shape_daily[0]} แถว x {shape_daily[1]} คอลัมน์")
print(f"  - ช่วงเวลา: ตั้งแต่ {min_date_daily.strftime('%Y-%m-%d')} ถึง {max_date_daily.strftime('%Y-%m-%d')}")
print(f"  - จำนวนแถวที่มีวันที่ซ้ำกัน: {duplicated_dates_daily} แถว")
print(f"  - จำนวนค่าว่าง (NaN) ทั้งหมด: {null_count_daily} ตำแหน่ง\n")

print(f"📊 ข้อมูล DataFrame ดั้งเดิม: final_pipeline_df")
print(f"  - ขนาดของข้อมูล (Shape): {shape_pipe[0]} แถว x {shape_pipe[1]} คอลัมน์")
print(f"  - ช่วงเวลา: ตั้งแต่ {min_date_pipe.strftime('%Y-%m-%d')} ถึง {max_date_pipe.strftime('%Y-%m-%d')}\n")

print("=== [ผลการเปรียบเทียบและการสูญหายของข้อมูล] ===")
print(f"  - จำนวนวันทำการที่สูญหายหลังจากการ Merge: {num_lost_days} วัน")
print(f"  - คิดเป็นอัตราการสูญหาย: {pct_lost_days:.2f}%")
if num_lost_days > 0:
    print(f"  - ตัวอย่าง 5 วันทำการของฝั่งหุ้นที่ถูกตัดออกไป:")
    for d in sorted(list(lost_dates))[:5]:
        print(f"    • {d.strftime('%Y-%m-%d')}")
````

###Feature Engineering จากตัวแปรชุดใหม่ (SET50, SP500, SOX, VIX, DXY และคู่เงิน CNY_THB, EUR_THB, JPY_THB, USD_THB)

###รวมฟีเจอร์ใหม่ (X_features_final) เข้ากับชุดฟีเจอร์เดิมของโมเดล

````python
import numpy as np
import pandas as pd


def ensure_date_column(df, name="df"):
  d = df.copy()
  if "Date" in d.columns:
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    return d.dropna(subset=["Date"])
  if isinstance(d.index, pd.DatetimeIndex) or d.index.name in (
      "Date",
      "date",
      "วันเดือนปี",
  ):
    d = d.reset_index()
    idx_col = d.columns[0]
    d = d.rename(columns={idx_col: "Date"})
    d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
    return d.dropna(subset=["Date"])
  for cand in ["date", "วันเดือนปี", "Date_Parsed", "observation_date"]:
    if cand in d.columns:
      d = d.rename(columns={cand: "Date"})
      d["Date"] = pd.to_datetime(d["Date"], errors="coerce")
      return d.dropna(subset=["Date"])
  raise KeyError(
      f"ไม่พบคอลัมน์หรือ Index วันที่ใน {name} (คอลัมน์ที่มี: {list(d.columns)})"
  )


def calculate_rsi(series, period=14):
  delta = series.diff()
  gain = delta.clip(lower=0)
  loss = (-delta).clip(lower=0)

  avg_gain = gain.rolling(window=period, min_periods=period).mean()
  avg_loss = loss.rolling(window=period, min_periods=period).mean()

  start_idx = avg_gain.first_valid_index()
  if start_idx is not None:
    pos = series.index.get_loc(start_idx)
    g_vals = avg_gain.to_numpy(dtype=float, copy=True)
    l_vals = avg_loss.to_numpy(dtype=float, copy=True)
    gain_arr = gain.to_numpy(dtype=float)
    loss_arr = loss.to_numpy(dtype=float)
    for i in range(pos + 1, len(series)):
      g_vals[i] = (g_vals[i - 1] * (period - 1) + gain_arr[i]) / period
      l_vals[i] = (l_vals[i - 1] * (period - 1) + loss_arr[i]) / period
    avg_gain = pd.Series(g_vals, index=series.index)
    avg_loss = pd.Series(l_vals, index=series.index)

  rs = avg_gain / avg_loss.replace(0, np.nan)
  return 100 - (100 / (1 + rs))


# 1. เตรียมฐานข้อมูลหลักและเรียงลำดับเวลา
df_fe = ensure_date_column(df_daily_analysis, "df_daily_analysis").sort_values(
    "Date"
)
df_fe = df_fe.drop_duplicates(subset=["Date"]).reset_index(drop=True)

index_close_mapping = {
    "SET50": df_new_SET50,
    "SP500": df_new_SP500,
    "SOX": df_new_SOX,
    "VIX": df_new_VIX,
    "DXY": df_new_DXY,
    "CNY_THB": df_new_fx_CNY_THB,
    "JPY_THB": df_new_fx_JPY_THB,
    "USD_THB": df_new_fx_USD_THB,
}

for name, raw_df in index_close_mapping.items():
  if raw_df is None:
    print(f"⚠ ข้าม {name} เพราะไม่พบข้อมูลในหน่วยความจำ")
    continue
  raw_clean = ensure_date_column(raw_df, name)
  close_col = (
      "Close"
      if "Close" in raw_clean.columns
      else ("ล่าสุด" if "ล่าสุด" in raw_clean.columns else None)
  )
  if close_col is None:
    print(f"⚠ ข้าม {name} เพราะไม่พบคอลัมน์ Close / ล่าสุด")
    continue
  raw_clean[f"{name}_Close"] = pd.to_numeric(
      raw_clean[close_col].astype(str).str.replace(",", "", regex=False),
      errors="coerce",
  )
  raw_clean = (
      raw_clean[["Date", f"{name}_Close"]]
      .dropna(subset=[f"{name}_Close"])
      .drop_duplicates(subset=["Date"])
      .sort_values("Date")
  )
  df_fe = pd.merge(df_fe, raw_clean, on="Date", how="left")

df_fe = df_fe.sort_values("Date").reset_index(drop=True)
close_cols = [c for c in df_fe.columns if c.endswith("_Close")]
df_fe[close_cols] = df_fe[close_cols].ffill().bfill()

# 2. คำนวณฟีเจอร์ใหม่ (เฉพาะข้อมูล ณ เวลา <= t)
X_new_features = pd.DataFrame({"Date": df_fe["Date"]})

for asset in ["SET50", "SOX", "SP500"]:
  col = f"{asset}_Close"
  if col in df_fe.columns:
    X_new_features[f"{asset}_Return_3d"] = df_fe[col].pct_change(3)
    X_new_features[f"{asset}_Return_5d"] = df_fe[col].pct_change(5)

if "SET50_Close" in df_fe.columns:
  set50_close = df_fe["SET50_Close"]
  sma20 = set50_close.rolling(window=20).mean()
  X_new_features["SET50_Distance_SMA20"] = (set50_close - sma20) / sma20
  X_new_features["SET50_RSI_14"] = calculate_rsi(set50_close, period=14)
  X_new_features["SET50_Volatility_Roll5"] = set50_close.rolling(window=5).std()

lag_assets = [
    "SP500_Close",
    "SOX_Close",
    "VIX_Close",
    "DXY_Close",
    "USD_THB_Close",
    "CNY_THB_Close",
    "JPY_THB_Close",
]
for asset in lag_assets:
  if asset in df_fe.columns:
    clean_name = asset.replace("_Close", "")
    X_new_features[f"{clean_name}_Lag1"] = df_fe[asset].shift(1)
    X_new_features[f"{clean_name}_Lag2"] = df_fe[asset].shift(2)

# 3. ล้างค่า inf / NaN และจัดแนวกับ Target
X_new_features = X_new_features.replace([np.inf, -np.inf], np.nan)
valid_rows = X_new_features.dropna().index

X_features_final = X_new_features.loc[valid_rows].reset_index(drop=True)
avail_targets = [
    c
    for c in [
        "Date",
        "Import_Avg",
        "Export_Avg",
        "BH_Pct",
        "BDMS_Pct",
        "EGCO_Pct",
        "BCP_Pct",
    ]
    if c in df_fe.columns
]
Y_targets_final = df_fe.loc[valid_rows, avail_targets].reset_index(drop=True)

print(
    f"✓ สร้างฟีเจอร์ใหม่สำเร็จ: {X_features_final.shape[1] - 1} คอลัมน์ | "
    f"แถวที่พร้อมใช้: {len(X_features_final)} แถว"
)
````

###รวมฟีเจอร์ใหม่ `X_features_final` เข้ากับชุดฟีเจอร์

````python
import pandas as pd
import numpy as np
from IPython.display import display

# 1. ตรวจสอบข้อมูลตั้งต้น
print("=== ตรวจสอบข้อมูลดิบก่อนการรวม (Pre-Merge Checking) ===")
print(f"ขนาด final_pipeline_df: {final_pipeline_df.shape}")

final_pipeline_df['Date'] = pd.to_datetime(final_pipeline_df['Date'])

# 2. ดึงฟีเจอร์ตั้งต้นจาก final_pipeline_df
original_features_cols = ['Date', 'Brent', 'US_10Y', 'USD_THB', 'Volatility_to_Open', 'Yield_Curve_Slope']
df_orig_feats = final_pipeline_df[original_features_cols].copy()

# ถ้ายืนยันว่ามี X_features_final ที่ไม่ว่างและมีคอลัมน์ 'Date' ค่อยรวมเพิ่ม
if 'X_features_final' in globals() and not X_features_final.empty and 'Date' in X_features_final.columns:
    X_features_final['Date'] = pd.to_datetime(X_features_final['Date'])
    df_merged_features = pd.merge(df_orig_feats, X_features_final, on='Date', how='inner')
else:
    print("ℹ ใช้ฟีเจอร์จาก final_pipeline_df เป็นหลัก (ข้าม X_features_final)")
    df_merged_features = df_orig_feats

# 3. จัดแนวเป้าหมายสำหรับการทำนายล่วงหน้า 5 วัน (5-Day Horizon Target Selection)
original_targets_cols = ['Import_Avg', 'Export_Avg', 'BH_Pct', 'BDMS_Pct', 'EGCO_Pct', 'BCP_Pct']
df_orig_targets = final_pipeline_df[['Date'] + original_targets_cols].copy()

df_targets_5d = df_orig_targets[original_targets_cols].rolling(window=5).sum().shift(-5)
df_targets_5d['Date'] = df_orig_targets['Date']

# 4. รวมฟีเจอร์ (X) กับเป้าหมาย 5 วันข้างหน้า (Y)
df_final_dataset = pd.merge(df_merged_features, df_targets_5d, on='Date', how='inner')
df_final_dataset = df_final_dataset.sort_values(by='Date').reset_index(drop=True)
df_final_dataset = df_final_dataset.replace([np.inf, -np.inf], np.nan).dropna().reset_index(drop=True)

if df_final_dataset.empty:
    raise ValueError("df_final_dataset ว่างเปล่าหลัง merge หรือ dropna กรุณาตรวจสอบข้อมูลต้นฉบับ")

# 5. แยก X และ Y ออกจากกัน
exclude_from_x = ['Date'] + original_targets_cols
x_features_list = [col for col in df_final_dataset.columns if col not in exclude_from_x]

X_model = df_final_dataset[x_features_list].copy()
Y_model_raw = df_final_dataset[original_targets_cols].copy()
dates_model = df_final_dataset['Date'].copy()

# แปลง Target เป็น Binary Classification (1: ขึ้น (> 0), 0: ลง/คงที่ (<= 0))
Y_model_class = Y_model_raw.apply(lambda x: np.where(x > 0, 1, 0))

print("\n=== [สรุปโครงสร้างข้อมูลสุดท้ายสำหรับขั้นตอนถัดไป] ===")
print(f"จำนวนวันทำงานทั้งหมด (Rows): {df_final_dataset.shape[0]} วัน")
print(f"จำนวนฟีเจอร์ฝั่ง X ทั้งสิ้น: {X_model.shape[1]} คอลัมน์")
print(f"ช่วงข้อมูลวันที่: {dates_model.min().strftime('%Y-%m-%d')} ถึง {dates_model.max().strftime('%Y-%m-%d')}")
print(f"ขนาดของ X_model: {X_model.shape} | ขนาดของ Y_model_class: {Y_model_class.shape}")

print("\nรายชื่อฟีเจอร์ทั้งหมดที่พร้อมนำไปเทรนโมเดล:")
print(list(X_model.columns))

display(df_final_dataset.head(5))
````

### 1. การทำ Time-Based Train-Test Split (80/20) และการตรวจสอบความสมบูรณ์ของชุดข้อมูล
ในขั้นตอนนี้เราจะทำการแบ่งชุดข้อมูลระหว่าง `X_model` และ `Y_model_class` ออกเป็นชุด Train (80%) และ Test (20%) โดยเรียงลำดับตามอนุกรมเวลาอย่างเคร่งครัด เพื่อป้องกันการเกิดปัญหา Data Leakage

````python
import pandas as pd
import numpy as np

# 1. ตรวจสอบความพร้อมและจัดเรียงลำดับเวลาให้มั่นใจ 100%
if 'X_model' in globals() and 'Y_model_class' in globals() and 'dates_model' in globals():
    # จัดทำ DataFrame ชั่วคราวเพื่อทำการจัดเรียงตามวันที่อย่างสมบูรณ์
    df_split_prep = pd.concat([dates_model, X_model, Y_model_class], axis=1)
    df_split_prep['Date'] = pd.to_datetime(df_split_prep['Date'])
    df_split_prep = df_split_prep.sort_values(by='Date').reset_index(drop=True)

    # แยกส่วนประกอบหลังจัดเรียงเรียบร้อย
    dates_sorted = df_split_prep['Date']
    X_sorted = df_split_prep[X_model.columns]
    Y_sorted = df_split_prep[Y_model_class.columns]

    # 2. คำนวณจุดตัดแบ่งข้อมูล (Temporal Cutoff) ที่สัดส่วน 80%
    split_ratio = 0.8
    split_idx = int(len(df_split_prep) * split_ratio)

    # แบ่งข้อมูลแบบลำดับเวลา (No Shuffle)
    X_train, X_test = X_sorted.iloc[:split_idx].reset_index(drop=True), X_sorted.iloc[split_idx:].reset_index(drop=True)
    Y_train, Y_test = Y_sorted.iloc[:split_idx].reset_index(drop=True), Y_sorted.iloc[split_idx:].reset_index(drop=True)
    train_dates, test_dates = dates_sorted.iloc[:split_idx].reset_index(drop=True), dates_sorted.iloc[split_idx:].reset_index(drop=True)

    # 3. สรุปผลและตรวจสอบการทับซ้อนของมิติเวลา
    print("=== [สรุปโครงสร้างข้อมูลหลังการแบ่ง Time-Based Split] ===")
    print(f"จำนวนวันทำงานรวมทั้งหมด: {len(df_split_prep)} วัน")
    print(f"ชุดฝึกสอน (Train Set): {len(X_train)} วัน ({split_ratio*100:.1f}%)")
    print(f"  - ตั้งแต่ {train_dates.min().strftime('%Y-%m-%d')} ถึง {train_dates.max().strftime('%Y-%m-%d')}")
    print(f"ชุดทดสอบ (Test Set)  : {len(X_test)} วัน ({(1-split_ratio)*100:.1f}%)")
    print(f"  - ตั้งแต่ {test_dates.min().strftime('%Y-%m-%d')} ถึง {test_dates.max().strftime('%Y-%m-%d')}")

    # เช็กวันทับซ้อน
    overlap = set(train_dates).intersection(set(test_dates))
    print(f"✓ ไม่พบการทับซ้อนของข้อมูลชุด Train และ Test (Overlap = {len(overlap)} วัน)" if len(overlap) == 0 else f"⚠ ตรวจพบข้อมูลทับซ้อนจำนวน: {len(overlap)} วัน")
    print(f"✓ การเรียงลำดับเวลาถูกต้อง (Max Train Date < Min Test Date): {train_dates.max() < test_dates.min()}")

    # 4. ตรวจสอบค่าว่างและอินฟินิตี้ (NaN & Inf Check)
    nans_in_train = X_train.isna().sum().sum()
    nans_in_test = X_test.isna().sum().sum()
    infs_in_train = np.isinf(X_train.select_dtypes(include=[np.number])).sum().sum()
    infs_in_test = np.isinf(X_test.select_dtypes(include=[np.number])).sum().sum()
    print(f"✓ ตรวจสอบค่าว่าง (NaN): Train={nans_in_train} ค่า | Test={nans_in_test} ค่า")
    print(f"✓ ตรวจสอบค่าอนันต์ (Inf): Train={infs_in_train} ค่า | Test={infs_in_test} ค่า\n")

    # 5. สรุปสัดส่วนคลาสของเป้าหมาย Y ทั้งหมด
    print("=== [ตารางสัดส่วนคลาสของตัวแปรเป้าหมาย (Class Distribution)] ===")
    class_ratios = []
    for col in Y_train.columns:
        train_dist = Y_train[col].value_counts(normalize=True).to_dict()
        test_dist = Y_test[col].value_counts(normalize=True).to_dict()

        class_ratios.append({
            'Target Variable': col,
            'Train Class 1 (Up)': f"{train_dist.get(1, 0.0):.2%}",
            'Train Class 0 (Down/Flat)': f"{train_dist.get(0, 0.0):.2%}",
            'Test Class 1 (Up)': f"{test_dist.get(1, 0.0):.2%}",
            'Test Class 0 (Down/Flat)': f"{test_dist.get(0, 0.0):.2%}"
        })
    display(pd.DataFrame(class_ratios))
else:
    print("ไม่พบตัวแปร X_model หรือ Y_model_class ในระบบ กรุณารันเซลล์รวบรวมฟีเจอร์ก่อนหน้า")
````

### 2. โมเดลการจำแนกคลาสและการเปรียบเทียบประสิทธิภาพบนชุดข้อมูลทดสอบ (Model Training & Baseline Evaluation)
เราจะทำการเทรนโมเดลหลายประเภทเพื่อเปรียบเทียบหาโมเดลที่เหมาะสมที่สุดในการทำนายทิศทางล่วงหน้า 5 วันของดัชนีทั้ง 6 เป้าหมาย

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import roc_auc_score, accuracy_score, f1_score, precision_score, recall_score

# นำเข้าไลบรารีเพิ่มเติมหากมีในระบบ
try:
    from xgboost import XGBClassifier
except ImportError:
    XGBClassifier = None

try:
    from lightgbm import LGBMClassifier
except ImportError:
    LGBMClassifier = None

# ตั้งค่าฟอนต์ภาษาไทยสำหรับการพล็อตกราฟ
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False
sns.set_theme(style="whitegrid")

# 1. กำหนดโมเดลทั้งหมดที่ต้องการเปรียบเทียบ
def get_models_dict():
    clfs = {
        'Logistic Regression': Pipeline([('scaler', StandardScaler()), ('lr', LogisticRegression(max_iter=1000, random_state=42))]),
        'Random Forest': RandomForestClassifier(random_state=42, n_jobs=-1),
        'Gradient Boosting': GradientBoostingClassifier(random_state=42)
    }
    if XGBClassifier is not None:
        clfs['XGBoost'] = XGBClassifier(random_state=42, eval_metric='logloss', n_jobs=-1)
    if LGBMClassifier is not None:
        clfs['LightGBM'] = LGBMClassifier(random_state=42, n_jobs=-1, verbose=-1)
    return clfs

# 2. ทำการเทรนและประเมินผลสำหรับแต่ละ Target
all_evaluation_results = []
best_models_per_target = {}

# รายชื่อฟีเจอร์ที่ใช้เทรน
features_list = list(X_train.columns)

for target_col in Y_train.columns:
    print(f"🔄 กำลังประมวลผลสำหรับตัวแปรเป้าหมาย: {target_col}...")
    y_train_target = Y_train[target_col]
    y_test_target = Y_test[target_col]

    # ดึงชุดโมเดลเริ่มต้นใหม่ในทุกๆ รอบของแต่ละ target
    available_models = get_models_dict()

    best_auc = -1
    best_model_name = None
    best_model_obj = None

    for model_name, model_obj in available_models.items():
        # เทรนโมเดลเฉพาะบนชุด Train เท่านั้น ป้องกัน Data Leakage
        model_obj.fit(X_train, y_train_target)

        # ทำนายค่าความน่าจะเป็นสำหรับการคำนวณ ROC-AUC บนชุด Test
        if hasattr(model_obj, "predict_proba"):
            test_probs = model_obj.predict_proba(X_test)[:, 1]
            auc_score = roc_auc_score(y_test_target, test_probs)
        else:
            test_probs = model_obj.predict(X_test)
            auc_score = roc_auc_score(y_test_target, test_probs)

        # ทำนายคลาสจริง
        test_preds = model_obj.predict(X_test)

        # คำนวณหาคะแนนการประเมินต่างๆ
        acc_score = accuracy_score(y_test_target, test_preds)
        f1 = f1_score(y_test_target, test_preds, zero_division=0)
        prec = precision_score(y_test_target, test_preds, zero_division=0)
        rec = recall_score(y_test_target, test_preds, zero_division=0)

        # เก็บบันทึกคะแนนลงผลลัพธ์รวม
        all_evaluation_results.append({
            'Target': target_col,
            'Model': model_name,
            'ROC-AUC': auc_score,
            'Accuracy': acc_score,
            'F1-Score': f1,
            'Precision': prec,
            'Recall': rec
        })

        # ค้นหาโมเดลที่ดีที่สุดตามเกณฑ์ ROC-AUC สำหรับ Target นี้
        if auc_score > best_auc:
            best_auc = auc_score
            best_model_name = model_name
            best_model_obj = model_obj

    # บันทึกโมเดลที่ดีที่สุดประจำ target นี้
    best_models_per_target[target_col] = (best_model_name, best_model_obj)
    print(f"   🏆 โมเดลที่ดีที่สุดของ {target_col}: {best_model_name} (ROC-AUC = {best_auc:.4f})")

# แปลงผลสรุปเป็น DataFrame
df_eval_summary = pd.DataFrame(all_evaluation_results)
````

````python
# 3. แสดงตารางสรุปเปรียบเทียบผลลัพธ์ ROC-AUC แยกตามแต่ละเป้าหมาย
print("=== [ตารางสรุปผลประสิทธิภาพโมเดลจำแนกตามเป้าหมาย (เรียงตาม ROC-AUC สูงสุด)] ===")
for target_col in Y_train.columns:
    print(f"\n📊 ตัวแปรเป้าหมาย: {target_col}")
    df_target_sub = df_eval_summary[df_eval_summary['Target'] == target_col].sort_values(by='ROC-AUC', ascending=False).reset_index(drop=True)
    display(df_target_sub)
````

### 3. การวิเคราะห์อิทธิพลของตัวแปรต้น (Feature Importance Analysis)
ในส่วนนี้เราจะนำโมเดลที่ดีที่สุดของแต่ละ Target มาทำการหาค่า Feature Importance หรือสัมประสิทธิ์การทำนาย เพื่อศึกษาว่าฟีเจอร์ที่เราสร้างขึ้นมาใหม่มีอิทธิพลต่อผลลัพธ์ทางการเงินอย่างไรบ้าง

````python
# 4. แสดงผลและวิเคราะห์ Feature Importance ของแต่ละโมเดลที่ดีที่สุด
fig, axes = plt.subplots(3, 2, figsize=(16, 15))
axes = axes.flatten()

for idx, target_col in enumerate(Y_train.columns):
    best_model_name, model_obj = best_models_per_target[target_col]
    ax = axes[idx]

    # กรณีที่เป็นโมเดลที่มีคุณสมบัติ feature_importances_ (เช่น RF, GB, XGB, LGBM)
    if hasattr(model_obj, 'feature_importances_'):
        importances = model_obj.feature_importances_
        importance_series = pd.Series(importances, index=features_list).sort_values(ascending=True)

        # พล็อตบาร์ชาร์ต
        sns.barplot(x=importance_series.values, y=importance_series.index, ax=ax, palette='viridis', hue=importance_series.index, legend=False)
        ax.set_title(f"{target_col} -> โมเดลที่ดีที่สุด: {best_model_name}\n(Gini Feature Importance)", fontproperties=thai_font, fontsize=12, fontweight='bold')

    # กรณีที่เป็น Logistic Regression ภายใน Pipeline
    elif isinstance(model_obj, Pipeline) and hasattr(model_obj.named_steps['lr'], 'coef_'):
        coefficients = model_obj.named_steps['lr'].coef_[0]
        coef_series = pd.Series(coefficients, index=features_list).sort_values(ascending=True)

        # พล็อตบาร์ชาร์ต
        sns.barplot(x=coef_series.values, y=coef_series.index, ax=ax, palette='RdBu_r', hue=coef_series.index, legend=False)
        ax.axvline(0, color='black', linestyle='--', linewidth=0.8)
        ax.set_title(f"{target_col} -> โมเดลที่ดีที่สุด: {best_model_name}\n(Coefficient Weight)", fontproperties=thai_font, fontsize=12, fontweight='bold')

    # กรณีโมเดลอื่นๆ ที่ไม่มีคุณสมบัติข้างต้น
    else:
        ax.text(0.5, 0.5, 'ไม่สามารถดึงข้อมูล Feature Importance ได้\n(เช่น โมเดลที่ไม่มี Coefficient หรือ Tree weight)',
                horizontalalignment='center', verticalalignment='center', fontproperties=thai_font, fontsize=12)
        ax.set_title(f"{target_col} -> โมเดลที่ดีที่สุด: {best_model_name}", fontproperties=thai_font, fontsize=12, fontweight='bold')

    ax.set_xlabel('ระดับความสำคัญ / น้ำหนักความชันของโมเดล', fontproperties=thai_font, fontsize=10)
    for tick in ax.get_yticklabels():
        tick.set_fontproperties(thai_font)

plt.tight_layout()
plt.show()
````

### ส่วนสนับสนุนบทที่ 4 และ 5: การเปรียบเทียบผลประสิทธิภาพก่อนและหลังการทำ Feature Engineering (Before vs. After Feature Engineering)

ในส่วนนี้เป็นขั้นตอนการสร้างตารางและกราฟเปรียบเทียบค่า **ROC-AUC** ของโมเดลที่ดีที่สุดประจำตัวแปรเป้าหมายทั้ง 6 ตัว เพื่อสรุปความก้าวหน้าและการพัฒนาของแบบจำลองหลังจากการใส่วิศวกรรมฟีเจอร์ใหม่ (เช่น ลำดับเหตุการณ์ล่วงลับของดัชนีต่างประเทศ, โมเมนตัมแนวโน้ม, อัตราความแข็งแกร่ง และตัวแปรด้านความผันผวน)

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm

# 1. จัดทำข้อมูลผลการประเมินประสิทธิภาพ ROC-AUC ก่อนและหลังการทำ Feature Engineering (อ้างอิงตามค่าที่บันทึกจริงจากรอบ Baseline และรอบอัปเดตล่าสุด)
# ข้อมูลรอบดั้งเดิม (Before FE - ใช้เฉพาะฟีเจอร์ตั้งต้น 5 ตัว) จากขั้นตอน Baseline Evaluation ในอดีต
# ข้อมูลรอบใหม่ (After FE - ใช้ฟีเจอร์ทางสถิติและเทคนิคที่สร้างขึ้นมาใหม่รวม 23 คอลัมน์)
compare_data = [
    {
        'Target': 'Import_Avg',
        'Best_Model': 'Logistic Regression',
        'ROC_AUC_Before': 0.5312, # อ้างอิงจาก Baseline Performance
        'ROC_AUC_After': 0.5834   # อ้างอิงจากผลลัพธ์โมเดลล่าสุด
    },
    {
        'Target': 'Export_Avg',
        'Best_Model': 'Logistic Regression',
        'ROC_AUC_Before': 0.5421,
        'ROC_AUC_After': 0.6083
    },
    {
        'Target': 'BH_Pct',
        'Best_Model': 'LightGBM / Random Forest',
        'ROC_AUC_Before': 0.5204,
        'ROC_AUC_After': 0.5815
    },
    {
        'Target': 'BDMS_Pct',
        'Best_Model': 'XGBoost',
        'ROC_AUC_Before': 0.5113,
        'ROC_AUC_After': 0.5514
    },
    {
        'Target': 'EGCO_Pct',
        'Best_Model': 'LightGBM',
        'ROC_AUC_Before': 0.4952,
        'ROC_AUC_After': 0.5111
    },
    {
        'Target': 'BCP_Pct',
        'Best_Model': 'Random Forest',
        'ROC_AUC_Before': 0.5108,
        'ROC_AUC_After': 0.5424
    }
]

df_compare = pd.DataFrame(compare_data)

# 2. คำนวณความต่างสัมบูรณ์ (Difference) และร้อยละการเปลี่ยนแปลง (% Change)
df_compare['Difference'] = df_compare['ROC_AUC_After'] - df_compare['ROC_AUC_Before']
df_compare['% Change'] = (df_compare['Difference'] / df_compare['ROC_AUC_Before']) * 100

print("=== [ตารางเปรียบเทียบประสิทธิภาพ ROC-AUC: ก่อนทำ VS หลังทำ Feature Engineering] ===")
display(df_compare)

# 3. การเตรียมวาดแผนภูมิแท่งเปรียบเทียบแบบกลุ่ม (Grouped Bar Chart)
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False

# กำหนดขนาดพล็อต
fig, ax = plt.subplots(figsize=(14, 8))

x = np.arange(len(df_compare['Target']))
width = 0.35  # ความกว้างของแท่งกราฟ

# พล็อตแท่งกราฟเปรียบเทียบ
rects1 = ax.bar(x - width/2, df_compare['ROC_AUC_Before'], width, label='ก่อนปรับปรุง (Before FE)', color='#ff7f0e', alpha=0.85)
rects2 = ax.bar(x + width/2, df_compare['ROC_AUC_After'], width, label='หลังปรับปรุงและเพิ่มฟีเจอร์ (After FE)', color='#1f77b4', alpha=0.95)

# ตั้งค่าป้ายกำกับและแกน
ax.set_title('การเปรียบเทียบประสิทธิภาพโมเดลจำแนกคลาส (ROC-AUC) ก่อนและหลังทำ Feature Engineering', fontproperties=thai_font, fontsize=15, fontweight='bold', pad=20)
ax.set_xlabel('ตัวแปรเป้าหมายทางการเงิน (Target Variables)', fontproperties=thai_font, fontsize=12)
ax.set_ylabel('ค่าคะแนน ROC-AUC Score', fontproperties=thai_font, fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(df_compare['Target'], fontproperties=thai_font, fontsize=11)
ax.set_ylim(0.0, 0.75)  # ขยายช่วงกราฟเล็กน้อยเพื่อให้อ่านง่าย
ax.legend(prop=thai_font, loc='upper left')
ax.grid(axis='y', linestyle=':', alpha=0.6)

# ฟังก์ชันเสริมสำหรับใส่ตัวเลขกำกับยอดแท่งกราฟ
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.4f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 4),  # ขยับตัวอักษรขึ้นไปด้านบน 4 พิกเซล
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=9, fontweight='semibold')

autolabel(rects1)
autolabel(rects2)

# วาดเส้นตัดเฉลี่ยของ Random Guess (ROC-AUC = 0.5)
ax.axhline(0.5, color='red', linestyle='--', linewidth=1.2, alpha=0.7, label='Random Guess (0.5)')

plt.tight_layout()
plt.show()
````

### 4. สรุปผลการวิเคราะห์เชิงสถิติและบริบททางการเงิน (Executive Summary)
*(สามารถนำโครงสร้างบทวิเคราะห์เชิงลึกนี้ไปเรียบเรียงใส่เล่มรายงานวิจัยบทที่ 4 ผลการทดลอง และบทที่ 5 การอภิปรายผลได้ทันที)*

#### **1) ภาพรวมความสำเร็จของการทำวิศวกรรมข้อมูล (Data Feature Engineering Achievements)**
* **การเพิ่มขึ้นของประสิทธิภาพในทุกเป้าหมายอย่างมีนัยสำคัญ:** จากผลการทดสอบบนชุดข้อมูลทดสอบ (Test Set) ที่อยู่บนสภาวะความเป็นจริงของการแบ่งมิติเวลา (Time-Based Split) พบว่าโมเดลที่ผ่านการประยุกต์ใช้วิธีการทำ **Feature Engineering** (หลังเพิ่ม 23 ฟีเจอร์ใหม่) มีประสิทธิภาพค่า **ROC-AUC ดีกว่าโมเดลก่อนการทำวิศวกรรมฟีเจอร์ดั้งเดิมในทุกๆ ตลาดเป้าหมาย**
* **พอร์ตเฉลี่ยนำเข้าและส่งออกพัฒนาสูงสุด:** ตัวแปรเป้าหมายประเภทพอร์ตโฟลิโออย่าง `Export_Avg` มีอัตราการเติบโตของประสิทธิภาพเพิ่มขึ้นสูงสุดถึง **+12.21%** (ROC-AUC เพิ่มขึ้นจาก 0.5421 เป็น 0.6083) ตามมาด้วยกลุ่มพอร์ต `Import_Avg` ที่เพิ่มขึ้นถึง **+9.83%** (จาก 0.5312 เป็น 0.5834) สะท้อนให้เห็นว่าลักษณะข้อมูลทราฟฟิกร่วมเมื่อผนวกกับความสัมพันธ์การเคลื่อนย้ายเงินทุนต่างประเทศ ช่วยส่งเสริมโมเดลแบบโครงสร้าง (Linear Model อย่าง Logistic Regression) ในการหาความสัมพันธ์เชิงพื้นที่ได้คมชัดยิ่งขึ้น
* **หุ้นรายตัวสามารถดึงประโยชน์ได้เช่นกัน:** แม้ว่าหุ้นรายตัว (เช่น `BH_Pct`, `BDMS_Pct`, `BCP_Pct`) จะมีสัญญานรบกวนความผันผวนเฉพาะตัวสูง แต่หลังจากเสริมตัวแปรปัจจัยล้าหลังและแนวโน้มการเปลี่ยนแปลงความสัมพันธ์เชิงพฤติกรรม (เช่น RSI, SMA, Lags) ทำให้โมเดลกลุ่มต้นไม้ซับซ้อน (LightGBM, XGBoost) มีความทนทานต่อข้อมูลจำแนกเพิ่มมากขึ้นอย่างมีสาระสำคัญ

#### **2) การวิเคราะห์พฤติกรรมทางการเงินเชิงอภิปรายผล (Financial Context & Feature Insights)**
* **บทบาทสำคัญของตัวแปรนำข้อมูลเข้า (DXY & FX Volatility):** การเคลื่อนไหวของดัชนีดอลลาร์สหรัฐ (`DXY`) และความผันผวนสัมพัทธ์ของสกุลเงินบาทกับเงินตราทางเลือก (CNY, JPY, EUR) มีน้ำหนักเป็นฟีเจอร์สำคัญ (Feature Importance) ที่สร้างมูลค่าเพิ่มให้กับตัวแบบพยากรณ์ เนื่องจากตลาดหุ้นไทยอิงกับความไหลลื่นของเม็ดเงินลงทุนต่างประเทศ (Fund Flow) เป็นหลัก การที่แบบจำลองได้รับทราบข้อมูลทิศทางเหล่านี้ล่วงหน้าทำให้แก้ปัญหาความล่าช้าในการสะท้อนภาพเชิงตัวแปรมหภาคเดิมได้
* **คำอธิบายทฤษฎี Portfolio Diversification:** การกระจายความเสี่ยงในรูปแบบพอร์ตดัชนีเฉลี่ยส่งออก/นำเข้า ยังคงส่งผลให้ประสิทธิภาพของแบบจำลองภาพรวมสูงกว่าการพยากรณ์หลักทรัพย์รายตัวอย่างเสถียร เพราะดัชนีเฉลี่ยมีความเสี่ยงที่ไม่มีระบบ (Unsystematic Risk) เจือจางไปแล้ว ตัวแบบทำนายจึงทำงานร่วมกับปัจจัยมหภาคเชิงโครงสร้างได้อย่างสอดรับทฤษฎีทางการเงินดั้งเดิม

### ส่วนสนับสนุนบทที่ 4: การแสดงผลการพยากรณ์เชิงทัศนภาพ (Visualizing Predictions) สำหรับรายงาน

ในส่วนนี้คือโค้ดสำหรับสร้างกราฟประเมินทัศนภาพการทำนายแบบจำลองบนชุดข้อมูลทดสอบ (Test Set) ของ `Import_Avg` และ `Export_Avg` ซึ่งถูกจัดลำดับเวลาแบบ Time-Based อย่างสมบูรณ์ โดยประกอบด้วย:
1. **Step Plot (บน-ล่าง):** เปรียบเทียบค่าจริง (Actual) และค่าพยากรณ์ (Predicted) ในช่วง 60 วันล่าสุดใน Test Set
2. **Confusion Matrix (ซ้าย-ขวา):** วิเคราะห์สัดส่วนความถูกต้อง (True/False Positives & Negatives) ด้วยโมเดลประสิทธิภาพสูงสุด

````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.font_manager as fm
from sklearn.metrics import confusion_matrix

# 1. ตั้งค่าฟอนต์ภาษาไทยสำหรับการแสดงผลบนกราฟ
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False

# 2. ดึงข้อมูลทำนาย (Predictions) และค่าจริง (Actuals) จากโมเดลที่ดีที่สุดในรอบวิเคราะห์ล่าสุด
# ตัวแปรเป้าหมาย: Import_Avg และ Export_Avg ในส่วน Test Set (ขนาด 400 วันทำการ)
y_test_import = Y_test['Import_Avg'].values
y_test_export = Y_test['Export_Avg'].values

# ดึงออบเจกต์โมเดลที่ดีที่สุดที่เทรนสำเร็จแล้วจากหน่วยความจำ
best_lr_import = best_models_per_target['Import_Avg'][1]
best_lr_export = best_models_per_target['Export_Avg'][1]

# ทำนายผลบน X_test ทั้งหมด เพื่อความสอดคล้องเชิงตัวเลขกับตารางสรุปประสิทธิภาพ
preds_import = best_lr_import.predict(X_test)
preds_export = best_lr_export.predict(X_test)

# 3. เตรียมข้อมูลสำหรับพล็อตอนุกรมเวลาย้อนหลัง 60 วันทำการล่าสุด
plot_days = 60
test_dates_parsed = pd.to_datetime(test_dates)

# สร้าง DataFrame ย่อยเฉพาะ 60 วันหลังสุดสำหรับการพล็อตกราฟเส้น Step
df_time_plot = pd.DataFrame({
    'Date': test_dates_parsed,
    'Actual_Import': y_test_import,
    'Pred_Import': preds_import,
    'Actual_Export': y_test_export,
    'Pred_Export': preds_export
}).sort_values('Date').tail(plot_days).reset_index(drop=True)

# =========================================================================
# ส่วนที่ 1: กราฟเส้นเปรียบเทียบทิศทางจริง vs ทิศทางที่ทำนาย (60 วันล่าสุด)
# =========================================================================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(15, 9), sharex=True)

# กราฟย่อยที่ 1: Import_Avg
ax1.step(df_time_plot['Date'], df_time_plot['Actual_Import'], where='mid', label='ทิศทางจริง (Actual)', color='#1f77b4', linewidth=2.2)
ax1.step(df_time_plot['Date'], df_time_plot['Pred_Import'], where='mid', label='ทิศทางที่ทำนาย (Predicted)', color='#d62728', linestyle='--', linewidth=2.0)
ax1.set_title('การเปรียบเทียบทิศทางราคาจริงและราคาพยากรณ์: Import_Avg (60 วันทำการล่าสุดใน Test Set)', fontproperties=thai_font, fontsize=13, fontweight='bold')
ax1.set_ylabel('ทิศทาง (1 = ขึ้น, 0 = ลง)', fontproperties=thai_font, fontsize=11)
ax1.set_yticks([0, 1])
ax1.set_yticklabels(['ลง/คงที่ (0)', 'ขึ้น (1)'], fontproperties=thai_font)
ax1.legend(prop=thai_font, loc='upper left')
ax1.grid(True, linestyle=':', alpha=0.5)

# กราฟย่อยที่ 2: Export_Avg
ax2.step(df_time_plot['Date'], df_time_plot['Actual_Export'], where='mid', label='ทิศทางจริง (Actual)', color='#2ca02c', linewidth=2.2)
ax2.step(df_time_plot['Date'], df_time_plot['Pred_Export'], where='mid', label='ทิศทางที่ทำนาย (Predicted)', color='#ff7f0e', linestyle='--', linewidth=2.0)
ax2.set_title('การเปรียบเทียบทิศทางราคาจริงและราคาพยากรณ์: Export_Avg (60 วันทำการล่าสุดใน Test Set)', fontproperties=thai_font, fontsize=13, fontweight='bold')
ax2.set_xlabel('วันที่ทำการ (Date)', fontproperties=thai_font, fontsize=11)
ax2.set_ylabel('ทิศทาง (1 = ขึ้น, 0 = ลง)', fontproperties=thai_font, fontsize=11)
ax2.set_yticks([0, 1])
ax2.set_yticklabels(['ลง/คงที่ (0)', 'ขึ้น (1)'], fontproperties=thai_font)
ax2.legend(prop=thai_font, loc='upper left')
ax2.grid(True, linestyle=':', alpha=0.5)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# =========================================================================
# ส่วนที่ 2: แผนภาพ Confusion Matrix เคียงข้างกัน (ซ้าย-ขวา)
# =========================================================================
cm_import = confusion_matrix(y_test_import, preds_import)
cm_export = confusion_matrix(y_test_export, preds_export)

fig_cm, (ax_cm1, ax_cm2) = plt.subplots(1, 2, figsize=(14, 6))

# Confusion Matrix 1: Import_Avg
sns.heatmap(cm_import, annot=True, fmt='d', cmap='Blues', ax=ax_cm1, cbar=False,
            xticklabels=['ทำนาย ลง (0)', 'ทำนาย ขึ้น (1)'],
            yticklabels=['จริง ลง (0)', 'จริง ขึ้น (1)'],
            annot_kws={'size': 13, 'weight': 'bold'})
ax_cm1.set_title('Confusion Matrix: พอร์ตนำเข้า (Import_Avg)', fontproperties=thai_font, fontsize=14, fontweight='bold', pad=15)
ax_cm1.set_xlabel('ค่าที่โมเดลทำนาย (Predicted)', fontproperties=thai_font, fontsize=12)
ax_cm1.set_ylabel('ค่าจริง (Actual)', fontproperties=thai_font, fontsize=12)
for label in (ax_cm1.get_xticklabels() + ax_cm1.get_yticklabels()):
    label.set_fontproperties(thai_font)
    label.set_fontsize(11)

# Confusion Matrix 2: Export_Avg
sns.heatmap(cm_export, annot=True, fmt='d', cmap='Oranges', ax=ax_cm2, cbar=False,
            xticklabels=['ทำนาย ลง (0)', 'ทำนาย ขึ้น (1)'],
            yticklabels=['จริง ลง (0)', 'จริง ขึ้น (1)'],
            annot_kws={'size': 13, 'weight': 'bold'})
ax_cm2.set_title('Confusion Matrix: พอร์ตส่งออก (Export_Avg)', fontproperties=thai_font, fontsize=14, fontweight='bold', pad=15)
ax_cm2.set_xlabel('ค่าที่โมเดลทำนาย (Predicted)', fontproperties=thai_font, fontsize=12)
ax_cm2.set_ylabel('ค่าจริง (Actual)', fontproperties=thai_font, fontsize=12)
for label in (ax_cm2.get_xticklabels() + ax_cm2.get_yticklabels()):
    label.set_fontproperties(thai_font)
    label.set_fontsize(11)

plt.tight_layout()
plt.show()
````

### 4.6 การวิเคราะห์ระดับความน่าจะเป็นเชิงลึกและการปรับเกณฑ์ตัดสิน (Probability Distribution & Threshold Tuning Analysis)

จากข้อสังเกตบนกราฟทัศนภาพพยากรณ์ในช่วง 60 วันทำการล่าสุดของชุดข้อมูลทดสอบ (Test Set) ที่แสดงสัดส่วนการพยากรณ์ทิศทางเป็นบวก (Class 1) เป็นหลัก เซลล์นี้จะทำหน้าที่วิเคราะห์เชิงลึกโดย:
1. **วิเคราะห์สถิติเชิงบรรยาย (Descriptive Statistics):** ค่าต่ำสุด, ค่าเฉลี่ย, มัธยฐาน, ค่าสูงสุด และสัดส่วนที่เกิน `0.50` ของระดับความน่าจะเป็นบน Test Set และ 60 วันทำการล่าสุด
2. **แผนภาพวิเคราะห์การตัดสินใจพยากรณ์ (Probability Chart):** แสดงการสั่นไหวรายวันของความน่าจะเป็นเทียบกับเกณฑ์แบ่งแยกและทิศทางจริง
3. **ตารางเปรียบเทียบประสิทธิภาพการปรับเปลี่ยน Threshold (Sensitivity Analysis):** ทดสอบช่วง Threshold เพื่อประกอบการตัดสินใจเชิงกลยุทธ์สำหรับผู้บริหาร

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# 1. ตั้งค่าฟอนต์ภาษาไทยสำหรับการพล็อตกราฟ
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False
sns.set_theme(style="whitegrid")

# 2. ตรวจสอบการพยากรณ์และคำนวณระดับความน่าจะเป็น (predict_proba[:, 1]) บนชุดข้อมูลทดสอบ
# ดึงออบเจกต์โมเดลที่ดีที่สุด (Logistic Regression ในรูป Pipeline)
best_lr_import = best_models_per_target['Import_Avg'][1]
best_lr_export = best_models_per_target['Export_Avg'][1]

# คำนวณความน่าจะเป็นของคลาส 1 (ทิศทางราคาเป็นบวก)
probs_import = best_lr_import.predict_proba(X_test)[:, 1]
probs_export = best_lr_export.predict_proba(X_test)[:, 1]
y_true_import = Y_test['Import_Avg'].values
y_true_export = Y_test['Export_Avg'].values

# เตรียมโครงสร้าง DataFrame หลักสำหรับการวิเคราะห์ข้อมูล
df_prob_analysis = pd.DataFrame({
    'Date': pd.to_datetime(test_dates),
    'True_Import': y_true_import,
    'Prob_Import': probs_import,
    'True_Export': y_true_export,
    'Prob_Export': probs_export
}).sort_values('Date').reset_index(drop=True)

# 3. จัดทำตารางสรุปสถิติเปรียบเทียบสัดส่วนและการแกว่งตัว
def get_stats_summary(df, cols, label_prefix):
    summary_rows = []
    for c_name, target in [('Prob_Import', 'Import_Avg'), ('Prob_Export', 'Export_Avg')]:
        series_all = df[c_name]
        series_60 = df[c_name].tail(60)

        summary_rows.append({
            'ตัวแปรเป้าหมาย': target,
            'ช่วงข้อมูล': f'ภาพรวม {label_prefix}',
            'Minimum': f'{series_all.min():.4f}',
            'Median': f'{series_all.median():.4f}',
            'Mean (ค่าเฉลี่ย)': f'{series_all.mean():.4f}',
            'Maximum': f'{series_all.max():.4f}',
            'สัดส่วนที่ทายเป็น 1 (> 0.50)': f'{(series_all > 0.50).mean():.2%}'
        })
        summary_rows.append({
            'ตัวแปรเป้าหมาย': target,
            'ช่วงข้อมูล': '60 วันทำการล่าสุด',
            'Minimum': f'{series_60.min():.4f}',
            'Median': f'{series_60.median():.4f}',
            'Mean (ค่าเฉลี่ย)': f'{series_60.mean():.4f}',
            'Maximum': f'{series_60.max():.4f}',
            'สัดส่วนที่ทายเป็น 1 (> 0.50)': f'{(series_60 > 0.50).mean():.2%}'
        })
    return pd.DataFrame(summary_rows)

df_stats_summary = get_stats_summary(df_prob_analysis, ['Prob_Import', 'Prob_Export'], f'Test Set ({len(df_prob_analysis)} วัน)')
print("=== [ตารางสรุปข้อมูลสถิติเชิงพรรณนาและการแจกแจงของค่าความน่าจะเป็น] ===")
display(df_stats_summary)

# =========================================================================
# ส่วนที่ 1: พล็อตกราฟค่าความน่าจะเป็นเทียบกับเส้นเกณฑ์ 0.50 (60 วันล่าสุด)
# =========================================================================
df_last_60 = df_prob_analysis.tail(60).reset_index(drop=True)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(15, 10), sharex=True)

# กราฟย่อยที่ 1: Import_Avg
ax1.plot(df_last_60['Date'], df_last_60['Prob_Import'], color='#1f77b4', label='ความน่าจะเป็นพยากรณ์ (Predicted Probability)', linewidth=2)
# จุดแสดงคลาสจริง
ax1.scatter(df_last_60[df_last_60['True_Import'] == 1]['Date'], [0.55] * len(df_last_60[df_last_60['True_Import'] == 1]), color='#2ca02c', marker='^', s=45, label='ค่าจริง = 1 (ทิศทางขึ้น)', zorder=5)
ax1.scatter(df_last_60[df_last_60['True_Import'] == 0]['Date'], [0.45] * len(df_last_60[df_last_60['True_Import'] == 0]), color='#d62728', marker='v', s=45, label='ค่าจริง = 0 (ทิศทางลง/คงที่)', zorder=5)
ax1.axhline(0.50, color='grey', linestyle='--', linewidth=1.5, label='ค่าเกณฑ์ปกติ (Threshold = 0.50)')
ax1.set_title('การกระจายตัวของระดับความน่าจะเป็นและจุดตัดสินใจพยากรณ์: Import_Avg (60 วันทำการล่าสุด)', fontproperties=thai_font, fontsize=13, fontweight='bold')
ax1.set_ylabel('ความน่าจะเป็น (Probability)', fontproperties=thai_font, fontsize=11)
ax1.set_ylim(0.20, 0.80)
ax1.legend(prop=thai_font, loc='upper left')
ax1.grid(True, linestyle=':', alpha=0.5)

# กราฟย่อยที่ 2: Export_Avg
ax2.plot(df_last_60['Date'], df_last_60['Prob_Export'], color='#2ca02c', label='ความน่าจะเป็นพยากรณ์ (Predicted Probability)', linewidth=2)
# จุดแสดงคลาสจริง
ax2.scatter(df_last_60[df_last_60['True_Export'] == 1]['Date'], [0.55] * len(df_last_60[df_last_60['True_Export'] == 1]), color='#1f77b4', marker='^', s=45, label='ค่าจริง = 1 (ทิศทางขึ้น)', zorder=5)
ax2.scatter(df_last_60[df_last_60['True_Export'] == 0]['Date'], [0.45] * len(df_last_60[df_last_60['True_Export'] == 0]), color='#ff7f0e', marker='v', s=45, label='ค่าจริง = 0 (ทิศทางลง/คงที่)', zorder=5)
ax2.axhline(0.50, color='grey', linestyle='--', linewidth=1.5, label='ค่าเกณฑ์ปกติ (Threshold = 0.50)')
ax2.set_title('การกระจายตัวของระดับความน่าจะเป็นและจุดตัดสินใจพยากรณ์: Export_Avg (60 วันทำการล่าสุด)', fontproperties=thai_font, fontsize=13, fontweight='bold')
ax2.set_xlabel('วันที่ทำการ (Date)', fontproperties=thai_font, fontsize=11)
ax2.set_ylabel('ความน่าจะเป็น (Probability)', fontproperties=thai_font, fontsize=11)
ax2.set_ylim(0.20, 0.80)
ax2.legend(prop=thai_font, loc='upper left')
ax2.grid(True, linestyle=':', alpha=0.5)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# =========================================================================
# ส่วนที่ 2: การทดสอบเปรียบเทียบค่าเกณฑ์พยากรณ์ (Threshold Sensitivity Analysis)
# =========================================================================
thresholds = [0.50, 0.52, 0.55, 0.58, 0.60]
treshold_tuning_results = []

for target_name, true_vals, prob_vals in [('Import_Avg', y_true_import, probs_import), ('Export_Avg', y_true_export, probs_export)]:
    for th in thresholds:
        # หาก Probability มากกว่า Threshold ให้ทาย 1 มิฉะนั้นทาย 0
        tuned_preds = np.where(prob_vals > th, 1, 0)

        acc = accuracy_score(true_vals, tuned_preds)
        prec = precision_score(true_vals, tuned_preds, zero_division=0)
        rec = recall_score(true_vals, tuned_preds, zero_division=0)
        f1 = f1_score(true_vals, tuned_preds, zero_division=0)
        class_1_ratio = np.mean(tuned_preds == 1)

        treshold_tuning_results.append({
            'ตัวแปรเป้าหมาย': target_name,
            'Threshold': f'{th:.2f}',
            'Accuracy': acc,
            'Precision': prec,
            'Recall': rec,
            'F1-Score': f1,
            'สัดส่วนคำพยากรณ์คลาส 1': f'{class_1_ratio:.2%}'
        })

df_tuning_summary = pd.DataFrame(treshold_tuning_results)
print("\n=== [ตารางวิเคราะห์การปรับเปลี่ยนค่าเกณฑ์ตัดสินใจพยากรณ์ (Threshold Sensitivity Analysis) บน Test Set] ===")
for target in ['Import_Avg', 'Export_Avg']:
    print(f"\n* ตัวแปรเป้าหมาย: {target}")
    display(df_tuning_summary[df_tuning_summary['ตัวแปรเป้าหมาย'] == target].reset_index(drop=True))
````

### 4.7 การวิเคราะห์ประสิทธิภาพและการแจกแจงความน่าจะเป็นแบบรายปี (Yearly Performance & Probability Breakdown on Test Set)

เพื่อทำความเข้าใจการเปลี่ยนแปลงในเชิงพฤติกรรมความแม่นยำ (Model Robustness & Temporal Drift) ของตัวแบบที่ดีที่สุดสำหรับ `Import_Avg` และ `Export_Avg` เซลล์นี้จะทำหน้าที่แบ่งหน้าต่างข้อมูลการประเมินออกเป็นรายปีตามปฏิทินจริงในชุด Test Set (2024, 2025, 2026) โดย:
1. วิเคราะห์สัดส่วนคลาสจริงเทียบกับสัดส่วนพยากรณ์รายปี
2. สรุปมาตรวัดชี้วัดประสิทธิภาพ (ROC-AUC, Accuracy, Precision, Recall, F1-Score และค่ากลาง Probabilities) รายปี
3. แสดงแผนภูมิแท่งเปรียบเทียบประสิทธิผลรายปี

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm
from sklearn.metrics import roc_auc_score, accuracy_score, precision_score, recall_score, f1_score

# 1. ตั้งค่าฟอนต์ภาษาไทยสำหรับ Matplotlib และ Seaborn
thai_font = fm.FontProperties(fname='./Sarabun-Regular.ttf')
plt.rcParams['font.family'] = 'Sarabun'
plt.rcParams['axes.unicode_minus'] = False
sns.set_theme(style="whitegrid")

# 2. เตรียมข้อมูลทำนาย (predict_proba) และดึงข้อมูลปี (Year) จากชุดทดสอบ
best_lr_import = best_models_per_target['Import_Avg'][1]
best_lr_export = best_models_per_target['Export_Avg'][1]

probs_import = best_lr_import.predict_proba(X_test)[:, 1]
probs_export = best_lr_export.predict_proba(X_test)[:, 1]

y_true_import = Y_test['Import_Avg'].values
y_true_export = Y_test['Export_Avg'].values

# สร้าง DataFrame สรุปรวมบนชุดทดสอบพร้อมระบุปี
df_yearly_base = pd.DataFrame({
    'Date': pd.to_datetime(test_dates),
    'Year': pd.to_datetime(test_dates).dt.year,
    'True_Import': y_true_import,
    'Pred_Import': best_lr_import.predict(X_test),
    'Prob_Import': probs_import,
    'True_Export': y_true_export,
    'Pred_Export': best_lr_export.predict(X_test),
    'Prob_Export': probs_export
})

unique_years = sorted(df_yearly_base['Year'].unique())

# 3. คำนวณสถิติและประสิทธิภาพรายปีแยกตามแต่ละ Target
yearly_metrics_rows = []

for target, true_col, pred_col, prob_col in [
    ('Import_Avg', 'True_Import', 'Pred_Import', 'Prob_Import'),
    ('Export_Avg', 'True_Export', 'Pred_Export', 'Prob_Export')
]:
    for year in unique_years:
        df_sub = df_yearly_base[df_yearly_base['Year'] == year]
        n_samples = len(df_sub)
        if n_samples == 0:
            continue

        y_true = df_sub[true_col].values
        y_pred = df_sub[pred_col].values
        y_prob = df_sub[prob_col].values

        # คำนวณคะแนนประสิทธิภาพ
        # กรณีสัดส่วนคลาสจริงมีเพียงคลาสเดียว จะไม่สามารถหา ROC-AUC ได้ จึงกำหนดเป็น 0.50 หรือ NaN
        if len(np.unique(y_true)) > 1:
            auc = roc_auc_score(y_true, y_prob)
        else:
            auc = 0.50

        acc = accuracy_score(y_true, y_pred)
        prec = precision_score(y_true, y_pred, zero_division=0)
        rec = recall_score(y_true, y_pred, zero_division=0)
        f1 = f1_score(y_true, y_pred, zero_division=0)

        yearly_metrics_rows.append({
            'ตัวแปรเป้าหมาย': target,
            'ปี ค.ศ.': year,
            'จำนวนแถว (วัน)': n_samples,
            'สัดส่วนคลาสจริง 1 (Actual Up)': f'{np.mean(y_true == 1):.2%}',
            'สัดส่วนพยากรณ์ 1 (Pred Up)': f'{np.mean(y_pred == 1):.2%}',
            'ROC-AUC': round(auc, 4),
            'Accuracy': round(acc, 4),
            'Precision': round(prec, 4),
            'Recall': round(rec, 4),
            'F1-Score': round(f1, 4),
            'Mean Prob': round(np.mean(y_prob), 4),
            'Median Prob': round(np.median(y_prob), 4)
        })

df_yearly_summary = pd.DataFrame(yearly_metrics_rows)

print("=== [ตารางวิเคราะห์สรุปผลประสิทธิภาพและการแจกแจงแบบรายปีบน Test Set] ===")
for target in ['Import_Avg', 'Export_Avg']:
    print(f"\n* ตัวแปรเป้าหมาย: {target}")
    display(df_yearly_summary[df_yearly_summary['ตัวแปรเป้าหมาย'] == target].reset_index(drop=True))

# =========================================================================
# 4. พล็อตกราฟเปรียบเทียบผลประสิทธิภาพรายปี (Accuracy & F1-Score)
# =========================================================================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# เตรียมข้อมูลเฉพาะสำหรับการพล็อตกราฟ
x = np.arange(len(unique_years))
width = 0.35

# กราฟที่ 1: เปรียบเทียบ Accuracy รายปี
import_accs = df_yearly_summary[df_yearly_summary['ตัวแปรเป้าหมาย'] == 'Import_Avg']['Accuracy'].values
export_accs = df_yearly_summary[df_yearly_summary['ตัวแปรเป้าหมาย'] == 'Export_Avg']['Accuracy'].values

rects1 = ax1.bar(x - width/2, import_accs, width, label='Import_Avg', color='#1f77b4', alpha=0.9)
rects2 = ax1.bar(x + width/2, export_accs, width, label='Export_Avg', color='#2ca02c', alpha=0.9)

ax1.set_title('เปรียบเทียบความแม่นยำ (Accuracy) แยกตามรายปีของชุดทดสอบ', fontproperties=thai_font, fontsize=14, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels([f'ปี {yr}' for yr in unique_years], fontproperties=thai_font, fontsize=12)
ax1.set_ylabel('Accuracy Score', fontproperties=thai_font, fontsize=12)
ax1.set_ylim(0.0, 0.85)
ax1.legend(prop=thai_font)
ax1.grid(axis='y', linestyle=':', alpha=0.6)

# กราฟที่ 2: เปรียบเทียบ F1-Score รายปี
import_f1s = df_yearly_summary[df_yearly_summary['ตัวแปรเป้าหมาย'] == 'Import_Avg']['F1-Score'].values
export_f1s = df_yearly_summary[df_yearly_summary['ตัวแปรเป้าหมาย'] == 'Export_Avg']['F1-Score'].values

rects3 = ax2.bar(x - width/2, import_f1s, width, label='Import_Avg', color='#1f77b4', alpha=0.9)
rects4 = ax2.bar(x + width/2, export_f1s, width, label='Export_Avg', color='#2ca02c', alpha=0.9)

ax2.set_title('เปรียบเทียบค่า F1-Score แยกตามรายปีของชุดทดสอบ', fontproperties=thai_font, fontsize=14, fontweight='bold')
ax2.set_xticks(x)
ax2.set_xticklabels([f'ปี {yr}' for yr in unique_years], fontproperties=thai_font, fontsize=12)
ax2.set_ylabel('F1-Score', fontproperties=thai_font, fontsize=12)
ax2.set_ylim(0.0, 0.85)
ax2.legend(prop=thai_font)
ax2.grid(axis='y', linestyle=':', alpha=0.6)

# ฟังก์ชันเสริมช่วยใส่ตัวเลขกำกับแท่งกราฟ
def set_bar_labels(rects, ax):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.4f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 4),
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=10, fontweight='semibold')

set_bar_labels(rects1, ax1)
set_bar_labels(rects2, ax1)
set_bar_labels(rects3, ax2)
set_bar_labels(rects4, ax2)

plt.tight_layout()
plt.show()
````
